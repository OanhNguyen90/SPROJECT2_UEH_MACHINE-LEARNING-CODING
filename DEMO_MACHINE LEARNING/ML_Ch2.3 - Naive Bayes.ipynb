#### 1. **Gaussian NB**
---

```python
## Tập tin dữ liệu IRIS
data = pd.read_excel(folder + '/Data/Iris.xls')
```

```python
## Các features
X = data.drop('species', axis = 1)
X.head()

## Biến target: species
y = data.species

print(pd.concat([X, y], axis = 1).head())

## Chia tập dữ liệu thành training, test sets theo tỷ lệ 80:20
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = .2, random_state = 1)
```

```python
##------------------------------------------------------------------------------
## Huấn luyện mô hình Gaussian NB
##------------------------------------------------------------------------------
Gmodel = GaussianNB()
Gmodel.fit(X_train, y_train)
```

```python
## Khai thác mô hình Gaussian NB đã được huấn luyện để đoán (Test set)
tiep = 'C'
while (tiep.upper() == 'C'):
    # Nhập dữ liệu
    spl    = float(eval(input(f'sepallength [4.3, 7.9]: ')))
    spw    = float(eval(input(f'sepalwidth  [2.0, 4.4]: ')))
    ptl    = float(eval(input(f'petallength [1.0, 6.9]: ')))
    ptw    = float(eval(input(f'petalwidth  [0.1, 2.5]: ')))
    X_new  = pd.DataFrame([spl, spw, ptl, ptw]).T

    # Dự đoán
    y_pred = Gmodel.predict(X_new)
    print(f'\nMẫu {X_new.values[0].tolist()} được dự đoán là {y_pred}\n')

    tiep = input('Tiếp tục (C/K) ? ')
```

```python
## Số trường hợp đoán ĐÚNG trong tập Test
df = pd.DataFrame({'yHat':Gmodel.predict(X_test), 'y': pd.Series(y_test)})
print(f'Accuracy (Test set) = {len(df[df.yHat == df.y])} / {len(X_test)}')
```

## 2. **Multinomial NB**
---

```python
## Tập tin dữ liệu Social_Network_Ads
data = pd.read_csv(folder + '/Data/Social_Network_Ads.csv')
```

```python
## Các features: Gender, Age, EstimatedSalary
X = data.drop(['User ID', 'EstimatedSalary_K', 'Purchased'], axis = 1)

## Chuyển Gender thành kiểu Numerical
sex = {'Male':1, 'Female':0}
X['Male'] = [sex[i] for i in data.Gender]
X         = X.drop(['Gender'], axis = 1)

## Biến target: Purchased
y = data.Purchased

print(pd.concat([X, y], axis = 1).head())

## Chia tập dữ liệu thành training, test sets theo tỷ lệ 80:20
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = .2, random_state = 1)
```

```python
##------------------------------------------------------------------------------
## Huấn luyện mô hình Multinomial NB
##------------------------------------------------------------------------------
Mmodel = MultinomialNB()
Mmodel.fit(X_train, y_train)
```

```python
## Khai thác mô hình Multinomial NB đã được huấn luyện để đoán (Test set)
labels = np.array(['KHÔNG mua', 'MUA'])

tiep = 'C'
while (tiep.upper() == 'C'):
    # Nhập dữ liệu
    age    = float(eval(input(f'Tuổi:             ')))
    sal    = float(eval(input(f'Mức lương:        ')))
    sex    = float(eval(input(f'Nam (1) / Nữ (0): ')))
    X_new  = pd.DataFrame([age, sal, sex]).T

    # Dự đoán
    y_pred = Mmodel.predict(X_new)
    print('Mẫu', X_new.values[0].tolist(), 'được dự đoán là', labels[y_pred])
    tiep = input('Tiếp tục (C/K) ? ')
```
