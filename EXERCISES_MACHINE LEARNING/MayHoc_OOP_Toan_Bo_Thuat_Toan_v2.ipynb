### 0B. DẤU HIỆU **CHẤT LƯỢNG DỮ LIỆU** → CHỌN KỸ THUẬT   *(thiếu · bất thường · không nhất quán · tích hợp)*
| Nhìn vào dữ liệu thấy… | → Chọn | Hàm | Nguồn |
|---|---|---|---|
| **THIẾU** – ô thiếu ở cột **không liên quan** yêu cầu bài | **Giữ nguyên** (không làm sai lệch, tránh dữ liệu thưa) | — | 📕 B4 §4.1 |
| Ô thiếu ở cột **liên quan TRỰC TIẾP** yêu cầu (vd so sánh nam/nữ mà thiếu giới tính) | **Bỏ cả dòng** | `drop_rows(cols)` | 📕 |
| **Cả cột** thiếu nặng (> ~**20%**) và **không tương quan** target | **Bỏ cột** (có tương quan → cân nhắc giữ) | `drop_cols(0.2)` | 📕 |
| Thiếu **lẻ tẻ**, nghi **MCAR**, cột **số** | Điền **mean** (đối xứng) / **median** (lệch, có outlier) | `fill_stat` | 📕 + 📙 |
| Cột **chữ** thiếu | Điền **mode**, hoặc giá trị đặc biệt `'unknown'` | `fill_stat` / `fill_constant` | 📕 |
| **MAR** (`diagnose()` ra thiếu liên quan cột khác) | Điền theo **NHÓM** có cùng tính chất (trung bình lân cận) | `fill_group(col, by)` | 📕 |
| **Nhiều cột số** thiếu, tương quan nhau; **số cột thiếu < số cột đủ** | **MICE** (hoặc **kNN**: số → hồi quy) | `fill_mice` / `fill_knn` | 📕 |
| **MNAR** (thiếu vì chính giá trị quá cao/thấp, vd khai lương) | Tìm **người tương đồng** (kNN) + **tri thức ngoài**; thêm **cột cờ** `*_isna` | `add_missing_flags` + `fill_knn` | 📕 |
| **Chuỗi thời gian** thiếu | Ngày trước · **nội suy** (nối trước–sau) · **cùng kỳ năm trước** | `fill_sequence` | 📕 (mục outlier chuỗi TG) |
| **BẤT THƯỜNG** – **1 biến**, hình chuông | **3-sigma** (`z > 3`, khử đơn vị) | `zscore_mask(k=3)` | 📕 §4.2 |
| **1 biến**, lệch / không chuẩn | **IQR (boxplot)**: ngoài `[Q1−1.5·IQR, Q3+1.5·IQR]` | `iqr_mask` | 📕 B7 + 📙 |
| **Không biết phân phối**, cần biết sẽ còn bao nhiêu % dữ liệu | **Chebyshev**: giữ ≥ `1 − 1/z²` (cận lệch → lấy `z` **nhỏ hơn**) | `chebyshev_keep(z)` | 📕 B7 |
| **2 biến CÙNG thang đo** | Khoảng cách **Euclid** tới **trọng tâm** (vùng *hình tròn*, bán kính R) | `centroid_mask` | 📕 §4.2 |
| **2 biến KHÁC thang đo / tương quan** | **Mahalanobis** (vùng *bầu dục*). Từ **3 chiều** trở lên kém hiệu quả | `mahalanobis_mask` | 📕 §4.2 |
| Cột có **độ biến thiên rất thấp** (≈0) | Outlier mức **feature** (vô ích) → **bỏ cột** | `drop_constant_cols` / `variance_filter` | 📕 §4.2 |
| Outlier là **lỗi** (tuổi = 200, điểm âm) | **Loại dòng** hoặc đặt **NaN** rồi điền như dữ liệu thiếu | `remove_rows` / `to_nan` | 📙 |
| Outlier là giá trị **thật**, **dương & lệch phải** | **Log** (có số 0 → `log1p`) | `log_transform` | 📕 |
| Cần kéo về **phân phối chuẩn** (dương) | **Box-Cox** ⚠️ *xem ghi chú dưới* | `boxcox_transform` | 📕 + 📙 |
| Có **cả âm lẫn dương** | **Yeo-Johnson** (không giới hạn dấu; phương sai gần dữ liệu thô) | `yeojohnson_transform` | 📕 |
| Muốn **giữ dòng**, chỉ kéo outlier về ngưỡng | **Winsorization** (percentile, hoặc thay bằng max/min **không-outlier**) | `winsorize` | 📕 |
| Muốn **làm trơn cục bộ** | **Binning** → thay bằng mean/median của bin | `bin_smooth` | 📕 |
| Outlier trong **chuỗi thời gian** | Ngày trước · **cùng kỳ năm trước** · **nội suy tuyến tính** | `ts_replace` | 📕 |
| **BẤT NHẤT** – cùng giá trị viết khác (`Nam / nam / NAM`, `Hà Nội / Ha Noi`) | **Chuẩn hoá chuỗi** (strip, hoa/thường, bỏ dấu) + **gộp** giá trị gần giống | `standardize_text`, `suggest_merge`, `map_values` | 📕 §4.3 + 📙 |
| **Sai kiểu** (ngày/số lưu dạng chữ, `'N/A'`) | Ép kiểu, lỗi → `NaT/NaN` rồi xử lý như thiếu | `to_datetime`, `to_numeric` | 📙 |
| **Vi phạm ràng buộc** (tuổi ngoài 0–120, ngày kết thúc < ngày bắt đầu) | Kiểm tra miền / luật nghiệp vụ | `check_range`, `check_rule` | 📙 |
| **Dòng trùng** | Bỏ trùng | `drop_duplicates` | 📙 |
| **TÍCH HỢP** nhiều nguồn | Kiến trúc: *consolidation* (gom 1 kho) · *propagation* (sao chép xuống) · *federation* (giao diện ảo) · *data warehouse* (theo chủ đề). Code: **merge / concat** theo khoá | `DataIntegrator.merge/stack` | 📕 B5 |
| Thông tin **dư thừa** – 2 cột **số** | **Pearson** (tuyến tính) / **Spearman** (thứ hạng) / hiệp phương sai | `redundant_numeric` | 📕 B5 |
| Thông tin dư thừa – 2 cột **chữ** | **Chi-bình-phương** (+ Cramér V) | `redundant_categorical` | 📕 B5 |

### 0C. DẤU HIỆU **ĐỂ BIẾN ĐỔI – MÃ HOÁ – THU GỌN** → CHỌN KỸ THUẬT
| Nhìn vào dữ liệu / yêu cầu thấy… | → Chọn | Hàm | Nguồn |
|---|---|---|---|
| **CHUẨN HOÁ** – feature **khác thang đo**, **gần chuẩn**, **ít outlier** | **z-score** `(x−μ)/σ` (khử đơn vị) | `zscore` | 📕 B5 §4 |

| **Không chuẩn**, **nhiều outlier** | **Robust** `(x−median)/IQR` | `robust` | 📕 |

| Quan hệ **cong** / cần **tương tác** giữa biến | **Features construction**: `x, x²,…`, `x1·x2` | `add_polynomial`, `add_interaction` | 📕 + 📗 tr.1 |
| Cần biến **0/1** từ biến liên tục (vd hồi quy logistic) | **Nhị phân hoá** theo ngưỡng | `binarize` | 📕 |
| Cần **làm tròn** | Làm tròn số nguyên / thập phân | `round_cols` | 📕 |
| Gom tuổi thành nhóm tuổi (**không có nhãn**) | **Cùng độ rộng** (nhạy outlier) · **cùng tần số** (bền hơn) · **theo tri thức lĩnh vực** | `equal_width` / `equal_freq` / `custom_bins` | 📕 |
| Gom nhóm **có nhãn** target | **Supervised binning** (cắt theo nhãn) | `supervised_bins` | 📕 (tên) + 📙 |
| **Chuỗi thời gian nhiễu** | **SMA** (k điểm trước) · **Centered SMA** (7 trước + 7 sau)/15 · **WMA** (gần nặng, xa nhẹ) | `sma` / `centered_sma` / `wma` | 📕 |

| Bảng **cross-tab** ↔ **features-values** | `crosstab/pivot` ↔ `melt` (đổi mã 0,1,2 về nhãn *Nam, Nữ…* – **thường dùng**) | `to_crosstab` / `to_features_values` | 📕 B6 |
| **MÃ HOÁ** – nhãn y dạng chữ | **Label** 0/1 (lớp dương = 1) | `label_target` | 📙 |

| Biến **định danh, ít giá trị** | **One-hot** (k cột) · **Dummy** (k−1 cột: tránh đa cộng tuyến hồi quy tuyến tính) | `onehot(drop_first=…)` | 📕 |
| Định danh, **rất nhiều giá trị** | **Frequency** · **Target/Mean/Leave-one-out** (⚠️ rò rỉ nhãn) · **Hashing** (độ dài cố định) | `frequency` / `target_encode` / `hashing` | 📕 |
| Mô hình **cây** | Label/ordinal là đủ (không cần one-hot) | `label` | 📙 |

| **THU GỌN** – quá nhiều **dòng** | **Lấy mẫu** (phân tầng giữ tỉ lệ nhãn) · **Clustering** (lưu trọng tâm) · **Histogram** · **Aggregation** · **Mô hình tham số** (lưu hàm hồi quy, không lưu tập train) | `sample`, `cluster_reduce`, `histogram_summary`, `parametric_regression` | 📕 |

```python
# ===== 0.1 IMPORT THƯ VIỆN =====
import os, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats                                  # kiểm định + phân phối xác suất (Notes tr.7, 8)
from scipy.spatial.distance import cdist                 # tính ma trận khoảng cách
from scipy.cluster.hierarchy import linkage, dendrogram  # dendrogram (Notes tr.14)

from sklearn.model_selection import (train_test_split, StratifiedKFold, cross_val_score,
                                     GridSearchCV, RandomizedSearchCV)
from sklearn.preprocessing import StandardScaler, MinMaxScaler, PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor, NearestNeighbors
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, export_text
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.svm import SVC
from sklearn.metrics import (confusion_matrix, accuracy_score, precision_score, recall_score,
                             f1_score, roc_curve, auc, silhouette_score, silhouette_samples,
                             calinski_harabasz_score, davies_bouldin_score, rand_score)
from sklearn.metrics.cluster import pair_confusion_matrix
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.decomposition import PCA                    # CHỈ dùng để vẽ hình 2D, không nằm trong notes

warnings.filterwarnings("ignore")
plt.rcParams["figure.figsize"] = (6, 4)
RANDOM_STATE = 42     # cố định để kết quả lặp lại được (rất quan trọng khi chấm bài)

# Đường dẫn file dữ liệu: nếu chạy trên Colab thì upload file rồi sửa đường dẫn ở đây
CSV_PATH = "Breast_Cancer_WI.csv"
if not os.path.exists(CSV_PATH):
    CSV_PATH = "/mnt/user-data/uploads/Breast_Cancer_WI.csv"
print("Đọc dữ liệu từ:", CSV_PATH)
```

```python
# ===== 0.2 LỚP DataManager: EDA + TIỀN XỬ LÝ + CHIA TẬP =====
class DataManager:
    """Quản lý dữ liệu: load -> kiểm tra ràng buộc -> EDA -> tiền xử lý -> chia tập.
    (Notes tr.4: bài kiểm tra bắt buộc có EDA + tiền xử lý + OOP + ràng buộc)"""

    def __init__(self, path, target="Class"):
        self.path = path          # đường dẫn file csv
        self.target = target      # tên cột nhãn (target attribute – Notes tr.2)
        self.df = None            # dữ liệu gốc
        self.X = None             # features (Notes tr.2: "thông tin khác là features")
        self.y = None             # nhãn dạng số 0/1
        self.label_map = {"Benign": 0, "Malignant": 1}   # mã hoá nhãn chữ -> số (Notes tr.3: masking)

    # ---------- RÀNG BUỘC (thêm điểm khi làm OOP) ----------
    def load(self):
        if not os.path.exists(self.path):
            raise FileNotFoundError(f"Không thấy file: {self.path}")
        self.df = pd.read_csv(self.path)
        if self.target not in self.df.columns:
            raise ValueError(f"Thiếu cột nhãn '{self.target}'")
        return self.df

    # ---------- EDA ----------
    def eda(self):
        df = self.df
        print("1) Kích thước (dòng, cột):", df.shape)
        print("2) Kiểu dữ liệu:", df.dtypes.value_counts().to_dict())
        print("3) Số ô thiếu (missing):", int(df.isna().sum().sum()))
        print("4) Số dòng trùng lặp:", int(df.duplicated().sum()))
        print("5) Phân bố nhãn:\n", df[self.target].value_counts())
        # Cột vô ích: phương sai = 0 (Notes tr.12 – outlier loại 3 "feature vô ích")
        num = df.drop(columns=[self.target])
        const_cols = num.columns[num.var() == 0].tolist()
        print("6) Cột phương sai = 0 (vô ích):", const_cols if const_cols else "không có")
        # Outliers theo IQR (Notes tr.12 – nhận diện dữ liệu bất thường)
        q1, q3 = num.quantile(0.25), num.quantile(0.75)
        iqr = q3 - q1
        n_out = ((num < q1 - 1.5 * iqr) | (num > q3 + 1.5 * iqr)).sum().sort_values(ascending=False)
        print("7) 5 cột nhiều outlier nhất (quy tắc IQR):\n", n_out.head(5))
        print("8) Thống kê mô tả 4 cột đầu:\n", num.iloc[:, :4].describe().round(3))
        # Biểu đồ nhanh
        fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
        df[self.target].value_counts().plot(kind="bar", ax=ax[0], title="Phân bố nhãn")
        for lb, g in df.groupby(self.target):
            ax[1].hist(g["radius_mean"], bins=20, alpha=0.6, label=lb)
        ax[1].set_title("radius_mean theo nhãn"); ax[1].legend()
        plt.tight_layout(); plt.show()

    # ---------- KIỂM ĐỊNH THỐNG KÊ (Notes tr.8) ----------
    def statistical_tests(self):
        df = self.df
        # 2 biến số  -> hệ số tương quan Pearson
        r, p = stats.pearsonr(df["radius_mean"], df["area_mean"])
        print(f"Pearson radius_mean–area_mean: r={r:.3f}, p={p:.3g}")
        # biến phân loại (2 nhóm) vs biến số -> T-test
        a = df.loc[df[self.target] == "Malignant", "radius_mean"]
        b = df.loc[df[self.target] == "Benign", "radius_mean"]
        t, p = stats.ttest_ind(a, b, equal_var=False)
        print(f"T-test radius_mean giữa 2 lớp: t={t:.2f}, p={p:.3g}")
        # biến phân loại (>=2 nhóm) vs biến số -> ANOVA
        f, p = stats.f_oneway(a, b)
        print(f"ANOVA radius_mean giữa các lớp: F={f:.2f}, p={p:.3g}")

    # ---------- TIỀN XỬ LÝ ----------
    def preprocess(self):
        df = self.df.drop_duplicates().dropna().copy()          # bỏ dòng trùng + dòng thiếu
        df["label"] = df[self.target].map(self.label_map)       # chữ -> số (Malignant=1: lớp dương "Positive")
        self.X = df.drop(columns=[self.target, "label"])
        self.y = df["label"].astype(int)
        print("Sau tiền xử lý: X", self.X.shape, "| y", self.y.shape)
        return self.X, self.y

    # ---------- CHIA TẬP ----------
    def split(self, test_size=0.2):
        """Hold-out 80-20 (Notes tr.11). stratify = giữ nguyên tỉ lệ nhãn ở 2 tập."""
        return train_test_split(self.X, self.y, test_size=test_size,
                                stratify=self.y, random_state=RANDOM_STATE)

    def split_train_val_test(self, val_size=0.2, test_size=0.2):
        """Train / Validation(development) / Test (Notes tr.9–10):
        train xây ứng viên – validation chọn mô hình/siêu tham số – test đánh giá cuối."""
        X_tmp, X_te, y_tmp, y_te = train_test_split(self.X, self.y, test_size=test_size,
                                                    stratify=self.y, random_state=RANDOM_STATE)
        rel = val_size / (1 - test_size)
        X_tr, X_va, y_tr, y_va = train_test_split(X_tmp, y_tmp, test_size=rel,
                                                  stratify=y_tmp, random_state=RANDOM_STATE)
        return X_tr, X_va, X_te, y_tr, y_va, y_te

# ----- chạy thử -----
dm = DataManager(CSV_PATH)
dm.load()
dm.eda()
dm.statistical_tests()
X, y = dm.preprocess()
X_train, X_test, y_train, y_test = dm.split()
print("Train:", X_train.shape, " Test:", X_test.shape)
```

### 0.3 Hàm tiện ích dùng chung: tìm điểm gãy ELBOW
**Notes tr.16:** *"Chọn mô hình theo phương pháp ELBOW: chọn điểm gãy… nếu có nhiều điểm gãy thì chọn k càng bé càng tốt mà sai số chấp nhận được."* Dùng lại ở K-NN, hồi quy, K-means.

```python
def find_elbow(xs, ys):
    """Điểm gãy = điểm xa nhất so với đường thẳng nối điểm đầu và điểm cuối của đường cong."""
    xs, ys = np.array(xs, float), np.array(ys, float)
    p1, p2 = np.array([xs[0], ys[0]]), np.array([xs[-1], ys[-1]])
    d = []
    for x, yv in zip(xs, ys):
        p = np.array([x, yv])
        d.append(abs(np.cross(p2 - p1, p1 - p)) / np.linalg.norm(p2 - p1))   # khoảng cách điểm-đường thẳng
    return xs[int(np.argmax(d))]
```

```python
# ===== 0.4 LỚP CƠ SỞ `PrepBase` + CÔNG CỤ DÙNG CHUNG (Word LTPTDL – Buổi 4: 3 quy tắc tiền xử lý) =====
# ----- IMPORT dùng cho TOÀN BỘ bộ công cụ tiền xử lý 0.4 → 0.12 (copy ô này trước khi copy các ô sau) -----
import itertools, re, unicodedata, difflib, warnings
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import skew, kurtosis, boxcox
from sklearn.preprocessing import (StandardScaler, MinMaxScaler, RobustScaler, PowerTransformer, PolynomialFeatures,
                                   OneHotEncoder, OrdinalEncoder)
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
warnings.filterwarnings("ignore")
pd.set_option("display.width", 200); pd.set_option("display.max_columns", 30)      # in bảng rộng không bị cắt
RANDOM_STATE = globals().get("RANDOM_STATE", 42)       # dùng lại hằng số ở ô 0.1 nếu đã có

# Quy tắc 1: KHÔNG chỉnh sửa dữ liệu thô  -> self.raw giữ nguyên, mọi thao tác làm trên self.df (bản sao)
# Quy tắc 2: so sánh kết quả TRƯỚC / SAU tiền xử lý -> compare()
# Quy tắc 3: linh hoạt biện pháp -> mỗi hàm trả về self (nối chuỗi) và ghi nhật ký vào self.log
class PrepBase:
    def __init__(self, df, target=None):
        self.raw = df.copy(deep=True)          # QUY TẮC 1: bản thô, không bao giờ sửa
        self.df = df.copy(deep=True)           # bản làm việc
        self.target = target                   # tên cột nhãn/đầu ra (nếu có)
        self.log = []                          # nhật ký các bước đã làm

    # ---------- tiện ích ----------
    def _note(self, msg):
        self.log.append(msg); print("•", msg)

    def num_cols(self, exclude_target=True):
        cols = [c for c in self.df.columns if pd.api.types.is_numeric_dtype(self.df[c])
                and not pd.api.types.is_bool_dtype(self.df[c])]
        return [c for c in cols if not (exclude_target and c == self.target)]

    def cat_cols(self, exclude_target=True):
        cols = [c for c in self.df.columns if not pd.api.types.is_numeric_dtype(self.df[c])
                and not pd.api.types.is_datetime64_any_dtype(self.df[c])]
        cols += [c for c in self.df.columns if pd.api.types.is_bool_dtype(self.df[c])]
        return [c for c in cols if not (exclude_target and c == self.target)]

    def get(self):
        return self.df

    def reset(self):
        self.df = self.raw.copy(deep=True); self.log = []; return self

    def compare(self):
        """QUY TẮC 2: so sánh TRƯỚC / SAU tiền xử lý."""
        a, b = self.raw, self.df
        print(f"Kích thước  : {a.shape}  ->  {b.shape}")
        print(f"Ô thiếu     : {int(a.isna().sum().sum())}  ->  {int(b.isna().sum().sum())}")
        print(f"Dòng trùng  : {int(a.duplicated().sum())}  ->  {int(b.duplicated().sum())}")
        common = [c for c in self.num_cols(False) if c in a.columns]
        if common:
            t = pd.DataFrame({"mean_trước": a[common].mean(), "mean_sau": b[common].mean(),
                              "std_trước": a[common].std(), "std_sau": b[common].std(),
                              "skew_trước": a[common].skew(), "skew_sau": b[common].skew()})
            print(t.round(3))

    # ---------- BIẾN ĐỔI TOÁN HỌC (dùng chung cho xử lý outlier & chỉnh dạng) ----------
    def log_transform(self, cols, shift_if_needed=True):
        """Log: dữ liệu DƯƠNG, lệch phải, biến thiên mạnh. Có số 0 -> log1p. Có số âm -> dịch trước (shift)."""
        for c in cols:
            mn = self.df[c].min()
            if mn <= 0:
                if not shift_if_needed: raise ValueError(f"{c} có giá trị <= 0, không log được")
                self.df[c] = np.log1p(self.df[c] - mn)            # dịch về >= 0 rồi log(1+x)
                self._note(f"log1p(x - min) cho '{c}' (có giá trị <= 0)")
            else:
                self.df[c] = np.log(self.df[c]); self._note(f"log(x) cho '{c}'")
        return self

    def boxcox_transform(self, cols):
        """Box-Cox: kéo về phân phối chuẩn. LƯU Ý 📙: bản chuẩn CHỈ nhận x > 0 (notes ghi 'âm hạn chế' = phải CỘNG HẰNG SỐ DỊCH trước)."""
        for c in cols:
            x = self.df[c]; shift = 0.0
            if x.min() <= 0:
                shift = -x.min() + 1e-6; x = x + shift             # shifted Box-Cox cho số âm ít
            y, lam = boxcox(x.dropna())
            self.df.loc[x.dropna().index, c] = y
            self._note(f"Box-Cox '{c}': lambda={lam:.3f}" + (f", đã dịch +{shift:.3g}" if shift else ""))
        return self

    def yeojohnson_transform(self, cols, standardize=False):
        """Yeo-Johnson: dùng được cả số ÂM lẫn DƯƠNG, không giới hạn; phương sai gần dữ liệu thô."""
        pt = PowerTransformer(method="yeo-johnson", standardize=standardize)
        self.df[cols] = pt.fit_transform(self.df[cols])
        self._note(f"Yeo-Johnson cho {cols}: lambdas={np.round(pt.lambdas_, 3).tolist()}")
        return self

    def skew_report(self):
        """Gợi ý phép biến đổi theo độ lệch (skewness): lệch phải mạnh -> log/Box-Cox/Yeo-Johnson."""
        rows = []
        for c in self.num_cols(False):
            x = self.df[c].dropna(); s = skew(x)
            if np.isnan(s): gy = "cột hằng (phương sai 0) -> bỏ cột"
            elif abs(s) < 0.5: gy = "gần đối xứng -> không cần"
            elif x.min() > 0: gy = "log hoặc Box-Cox" if s > 0 else "bình phương/mũ hoặc Yeo-Johnson"
            else: gy = "Yeo-Johnson (có số <= 0)"
            rows.append((c, round(float(s), 3), round(x.min(), 3), gy))
        return pd.DataFrame(rows, columns=["cột", "skewness", "min", "gợi ý"]).set_index("cột")

# ===== KIỂM ĐỊNH GIẢ THUYẾT (Word Buổi 4 – mục 6) – dùng cho EDA 2 biến & kiểm tra dư thừa khi tích hợp =====
class HypothesisTester:
    """H0 = giả thuyết muốn BÁC BỎ / có dấu '='. p < alpha -> bác bỏ H0 (Word: Sai lầm loại I = bác bỏ H0 đúng)."""
    @staticmethod
    def decide(p, alpha=0.05):
        return f"p={p:.4g} {'<' if p < alpha else '>='} alpha={alpha} -> " + \
               ("BÁC BỎ H0" if p < alpha else "KHÔNG đủ bằng chứng bác bỏ H0")

    @staticmethod
    def z_test(x, mu0, sigma=None, alternative="two-sided", alpha=0.05):
        """z-test 1 mẫu (biết sigma tổng thể hoặc n lớn). alternative: 'two-sided' | 'greater' | 'less'."""
        x = np.asarray(x, float); n = len(x)
        sigma = sigma if sigma is not None else x.std(ddof=1)
        z = (x.mean() - mu0) / (sigma / np.sqrt(n))
        p = {"two-sided": 2 * (1 - stats.norm.cdf(abs(z))), "greater": 1 - stats.norm.cdf(z),
             "less": stats.norm.cdf(z)}[alternative]
        print(f"z={z:.3f}; ", HypothesisTester.decide(p, alpha)); return z, p

    @staticmethod
    def t_test_1sample(x, mu0, alternative="two-sided", alpha=0.05):
        t, p = stats.ttest_1samp(x, mu0, alternative=alternative)
        print(f"t={t:.3f}; ", HypothesisTester.decide(p, alpha)); return t, p

    @staticmethod
    def t_test_2groups(a, b, equal_var=False, alpha=0.05):
        t, p = stats.ttest_ind(a, b, equal_var=equal_var)        # Welch (equal_var=False) an toàn hơn
        print(f"t={t:.3f}; ", HypothesisTester.decide(p, alpha)); return t, p

    @staticmethod
    def anova(*groups, alpha=0.05):
        f, p = stats.f_oneway(*groups)
        print(f"F={f:.3f}; ", HypothesisTester.decide(p, alpha)); return f, p

    @staticmethod
    def chi_square(df, a, b, alpha=0.05):
        """2 biến categorical: bảng chéo (cross-table) + chi-bình-phương độc lập."""
        tab = pd.crosstab(df[a], df[b])
        chi2, p, dof, _ = stats.chi2_contingency(tab)
        print(tab); print(f"chi2={chi2:.3f}, dof={dof}; ", HypothesisTester.decide(p, alpha)); return chi2, p

print("Đã nạp PrepBase + HypothesisTester")
```

#### 0.4b DỮ LIỆU "BẨN" GIẢ LẬP – chỉ để chạy thử các ô bên dưới (đi thi: thay bằng `df` của đề)

### 0.5 EDA TỔNG QUÁT  *(BUỔI 4 §2: nhận diện cấu trúc → chức năng → chất lượng → đơn biến/đa biến)*
**Tiêu chí dữ liệu tốt (📕):** *nội tại* = **chính xác · đầy đủ · nhất quán** · *ngữ cảnh* = **cập thời · hữu dụng** · *truy cập* = **đúng đắn · toàn vẹn · trực quan · an toàn**.
**EDA ≠ mô tả (📕):** *khám phá* tập trung **chất lượng** (kích thước, quy mô, phân bố có ổn không); *mô tả* tập trung **ngữ nghĩa**.

| Cần xem | Hàm | Kết luận thế nào |
|---|---|---|
| Kiểu biến + input/output | `identify_variables()` | Chỉ là **gợi ý** – đọc đề để chốt (mã vùng `700000` là định danh, không phải số) |
| Thiếu / outlier / bất nhất | `quality()` | Ghi lại **số lượng + %** rồi chọn kỹ thuật theo **Bảng 0B** |
| Độ lệch → có cần biến đổi? | `skew_report()` | Lệch phải mạnh → log / Box-Cox / Yeo-Johnson |
| 2 biến bất kỳ | `bivariate(a, b)` | **Tự chọn** Pearson+Spearman / chi² / t-test / ANOVA theo kiểu 2 biến |

```python
# ===== 0.5 EDA TỔNG QUÁT: nhận diện biến -> chất lượng dữ liệu -> đơn biến -> đa biến =====
# Word Buổi 4 (mục 2): Bước 1 nhận diện CẤU TRÚC · Bước 2 nhận diện CHỨC NĂNG · Bước 3 đánh giá chất lượng · Bước 4 thu gọn
class EDAReport(PrepBase):

    # ---- Bước 1: nhận diện cấu trúc (kiểu dữ liệu) ----
    def identify_variables(self, discrete_max=10):
        """Đoán loại biến. ĐÂY CHỈ LÀ GỢI Ý – phải đọc đề để kết luận (vd: mã vùng 700000 là định danh, không phải số)."""
        rows = []
        for c in self.df.columns:
            s = self.df[c]; nun = s.nunique(dropna=True)
            if pd.api.types.is_datetime64_any_dtype(s): kind = "thời gian (date-time)"
            elif pd.api.types.is_bool_dtype(s) or nun == 2: kind = "nhị phân (binary)"
            elif pd.api.types.is_numeric_dtype(s):
                if nun == len(s.dropna()) and nun > 20 and (s.dropna() % 1 == 0).all(): kind = "định danh? (giống ID)"
                elif (s.dropna() % 1 == 0).all() and nun <= discrete_max: kind = "định lượng RỜI RẠC (đếm được)"
                else: kind = "định lượng LIÊN TỤC (đo được)"
            else:
                kind = "định danh/định tính (categorical: nominal hoặc ordinal)"
            role = "OUTPUT (target)" if c == self.target else "INPUT (feature)"   # Bước 2: chức năng
            rows.append((c, str(s.dtype), nun, kind, role))
        return pd.DataFrame(rows, columns=["cột", "dtype", "số giá trị khác nhau", "loại (gợi ý)", "chức năng"]).set_index("cột")

    # ---- Bước 3: đánh giá chất lượng (thiếu / bất thường / không nhất quán) ----
    def missing_report(self):
        m = self.df.isna().sum(); t = pd.DataFrame({"số ô thiếu": m, "% thiếu": (m / len(self.df) * 100).round(2)})
        return t[t["số ô thiếu"] > 0].sort_values("% thiếu", ascending=False)

    def outlier_report(self, z=3, k=1.5):
        rows = []
        for c in self.num_cols(False):
            x = self.df[c].dropna()
            zc = ((x - x.mean()) / x.std(ddof=1)).abs() > z                       # 3-sigma
            q1, q3 = x.quantile([.25, .75]); i = q3 - q1
            ic = (x < q1 - k * i) | (x > q3 + k * i)                              # IQR / boxplot
            rows.append((c, int(zc.sum()), int(ic.sum()), round(skew(x), 2)))
        return pd.DataFrame(rows, columns=["cột", f"outlier {z}-sigma", f"outlier IQR×{k}", "skewness"]).set_index("cột")

    def inconsistency_report(self):
        """Phát hiện KHÔNG NHẤT QUÁN: cùng 1 giá trị nhưng viết khác (hoa/thường, khoảng trắng, gần giống nhau)."""
        out = {}
        for c in self.cat_cols():
            s = self.df[c].dropna().astype(str)
            norm = s.str.strip().str.lower().str.replace(r"\s+", " ", regex=True)
            if s.nunique() != norm.nunique():
                out[c] = f"{s.nunique()} giá trị gốc -> {norm.nunique()} sau khi chuẩn hoá hoa/thường/khoảng trắng"
            vals = list(norm.unique())
            sim = [(a, b) for a, b in itertools.combinations(vals, 2) if difflib.SequenceMatcher(None, a, b).ratio() > 0.85]
            if sim: out[c + " (gần giống)"] = sim[:5]
        return out if out else "không thấy bất nhất rõ ràng (vẫn nên kiểm tra ràng buộc nghiệp vụ: khoảng giá trị, ngày…)"

    def quality(self):
        print("Kích thước:", self.df.shape, "| Dòng trùng:", int(self.df.duplicated().sum()))
        const = [c for c in self.df.columns if self.df[c].nunique(dropna=False) <= 1]
        print("Cột hằng (phương sai = 0, vô ích):", const or "không có")
        print("\n-- Dữ liệu THIẾU --\n", self.missing_report() if len(self.missing_report()) else "không thiếu")
        print("\n-- Dữ liệu BẤT THƯỜNG (outlier) --\n", self.outlier_report())
        print("\n-- Dữ liệu KHÔNG NHẤT QUÁN --\n", self.inconsistency_report())

    # ---- Phân tích ĐƠN biến ----
    def univariate_numeric(self):
        d = self.df[self.num_cols(False)]
        t = d.describe().T
        t["median"] = d.median(); t["skew"] = d.skew(); t["kurt(3=chuẩn)"] = d.apply(lambda s: kurtosis(s.dropna(), fisher=False))
        return t.round(3)

    def univariate_categorical(self, top=5):
        for c in self.cat_cols(False):
            print(f"\n[{c}]\n", self.df[c].value_counts(dropna=False).head(top))

    def plots(self, cols=None, ncols=4):
        cols = cols or self.num_cols(False)
        n = len(cols); nr = int(np.ceil(n / ncols))
        fig, ax = plt.subplots(nr * 2, ncols, figsize=(3.2 * ncols, 5 * nr), squeeze=False)
        for i, c in enumerate(cols):
            r, k = divmod(i, ncols)
            ax[2 * r][k].hist(self.df[c].dropna(), bins=20); ax[2 * r][k].set_title(c, fontsize=9)
            ax[2 * r + 1][k].boxplot(self.df[c].dropna(), vert=False)
        for j in range(n, nr * ncols):
            ax[2 * (j // ncols)][j % ncols].axis("off"); ax[2 * (j // ncols) + 1][j % ncols].axis("off")
        plt.tight_layout(); plt.show()

    # ---- Phân tích ĐA biến: tự CHỌN công cụ theo kiểu 2 biến (Word Buổi 4 mục 2 & Buổi 5 mục 3) ----
    def bivariate(self, a, b, alpha=0.05):
        A, B = self.df[a], self.df[b]; na, nb = pd.api.types.is_numeric_dtype(A), pd.api.types.is_numeric_dtype(B)
        if na and nb:                                         # số – số
            d = self.df[[a, b]].dropna()
            r, p = stats.pearsonr(d[a], d[b]); rho, p2 = stats.spearmanr(d[a], d[b])
            print(f"[số–số] Pearson r={r:.3f} (p={p:.3g}) | Spearman rho={rho:.3f} (p={p2:.3g}) | Cov={d[a].cov(d[b]):.4g}")
            print("  -> Pearson: quan hệ TUYẾN TÍNH, dữ liệu số gần chuẩn. Spearman: thứ hạng / có outlier / không chuẩn. Cov phụ thuộc đơn vị -> chỉ xem chiều.")
        elif not na and not nb:                               # chữ – chữ
            print("[chữ–chữ] cross-table + chi-bình-phương"); HypothesisTester.chi_square(self.df, a, b, alpha)
        else:                                                 # số – chữ
            num, cat = (a, b) if na else (b, a)
            groups = [g[num].dropna().values for _, g in self.df.groupby(cat)]
            print(f"[số–chữ] '{num}' theo {len(groups)} nhóm của '{cat}'")
            if len(groups) == 2: HypothesisTester.t_test_2groups(*groups, alpha=alpha)
            elif len(groups) > 2: HypothesisTester.anova(*groups, alpha=alpha)
            self.df.boxplot(column=num, by=cat); plt.suptitle(""); plt.show()      # parallel boxplots

    def correlation_matrix(self, method="pearson", plot=True):
        c = self.df[self.num_cols(False)].corr(method=method)
        if plot:
            plt.figure(figsize=(5, 4)); plt.imshow(c, cmap="coolwarm", vmin=-1, vmax=1); plt.colorbar()
            plt.xticks(range(len(c)), c.columns, rotation=90, fontsize=7); plt.yticks(range(len(c)), c.columns, fontsize=7)
            plt.title(f"Ma trận tương quan ({method}) – đường chéo = 1"); plt.tight_layout(); plt.show()
        return c.round(3)

    def run_all(self):
        print("=== BƯỚC 1–2: CẤU TRÚC & CHỨC NĂNG ===\n", self.identify_variables())
        print("\n=== BƯỚC 3: CHẤT LƯỢNG DỮ LIỆU ==="); self.quality()
        print("\n=== ĐƠN BIẾN (số) ===\n", self.univariate_numeric())
        return self
# ----- chạy thử -----
eda = EDAReport(dirty, target="buy")
eda.run_all()                                   # bước 1-2-3 + đơn biến
print(eda.skew_report())                        # gợi ý biến đổi theo độ lệch
eda.bivariate("age", "income")                  # số–số   -> Pearson / Spearman / Cov
eda.bivariate("gender", "buy")                  # chữ–chữ -> bảng chéo + chi-bình-phương
eda.bivariate("income", "buy")                  # số–chữ  -> t-test (2 nhóm); >=3 nhóm -> ANOVA
print(eda.correlation_matrix(plot=False))
eda.plots(["age", "income", "score"], ncols=3)  # histogram + boxplot
HypothesisTester.z_test(dirty["score"].dropna(), mu0=60, alternative="greater")   # H0: μ <= 60 ; Ha: μ > 60
```

### 0.6 XỬ LÝ DỮ LIỆU THIẾU  *(BUỔI 4 §4.1)*
**Tác động của thiếu (📕):** kết quả **thiên lệch (bias)**; vi phạm điều kiện đầu vào thuật toán (vd thiếu làm lệch phân phối chuẩn của hồi quy tuyến tính).

| Phương pháp | Ưu | Nhược / điều kiện | Hàm |
|---|---|---|---|
| **Giữ nguyên** | Không làm sai lệch, đỡ dữ liệu thưa | Chỉ khi ô thiếu **không liên quan** yêu cầu & thuật toán chịu được NaN | — |
| **Bỏ dòng** | Mẫu cùng quy mô, công bằng | Mất dữ liệu → *hạn chế dùng* | `drop_rows` |
| **Bỏ cột** | Gọn | Chỉ khi thiếu **> ~20%** và không tương quan target | `drop_cols` |
| **Giá trị đặc biệt / thống kê** (`unknown`, mean, median, mode) | Nhanh | Làm **co** phương sai; mean bị outlier kéo | `fill_constant`, `fill_stat` |
| **Theo nhóm** | Giữ đặc trưng nhóm (**MAR**) | Cần cột nhóm hợp lý | `fill_group` |
| **kNN** | Dùng dòng gần nhất | Phải **chuẩn hoá**; chậm với dữ liệu lớn | `fill_knn` |
| **MICE** | Tận dụng quan hệ giữa các cột | Cần **số cột thiếu < số cột đủ**; chạy lặp | `fill_mice` |
| **Chuỗi thời gian** | Giữ xu thế | Dữ liệu phải **đã sắp xếp** | `fill_sequence` |

| Cơ chế thiếu (📕) | Ý nghĩa | Xử lý |
|---|---|---|
| **MCAR** | Thiếu hoàn toàn ngẫu nhiên, lẻ tẻ | **Không thay** hoặc điền mean/mode (không thiên lệch); mô hình/MICE hao công & có thể lệch |
| **MAR** | Thiếu ngẫu nhiên nhưng **phụ thuộc cột khác** | Trung bình **lân cận cùng tính chất** (theo nhóm) |
| **MNAR** | Thiếu **phụ thuộc chính giá trị đó** (lương quá cao/thấp) | Mô hình tìm **người tương đồng**; **bổ sung tri thức ngoài** |

*(📕 còn nêu: thiếu do **mất sau khi thu thập** — lỗi trích rút/logic, có quy luật · dữ liệu **chưa biết** · **không thể thu thập**.)*

```python
# ===== 0.6 XỬ LÝ DỮ LIỆU THIẾU (Word Buổi 4 – mục 4.1) =====
from sklearn.impute import KNNImputer
from sklearn.experimental import enable_iterative_imputer          # noqa  (bắt buộc để import IterativeImputer)
from sklearn.impute import IterativeImputer

class MissingHandler(PrepBase):

    def report(self):
        m = self.df.isna().sum()
        t = pd.DataFrame({"số ô thiếu": m, "% thiếu": (m / len(self.df) * 100).round(2)})
        return t[t["số ô thiếu"] > 0].sort_values("% thiếu", ascending=False)

    # ---- đoán CƠ CHẾ thiếu: MCAR / MAR / MNAR ----
    def diagnose(self, col, alpha=0.05):
        """So sánh các cột khác giữa nhóm 'ô bị thiếu' và 'ô không thiếu'. Có khác biệt -> MAR; không khác -> MCAR (hoặc MNAR!).
        MNAR (thiếu vì CHÍNH giá trị đó quá cao/thấp, vd khai lương) KHÔNG kiểm tra được từ dữ liệu -> phải dùng tri thức ngoài."""
        miss = self.df[col].isna(); hits = []
        for c in self.df.columns:
            if c == col or self.df[c].isna().all(): continue
            if pd.api.types.is_numeric_dtype(self.df[c]):
                a, b = self.df.loc[miss, c].dropna(), self.df.loc[~miss, c].dropna()
                if len(a) > 2 and len(b) > 2 and stats.ttest_ind(a, b, equal_var=False)[1] < alpha: hits.append(c)
            else:
                tab = pd.crosstab(miss, self.df[c])
                if tab.shape[0] > 1 and tab.shape[1] > 1 and stats.chi2_contingency(tab)[1] < alpha: hits.append(c)
        if hits: print(f"'{col}': thiếu có liên quan tới {hits} -> nghi MAR -> điền THEO NHÓM / kNN / MICE")
        else: print(f"'{col}': chưa thấy liên quan cột khác -> nghi MCAR (hoặc MNAR) -> điền mean/median/mode; nếu MNAR thì thêm cột cờ + tri thức ngoài")
        return hits

    def suggest(self, max_missing=0.20):
        """Gợi ý nhanh từng cột thiếu (đúng thứ tự ưu tiên trong Word)."""
        for c, r in (self.df.isna().mean()[lambda s: s > 0]).items():
            num = pd.api.types.is_numeric_dtype(self.df[c])
            if r > max_missing: g = f"thiếu {r:.0%} > {max_missing:.0%} -> cân nhắc BỎ CỘT (nếu không tương quan target)"
            elif not num: g = "categorical -> mode (hoặc 'unknown' nếu thiếu mang ý nghĩa)"
            else: g = "số -> median (lệch/outlier) hoặc mean (đối xứng); MAR -> theo nhóm/kNN/MICE"
            print(f"  {c}: {r:.1%} | {g}")

    # ---- 1) GIỮ NGUYÊN: không làm gì (nếu ô thiếu KHÔNG liên quan yêu cầu bài) ----

    # ---- 2) LOẠI BỎ dòng / cột ----
    def drop_rows(self, cols=None, how="any"):
        """cols=None: xét mọi cột. Dùng khi ô thiếu nằm ở cột LIÊN QUAN TRỰC TIẾP yêu cầu (vd phân tích theo giới tính thì bỏ dòng thiếu giới tính)."""
        n0 = len(self.df); self.df = self.df.dropna(subset=cols, how=how)
        self._note(f"Bỏ {n0 - len(self.df)} dòng thiếu (cols={cols}, how={how})"); return self

    def drop_cols(self, max_missing=0.20, keep=()):
        """Bỏ cột thiếu quá ngưỡng (mức hợp lý 20%). Trước khi bỏ nên xem tương quan với target (EDAReport.bivariate)."""
        bad = [c for c in self.df.columns if self.df[c].isna().mean() > max_missing and c not in keep and c != self.target]
        self.df = self.df.drop(columns=bad); self._note(f"Bỏ cột thiếu > {max_missing:.0%}: {bad}"); return self

    # ---- 3) THAY bằng giá trị đặc biệt / thống kê mô tả ----
    def fill_constant(self, cols, value="unknown"):
        for c in cols: self.df[c] = self.df[c].fillna(value)
        self._note(f"Điền hằng '{value}' cho {cols}"); return self

    def fill_stat(self, cols=None, strategy="auto"):
        """strategy: 'mean' | 'median' | 'mode' | 'auto' (số: mean nếu |skew|<0.5 ngược lại median; chữ: mode)."""
        for c in (cols or self.df.columns[self.df.isna().any()].tolist()):
            if c == self.target: continue
            s = self.df[c]; num = pd.api.types.is_numeric_dtype(s)
            st = strategy if strategy != "auto" else (("mean" if abs(skew(s.dropna())) < 0.5 else "median") if num else "mode")
            v = {"mean": s.mean, "median": s.median}.get(st, lambda: s.mode().iloc[0])()
            self.df[c] = s.fillna(v); self._note(f"'{c}': điền {st} = {v:.4g}" if num else f"'{c}': điền mode = {v}")
        return self

    # ---- 4) THAY THEO NHÓM (MAR: thiếu phụ thuộc cột khác) ----
    def fill_group(self, col, by, stat="mean"):
        """Điền theo trung bình / trung vị / mode của NHÓM có cùng tính chất (cột `by`)."""
        if pd.api.types.is_numeric_dtype(self.df[col]):
            g = self.df.groupby(by)[col].transform(stat)
            self.df[col] = self.df[col].fillna(g).fillna(getattr(self.df[col], stat)())
        else:
            g = self.df.groupby(by)[col].transform(lambda s: s.mode().iloc[0] if s.notna().any() else np.nan)
            self.df[col] = self.df[col].fillna(g).fillna(self.df[col].mode().iloc[0])
        self._note(f"'{col}': điền {stat} theo nhóm '{by}'"); return self

    # ---- 5) DỮ LIỆU CHUỖI / THỜI GIAN ----
    def fill_sequence(self, cols, method="ffill", period=None):
        """'ffill' = lấy giá trị ngày trước; 'bfill' = ngày sau; 'interpolate' = nối đường thẳng giữa trước & sau (nội suy tuyến tính);
        'same_period' = cùng kỳ năm trước (period = số dòng của 1 chu kỳ, vd 12 tháng / 365 ngày). Dữ liệu phải ĐÃ SẮP XẾP theo thời gian."""
        for c in cols:
            if method == "interpolate": self.df[c] = self.df[c].interpolate(method="linear", limit_direction="both")
            elif method == "same_period": self.df[c] = self.df[c].fillna(self.df[c].shift(period))
            else: self.df[c] = getattr(self.df[c], method)()
        self._note(f"Điền chuỗi thời gian {cols} bằng {method}"); return self

    # ---- 6) kNN: đoán ô thiếu từ các dòng GẦN NHẤT ----
    def fill_knn(self, cols=None, n_neighbors=5):
        """Chỉ cột SỐ. Chuẩn hoá trước (khoảng cách nhạy thang đo) rồi trả về thang gốc."""
        cols = cols or self.num_cols(False); sc = StandardScaler()
        Z = sc.fit_transform(self.df[cols])                                     # StandardScaler bỏ qua NaN khi tính mean/std
        Z = KNNImputer(n_neighbors=n_neighbors, weights="distance").fit_transform(Z)
        self.df[cols] = sc.inverse_transform(Z); self._note(f"kNN-impute (k={n_neighbors}) cho {cols}"); return self

    # ---- 7) MICE (điền lặp bằng hồi quy) ----
    def fill_mice(self, cols=None, max_iter=10):
        """Quy trình MICE: điền tạm ngẫu nhiên -> lần lượt coi mỗi cột thiếu là target, hồi quy từ các cột còn lại -> điền lại -> lặp.
        Điều kiện (Word): số cột thiếu < số cột đầy đủ."""
        cols = cols or self.num_cols(False)
        n_miss = int(self.df[cols].isna().any().sum())
        if n_miss >= len(cols) - n_miss: print("⚠️ Cột thiếu >= cột đầy đủ: MICE kém tin cậy")
        imp = IterativeImputer(max_iter=max_iter, random_state=RANDOM_STATE, sample_posterior=False)
        self.df[cols] = imp.fit_transform(self.df[cols]); self._note(f"MICE (max_iter={max_iter}) cho {cols}"); return self

    # ---- 8) MNAR: giữ lại THÔNG TIN "đã từng thiếu" ----
    def add_missing_flags(self, cols=None):
        for c in (cols or self.df.columns[self.df.isna().any()].tolist()):
            self.df[c + "_isna"] = self.df[c].isna().astype(int)
        self._note(f"Thêm cột cờ *_isna (giữ thông tin 'bị thiếu' – hữu ích khi MNAR). Gọi TRƯỚC khi điền."); return self
# ----- chạy thử -----
m = MissingHandler(dirty, target="buy")
print(m.report()); m.suggest()
for c in ["age", "income", "city"]: m.diagnose(c)                      # đoán cơ chế thiếu
m1 = (MissingHandler(dirty, "buy").fill_group("income", "edu", "median")   # MAR: điền theo nhóm
      .fill_knn(["age", "score", "spend"]).fill_stat(["city"]))            # kNN cho số, mode cho chữ
m1.compare()
m2 = MissingHandler(dirty, "buy").fill_mice(["age", "income", "score", "spend"])   # MICE
m3 = MissingHandler(dirty, "buy").drop_cols(0.10).drop_rows(["age"])                # bỏ cột >10% thiếu, bỏ dòng thiếu age
print("Sau drop:", m3.df.shape, "| dữ liệu thô vẫn nguyên:", m3.raw.isna().sum().sum(), "ô thiếu")
```

### 0.7 XỬ LÝ DỮ LIỆU BẤT THƯỜNG – OUTLIER  *(BUỔI 4 §4.2)*
**3 mức bất thường (📕):** ① **attribute**: khác tổng thể · ② **item**: không khác tổng thể nhưng khác **lân cận** · ③ **feature**: vô ích (**biến thiên quá thấp**).
EDA chỉ **nhận diện**; còn *tích cực hay tiêu cực* thì **quyết định theo bản chất** (lỗi nhập hay giá trị thật).

| Phát hiện bằng | Dùng khi | Lưu ý |
|---|---|---|
| **3-sigma** (z-test) | 1 biến **gần chuẩn** | Khử đơn vị; dùng `ddof=1` |
| **IQR** (boxplot) | 1 biến **lệch** | Không giả định chuẩn |
| **Euclid tới trọng tâm** | 2+ biến **cùng thang đo** | Vùng hình tròn, `R = mean(d)+k·std(d)` |
| **Mahalanobis** | 2+ biến **khác thang / tương quan** | Vùng bầu dục; **≥ 3 chiều kém hiệu quả** |
| **Chebyshev** | **Không biết phân phối** | Chỉ cho **tỉ lệ giữ lại tối thiểu** |

| Xử lý bằng | Bản chất | Điều kiện | Hàm |
|---|---|---|---|
| **Log** | Nén khoảng biến thiên, kéo lệch phải về chuẩn | **Chỉ số dương** (có 0 → log1p) | `log_transform` |
| **Box-Cox** | Kéo về chuẩn | Chuẩn: `x>0` (âm nhỏ: dịch hằng số) | `boxcox_transform` |
| **Yeo-Johnson** | Kéo về chuẩn, phương sai gần dữ liệu thô | **Âm/dương đều được**; hợp outlier & cộng tuyến | `yeojohnson_transform` |
| **Binning** | Làm trơn cục bộ (mean/median bin) | Chọn số bin | `bin_smooth` |
| **Winsorization** | Kéo về ngưỡng (percentile / max-min không-outlier) | Cần chọn ngưỡng | `winsorize` |
| **Chuỗi thời gian** | Hôm trước / cùng kỳ năm trước / nội suy | Đã sắp xếp theo thời gian | `ts_replace` |

```python
# ===== 0.7 XỬ LÝ DỮ LIỆU BẤT THƯỜNG – OUTLIER (Word Buổi 4 – mục 4.2) =====
# EDA chỉ NHẬN DIỆN bất thường (không quan tâm tốt/xấu). Quyết định bỏ / giữ / biến đổi phụ thuộc mục tiêu & bản chất (lỗi nhập hay giá trị thật).
class OutlierHandler(PrepBase):          # kế thừa log_transform / boxcox_transform / yeojohnson_transform từ PrepBase

    # ---------- PHÁT HIỆN ----------
    def zscore_mask(self, cols=None, k=3):
        """Quy tắc 3-SIGMA (z-test, khử đơn vị): |z| > k -> outlier. Hợp khi 1 biến GẦN CHUẨN. Dùng ddof=1 (mẫu)."""
        cols = cols or self.num_cols(); x = self.df[cols]
        return ((x - x.mean()) / x.std(ddof=1)).abs() > k

    def iqr_mask(self, cols=None, k=1.5):
        """Quy tắc IQR / boxplot: ngoài [Q1 - k·IQR, Q3 + k·IQR]. Hợp khi dữ liệu LỆCH / không chuẩn."""
        cols = cols or self.num_cols(); x = self.df[cols]
        q1, q3 = x.quantile(.25), x.quantile(.75); i = q3 - q1
        return (x < q1 - k * i) | (x > q3 + k * i)

    def centroid_mask(self, cols, k=3):
        """2+ biến CÙNG THANG ĐO: vẽ 'hình tròn' – tính trọng tâm (vector trung bình), khoảng cách Euclid tới tâm, R = mean(d) + k·std(d)."""
        X = self.df[cols].dropna(); d = np.linalg.norm(X.values - X.values.mean(axis=0), axis=1)
        R = d.mean() + k * d.std(ddof=1)
        return pd.Series(d > R, index=X.index).reindex(self.df.index, fill_value=False)

    def mahalanobis_mask(self, cols, alpha=0.001):
        """2+ biến KHÁC THANG ĐO / có tương quan: 'hình bầu dục' (Mahalanobis). d² ~ chi-bình-phương(bậc tự do = số cột).
        📕 Nhiều chiều (>=3) kém hiệu quả: gần nhau ở 2 chiều vẫn có thể rất xa ở chiều thứ 3."""
        X = self.df[cols].dropna(); mu = X.values.mean(axis=0); VI = np.linalg.pinv(np.cov(X.values.T))
        diff = X.values - mu; d2 = np.einsum("ij,jk,ik->i", diff, VI, diff)
        return pd.Series(d2 > stats.chi2.ppf(1 - alpha, df=len(cols)), index=X.index).reindex(self.df.index, fill_value=False)

    @staticmethod
    def chebyshev_keep(z):
        """Bất đẳng thức Chebyshev (không cần biết phân phối): tỷ lệ dữ liệu NẰM TRONG z độ lệch chuẩn >= 1 - 1/z². (z >= 1)
        Nếu cận trái/phải khác nhau -> lấy z NHỎ HƠN (Word)."""
        return 1 - 1 / z ** 2

    def summary(self):
        a, b = self.zscore_mask(), self.iqr_mask()
        return pd.DataFrame({"3-sigma": a.sum(), "IQR×1.5": b.sum(), "skew": self.df[self.num_cols()].skew().round(2)})

    # ---------- XỬ LÝ ----------
    def remove_rows(self, mask):
        """Loại dòng outlier (chỉ khi chắc là LỖI). mask: Series bool hoặc DataFrame bool (dòng nào có True ở bất kỳ cột)."""
        m = mask.any(axis=1) if isinstance(mask, pd.DataFrame) else mask
        n0 = len(self.df); self.df = self.df[~m.reindex(self.df.index, fill_value=False)]
        self._note(f"Loại {n0 - len(self.df)} dòng outlier"); return self

    def to_nan(self, mask):
        """Biến outlier thành NaN rồi điền bằng MissingHandler (hợp với dữ liệu thời gian / giá trị sai logic như age=200)."""
        for c in mask.columns: self.df[c] = self.df[c].mask(mask[c])
        self._note(f"Đặt NaN cho {int(mask.values.sum())} ô outlier (xử lý tiếp như dữ liệu thiếu)"); return self

    def winsorize(self, cols, method="percentile", lower=0.05, upper=0.95, k=1.5):
        """'Kìm hãm' (winsorization) – kéo outlier về NGƯỠNG.
        method='percentile': kẹp ở bách phân vị lower/upper (vd 5%–95%).
        method='iqr': thay bằng GIÁ TRỊ LỚN/NHỎ NHẤT KHÔNG PHẢI outlier (đúng định nghĩa trong Word)."""
        for c in cols:
            x = self.df[c]
            if method == "percentile": lo, hi = x.quantile(lower), x.quantile(upper)
            else:
                q1, q3 = x.quantile(.25), x.quantile(.75); i = q3 - q1
                lo, hi = x[x >= q1 - k * i].min(), x[x <= q3 + k * i].max()
            self.df[c] = x.clip(lo, hi); self._note(f"Winsorize '{c}' [{method}] về [{lo:.4g}, {hi:.4g}]")
        return self

    def bin_smooth(self, col, bins=5, stat="mean", strategy="quantile"):
        """LÀM TRƠN CỤC BỘ (binning): chia thành bins (cùng tần số 'quantile' hoặc cùng độ rộng 'uniform') rồi thay mỗi giá trị bằng mean/median của bin."""
        b = pd.qcut(self.df[col], bins, duplicates="drop") if strategy == "quantile" else pd.cut(self.df[col], bins)
        self.df[col] = self.df.groupby(b, observed=True)[col].transform(stat)
        self._note(f"Làm trơn '{col}' bằng {stat} trong {bins} bins ({strategy})"); return self

    def ts_replace(self, col, mask, method="prev", period=None):
        """Dữ liệu THỜI GIAN: thay outlier bằng ngày trước ('prev') / nội suy giữa trước-sau ('linear') / cùng kỳ năm trước ('same_period', cần period)."""
        s = self.df[col].mask(mask)
        if method == "prev": s = s.ffill()
        elif method == "linear": s = s.interpolate(limit_direction="both")
        else: s = s.fillna(s.shift(period)).ffill()
        self.df[col] = s; self._note(f"Thay outlier của '{col}' bằng {method}"); return self
# ----- chạy thử -----
o = OutlierHandler(dirty, target="buy")
print(o.summary()); print("Chebyshev z=2 giữ lại >=", OutlierHandler.chebyshev_keep(2))
print("age: 3-sigma =", int(o.zscore_mask(["age"]).values.sum()), "| IQR =", int(o.iqr_mask(["age"]).values.sum()))
print("centroid:", int(o.centroid_mask(["age", "score"]).sum()), "| Mahalanobis:", int(o.mahalanobis_mask(["age", "income", "spend"]).sum()))
a = OutlierHandler(dirty, "buy").winsorize(["age", "income"], method="iqr")        # kìm hãm: về max/min KHÔNG phải outlier
b = OutlierHandler(dirty, "buy").log_transform(["income"]).yeojohnson_transform(["age", "spend"])
print(a.df[["age", "income"]].agg(["min", "max"]).T, "\nskew sau log/Yeo-Johnson:", b.df[["income", "age", "spend"]].skew().round(2).to_dict())
```

### 0.8 KHÔNG NHẤT QUÁN + TÍCH HỢP DỮ LIỆU  *(BUỔI 4 §4.3, §5 · BUỔI 5 §3)*
- **Dạng chuẩn 3 (3NF)** khi thiết kế CSDL: không phụ thuộc **bắc cầu** → không trùng lấp thông tin → hạn chế bất nhất. **Metadata** = dữ liệu mô tả dữ liệu (ngày tạo, kích thước…).
- **Tích hợp ngoài** phải **tiền xử lý** trước (khác kiểu, khác đơn vị, khác cách viết).

| Lỗi bất nhất thường gặp | Cách phát hiện | Cách sửa |
|---|---|---|
| Hoa/thường, khoảng trắng, dấu | `EDAReport.inconsistency_report()` | `standardize_text` |
| Viết khác nhưng cùng nghĩa (`HN`, `Ha Noi`) | `suggest_merge` | `map_values` |
| Sai kiểu / sai định dạng ngày | `identify_variables()` | `to_datetime`, `to_numeric` |
| Ngoài miền / vi phạm luật | `check_range`, `check_rule` | đặt `NaN`, hoặc bỏ dòng |
| Trùng dòng | `quality()` | `drop_duplicates` |

```python
# ===== 0.8 DỮ LIỆU KHÔNG NHẤT QUÁN + TÍCH HỢP DỮ LIỆU (Word Buổi 4 – mục 4.3, 5; Buổi 5 – mục 3) =====
class ConsistencyHandler(PrepBase):
    """Làm sạch KHÔNG NHẤT QUÁN: cùng ý nghĩa nhưng viết khác nhau, sai ràng buộc nghiệp vụ, sai kiểu, trùng lặp."""

    @staticmethod
    def _strip_accents(s):
        return "".join(ch for ch in unicodedata.normalize("NFD", s) if unicodedata.category(ch) != "Mn").replace("đ", "d").replace("Đ", "D")

    def standardize_text(self, cols=None, case="lower", strip_accents=False):
        """Cắt khoảng trắng đầu/cuối, gộp khoảng trắng thừa, đổi hoa/thường (case: 'lower'|'upper'|'title'|None)."""
        for c in (cols or self.cat_cols()):
            s = self.df[c].astype("string").str.strip().str.replace(r"\s+", " ", regex=True)
            if strip_accents: s = s.map(lambda v: self._strip_accents(v) if isinstance(v, str) else v)
            if case: s = getattr(s.str, case)()
            self.df[c] = s.astype(object).where(s.notna(), np.nan)
        self._note(f"Chuẩn hoá văn bản {cols or 'mọi cột chữ'} (case={case}, bỏ dấu={strip_accents})"); return self

    def map_values(self, col, mapping):
        """Gộp cách viết khác nhau về 1 giá trị chuẩn: mapping = {'hanoi': 'ha noi', 'hn': 'ha noi'}."""
        self.df[col] = self.df[col].replace(mapping); self._note(f"Gộp giá trị '{col}' theo mapping {mapping}"); return self

    def suggest_merge(self, col, cutoff=0.85):
        """Gợi ý các giá trị 'gần giống' cần gộp (so khớp chuỗi mờ, không phân biệt hoa/thường/dấu cách)."""
        key = lambda v: self._strip_accents(str(v)).lower().strip()
        vals = self.df[col].dropna().astype(str).unique().tolist(); out = []
        for v in vals:
            sim = [w for w in vals if w != v and (key(w) == key(v) or
                   difflib.SequenceMatcher(None, key(v).replace(" ", ""), key(w).replace(" ", "")).ratio() >= cutoff)]
            if sim and (v, sim) not in out and (sim[0], [v]) not in out: out.append((v, sim))
        return out

    def to_datetime(self, col, fmt=None, dayfirst=True):
        """Thống nhất định dạng ngày (lỗi -> NaT). Timestamp = số giây từ 01/01/1970 UTC -> pd.to_datetime(x, unit='s')."""
        self.df[col] = pd.to_datetime(self.df[col], format=fmt, dayfirst=dayfirst, errors="coerce")
        self._note(f"Chuyển '{col}' sang datetime ({int(self.df[col].isna().sum())} ô không hợp lệ -> NaT)"); return self

    def to_numeric(self, cols):
        """Ép kiểu số (chuỗi lỗi như '1,5tr', 'N/A' -> NaN) rồi xử lý như dữ liệu thiếu."""
        for c in cols: self.df[c] = pd.to_numeric(self.df[c], errors="coerce")
        self._note(f"Ép kiểu số cho {cols}"); return self

    def check_range(self, col, lo=None, hi=None, action="nan"):
        """Ràng buộc miền giá trị (vd tuổi 0–120, điểm 0–10). action: 'nan' (đặt NaN) | 'drop' (bỏ dòng) | 'report' (chỉ báo cáo)."""
        bad = pd.Series(False, index=self.df.index)
        if lo is not None: bad |= self.df[col] < lo
        if hi is not None: bad |= self.df[col] > hi
        print(f"'{col}': {int(bad.sum())} giá trị ngoài [{lo}, {hi}]")
        if action == "nan": self.df.loc[bad, col] = np.nan
        elif action == "drop": self.df = self.df[~bad]
        return self

    def check_rule(self, rule, name="ràng buộc", action="report"):
        """Ràng buộc nghiệp vụ tuỳ ý. rule(df) -> Series bool (True = HỢP LỆ). Vd: lambda d: d['ngay_ket_thuc'] >= d['ngay_bat_dau']."""
        ok = rule(self.df); print(f"{name}: {int((~ok).sum())} dòng vi phạm")
        if action == "drop": self.df = self.df[ok]
        return self

    def drop_duplicates(self, subset=None, keep="first"):
        n0 = len(self.df); self.df = self.df.drop_duplicates(subset=subset, keep=keep)
        self._note(f"Bỏ {n0 - len(self.df)} dòng trùng (subset={subset})"); return self

    def drop_constant_cols(self):
        const = [c for c in self.df.columns if self.df[c].nunique(dropna=False) <= 1 and c != self.target]
        self.df = self.df.drop(columns=const); self._note(f"Bỏ cột hằng (phương sai 0): {const}"); return self


class DataIntegrator:
    """TÍCH HỢP nhiều nguồn + nhận diện THÔNG TIN DƯ THỪA (Word Buổi 5 – mục 3)."""

    @staticmethod
    def merge(left, right, on, how="inner", suffixes=("_a", "_b"), validate=None):
        """Nối theo khoá (JOIN). how: inner | left | right | outer. validate='one_to_one'/'one_to_many' để bắt khoá trùng. Kiểm tra: kiểu khoá phải giống nhau."""
        out = left.merge(right, on=on, how=how, suffixes=suffixes, validate=validate)
        print(f"merge[{how}] on {on}: {left.shape} + {right.shape} -> {out.shape}"); return out

    @staticmethod
    def stack(dfs, ignore_index=True):
        """Nối THEO DÒNG các bảng cùng cấu trúc (UNION). Kiểm tra tên cột/kiểu giống nhau."""
        cols = [set(d.columns) for d in dfs]
        if any(c != cols[0] for c in cols): print("⚠️ Các bảng KHÔNG cùng bộ cột:", [sorted(c ^ cols[0]) for c in cols[1:]])
        return pd.concat(dfs, ignore_index=ignore_index)

    @staticmethod
    def redundant_numeric(df, thr=0.9, method="pearson"):
        """Cặp cột SỐ dư thừa: |tương quan| > thr (Pearson: tuyến tính; Spearman: thứ hạng/không chuẩn). Cũng in hiệp phương sai (phụ thuộc đơn vị)."""
        c = df.select_dtypes("number").corr(method=method); rows = []
        for a, b in itertools.combinations(c.columns, 2):
            if abs(c.loc[a, b]) > thr: rows.append((a, b, round(c.loc[a, b], 3), round(df[a].cov(df[b]), 4)))
        return pd.DataFrame(rows, columns=["cột A", "cột B", f"{method} r", "covariance"])

    @staticmethod
    def redundant_categorical(df, alpha=0.05):
        """Cặp cột CHỮ phụ thuộc nhau (chi-bình-phương, p < alpha) + Cramér's V để đo MỨC ĐỘ."""
        cats = [c for c in df.columns if not pd.api.types.is_numeric_dtype(df[c])]; rows = []
        for a, b in itertools.combinations(cats, 2):
            t = pd.crosstab(df[a], df[b]); chi2, p, _, _ = stats.chi2_contingency(t)
            v = np.sqrt(chi2 / (t.values.sum() * (min(t.shape) - 1))) if min(t.shape) > 1 else np.nan
            if p < alpha: rows.append((a, b, round(p, 5), round(v, 3)))
        return pd.DataFrame(rows, columns=["cột A", "cột B", "p-value", "Cramér V"])
# ----- chạy thử -----
c = ConsistencyHandler(dirty, target="buy")
print("Gợi ý gộp city:", c.suggest_merge("city"))
c.standardize_text(["gender", "city"], case="lower", strip_accents=True).map_values("city", {"hanoi": "ha noi"})
print(c.df["gender"].value_counts().to_dict(), c.df["city"].value_counts().to_dict())
c.check_range("age", 0, 100, "nan").drop_duplicates().drop_constant_cols()
print("Ngày tháng:\n", ConsistencyHandler(pd.DataFrame({"d": ["01/02/2025", "31/12/2024", "abc"]})).to_datetime("d").df)
A = pd.DataFrame({"k": [1, 2, 3], "x": [1, 2, 3]}); B = pd.DataFrame({"k": [1, 2, 4], "y": list("abc")})
DataIntegrator.merge(A, B, on="k", how="left")
print(DataIntegrator.redundant_numeric(dirty, 0.9)); print(DataIntegrator.redundant_categorical(c.df))
```

### 0.9 BIẾN ĐỔI DỮ LIỆU DẠNG SỐ  *(BUỔI 5 §4 · BUỔI 6)*
**Khi nào phải biến đổi (📕):** ① cần xét tương quan mà feature **không cùng thang đo** · ② phân phối **quá đặc biệt** (lệch nhiều, biến thiên mạnh) · ③ **đáp ứng đầu vào thuật toán** (hồi quy tuyến tính cần chuẩn). **Phải đảm bảo KHÔNG đổi bản chất bài toán.**

**BẢNG PHÂN BIỆT 4 CÁCH CHUẨN HOÁ**
| | **z-score** | **Min-Max** | **Robust** | **Decimal** |
|---|---|---|---|---|
| Công thức | `(x−μ)/σ` | `(x−L)/(U−L)·(Unew−Lnew)+Lnew` | `(x−median)/IQR` | `x/10^k` |
| Kết quả | mean 0, std 1 (không bị chặn) | **[L_new, U_new]** (thường 0–1) | median 0, IQR 1 | `(−1, 1)` |
| **Điều kiện (📕)** | **Gần chuẩn**, ít outlier | **Không chuẩn**, ít outlier | **Không chuẩn**, **nhiều outlier** | Chỉ cần co về `(−1,1)` |
| Nhạy outlier | Có (μ, σ bị kéo) | **Rất** (L, U là đầu mút) | **Không** (median, IQR bền) | Có |
| Dùng cho | K-NN, SVM, K-means, PCA, GD | Multinomial/Bernoulli NB, mạng nơ-ron | Dữ liệu nhiều ngoại lai | Minh hoạ |

**BẢNG PHÂN BIỆT 3 CÁCH CHỈNH DẠNG BẰNG TOÁN HỌC** *(kéo dữ liệu lệch phải về chuẩn)*
| | **Log** | **Box-Cox** | **Yeo-Johnson** |
|---|---|---|---|
| Dữ liệu | **Dương**, biến thiên mạnh | Dương (⚠️ âm nhỏ: dịch hằng số) | **Dương và âm**, không giới hạn |
| Tham số | không | `λ` (ước lượng) | `λ` (ước lượng) |
| Ghi chú | Có 0 → `log1p` | Kéo mạnh về chuẩn | Phương sai gần dữ liệu thô, ổn định |

**BẢNG PHÂN BIỆT RỜI RẠC HOÁ (binning)**
| | Cùng **độ rộng** | Cùng **tần số** | Theo **tri thức lĩnh vực** | **Supervised** |
|---|---|---|---|---|
| Ý tưởng | Chia miền giá trị thành k đoạn bằng nhau | Mỗi nhóm ~ cùng số quan sát | Ngưỡng có ý nghĩa (18, 35, 60 tuổi) | Cắt sao cho nhóm **thuần theo nhãn** |
| Dùng nhãn? | Không (*unsupervised*) | Không | Không | **Có** |
| Nhạy outlier | **Có** | Không | Không | Không |
| Hàm | `equal_width` (`pd.cut`) | `equal_freq` (`pd.qcut`) | `custom_bins` | `supervised_bins` |

**Làm trơn chuỗi thời gian:** **SMA** = trung bình `k` điểm trước (window **càng lớn càng "duỗi thẳng"**, nhưng không phân biệt gần/xa) · **Centered SMA** = 7 trước + chính nó + 7 sau, chia **15** · **WMA** = có **trọng số** (gần lớn, xa nhỏ).

```python
# ===== 0.9 BIẾN ĐỔI DỮ LIỆU DẠNG SỐ (Word Buổi 5 – mục 4): chuẩn hoá · chỉnh dạng · rời rạc hoá · làm trơn =====
from sklearn.tree import DecisionTreeClassifier as _DTC

class NumericTransformer(PrepBase):     # kế thừa log / Box-Cox / Yeo-Johnson từ PrepBase

    # ---------- 1.1 CHUẨN HOÁ (normalization / scaling) ----------
    def minmax(self, cols, new_min=0.0, new_max=1.0):
        """Min-Max: f(x) = (x-L)/(U-L)·(Unew-Lnew) + Lnew. ĐK: KHÔNG chuẩn, ÍT outlier (outlier làm co cụm các điểm còn lại)."""
        for c in cols:
            L, U = self.df[c].min(), self.df[c].max()
            if U == L: continue                                            # cột hằng: bỏ qua (tránh chia 0)
            self.df[c] = (self.df[c] - L) / (U - L) * (new_max - new_min) + new_min
        self._note(f"Min-Max -> [{new_min}, {new_max}] cho {cols}"); return self

    def decimal_scaling(self, cols):
        """Thập phân: f(x) = x / 10^k, k = số chữ số của phần tử LỚN NHẤT (|x|) -> mọi giá trị nằm trong (-1, 1)."""
        for c in cols:
            mx = self.df[c].abs().max(); k = len(str(int(mx))) if mx >= 1 else 0
            self.df[c] = self.df[c] / 10 ** k; self._note(f"Decimal scaling '{c}': chia 10^{k}")
        return self

    def zscore(self, cols, ddof=0):
        """z-score: (x-μ)/σ, khử đơn vị. ĐK: GẦN CHUẨN, ÍT outlier. ddof=0 (như StandardScaler); ddof=1 nếu đề nói độ lệch chuẩn MẪU."""
        for c in cols: self.df[c] = (self.df[c] - self.df[c].mean()) / self.df[c].std(ddof=ddof)
        self._note(f"z-score (ddof={ddof}) cho {cols}"); return self

    def robust(self, cols):
        """Robust: (x - median) / IQR. ĐK: KHÔNG chuẩn, NHIỀU outlier (median & IQR không bị outlier kéo)."""
        for c in cols:
            q1, q3 = self.df[c].quantile([.25, .75]); self.df[c] = (self.df[c] - self.df[c].median()) / (q3 - q1)
        self._note(f"Robust (median/IQR) cho {cols}"); return self

    def auto_scale(self, cols=None):
        """Tự chọn theo ĐK trong Word: gần chuẩn & ít outlier -> z-score; không chuẩn & ít outlier -> Min-Max; nhiều outlier -> Robust."""
        for c in (cols or self.num_cols()):
            if self.df[c].nunique() <= 1: continue
            x = self.df[c].dropna(); q1, q3 = x.quantile([.25, .75]); i = q3 - q1
            p_out = ((x < q1 - 1.5 * i) | (x > q3 + 1.5 * i)).mean(); sk = abs(skew(x))
            if p_out > 0.05: self.robust([c])
            elif sk < 0.5: self.zscore([c])
            else: self.minmax([c])
        return self

    # ---------- 1.2 CHỈNH DẠNG ----------
    # (log / Box-Cox / Yeo-Johnson: dùng hàm kế thừa từ PrepBase)
    def add_polynomial(self, cols, degree=2, interaction_only=False):
        """Features construction: x -> (x, x², …) và tích chéo x1·x2 (interaction). Dùng cho quan hệ CONG (hồi quy đa thức)."""
        pf = PolynomialFeatures(degree, include_bias=False, interaction_only=interaction_only)
        P = pf.fit_transform(self.df[cols]); names = [n.replace(" ", "*") for n in pf.get_feature_names_out(cols)]   # 'a b' -> 'a*b' 
        new = pd.DataFrame(P, columns=names, index=self.df.index).drop(columns=cols)
        self.df = pd.concat([self.df, new], axis=1); self._note(f"Thêm {len(names) - len(cols)} đặc trưng đa thức/tương tác: {list(new.columns)}"); return self

    def add_interaction(self, a, b, op="*"):
        self.df[f"{a}{op}{b}"] = {"*": self.df[a] * self.df[b], "/": self.df[a] / self.df[b], "+": self.df[a] + self.df[b], "-": self.df[a] - self.df[b]}[op]
        self._note(f"Thêm đặc trưng {a}{op}{b}"); return self

    def split_datetime(self, col):
        d = pd.to_datetime(self.df[col])
        for part in ["year", "month", "day", "dayofweek", "hour"]: self.df[f"{col}_{part}"] = getattr(d.dt, part)
        self._note(f"Tách '{col}' thành year/month/day/dayofweek/hour"); return self

    def binarize(self, col, threshold, new_col=None):
        """Nhị phân hoá: > ngưỡng -> 1, ngược lại 0 (vd làm nhãn cho hồi quy logistic)."""
        self.df[new_col or f"{col}_bin"] = (self.df[col] > threshold).astype(int); self._note(f"Nhị phân hoá '{col}' > {threshold}"); return self

    def round_cols(self, cols, decimals=0):
        for c in cols: self.df[c] = self.df[c].round(decimals)
        self._note(f"Làm tròn {cols} {decimals} chữ số"); return self

    # ---------- 1.3 TỔNG HỢP (aggregation: roll-up / drill-down) ----------
    def aggregate(self, by, agg):
        """Roll-up: gộp lên mức trừu tượng cao hơn. agg ví dụ {'income': ['mean','sum'], 'age': 'median'}. Drill-down = groupby thêm cột chi tiết. TRẢ VỀ bảng mới."""
        return self.df.groupby(by).agg(agg)

    # ---------- 1.4 RỜI RẠC HOÁ (discretization) ----------
    def equal_width(self, col, k=5, labels=None, new_col=None):
        """Cùng ĐỘ RỘNG: chia miền giá trị thành k đoạn bằng nhau (pd.cut). Nhạy outlier."""
        self.df[new_col or f"{col}_w"] = pd.cut(self.df[col], k, labels=labels); self._note(f"Equal-width {k} bins cho '{col}'"); return self

    def equal_freq(self, col, k=5, labels=None, new_col=None):
        """Cùng TẦN SỐ: mỗi bin chứa ~ cùng số quan sát (pd.qcut). Bền với outlier hơn."""
        self.df[new_col or f"{col}_f"] = pd.qcut(self.df[col], k, labels=labels, duplicates="drop"); self._note(f"Equal-frequency {k} bins cho '{col}'"); return self

    def custom_bins(self, col, edges, labels, new_col=None, right=True):
        """Chia theo TRI THỨC LĨNH VỰC (vd tuổi: [0,18,35,60,120] -> trẻ em/thanh niên/trung niên/cao tuổi)."""
        self.df[new_col or f"{col}_g"] = pd.cut(self.df[col], bins=edges, labels=labels, right=right); self._note(f"Chia nhóm '{col}' theo ngưỡng {edges}"); return self

    def supervised_bins(self, col, target, max_bins=4, new_col=None):
        """Supervised binning 📙: dùng CÂY quyết định 1 biến để chọn ngưỡng cắt sao cho các bin 'thuần' theo nhãn."""
        d = self.df[[col, target]].dropna(); t = _DTC(max_leaf_nodes=max_bins, random_state=RANDOM_STATE).fit(d[[col]], d[target])
        cuts = sorted(t.tree_.threshold[t.tree_.threshold != -2]); edges = [-np.inf] + cuts + [np.inf]
        self.df[new_col or f"{col}_s"] = pd.cut(self.df[col], edges); self._note(f"Supervised bins '{col}' theo {target}: ngưỡng {np.round(cuts, 3).tolist()}"); return self

    # ---------- 1.5 LÀM TRƠN chuỗi thời gian (smoothing) ----------
    def sma(self, col, window=7, new_col=None):
        """SMA: trung bình k điểm PHÍA TRƯỚC (gồm hiện tại). Window càng lớn càng 'duỗi thẳng' (mất chi tiết)."""
        self.df[new_col or f"{col}_sma{window}"] = self.df[col].rolling(window, min_periods=1).mean(); return self

    def centered_sma(self, col, half=7, new_col=None):
        """Centered SMA: (half điểm trước + chính nó + half điểm sau) / (2·half+1). half=7 -> chia 15 (đúng ví dụ Word)."""
        self.df[new_col or f"{col}_csma{2 * half + 1}"] = self.df[col].rolling(2 * half + 1, center=True, min_periods=1).mean(); return self

    def wma(self, col, window=5, weights=None, new_col=None):
        """WMA: trung bình CÓ TRỌNG SỐ – gần trọng số lớn, xa trọng số nhỏ (mặc định 1,2,…,window; điểm mới nhất nặng nhất)."""
        w = np.arange(1, window + 1) if weights is None else np.asarray(weights); w = w / w.sum()
        self.df[new_col or f"{col}_wma{window}"] = self.df[col].rolling(window).apply(lambda v: np.dot(v, w), raw=True); return self

    # ---------- 1.6 CROSS-TAB <-> FEATURES-VALUES ----------
    @staticmethod
    def to_crosstab(df, row, col, value=None, aggfunc="count"):
        """Features-Values -> Cross-tab: tên cột của bảng mới = các GIÁ TRỊ của `col`; ô = tần suất (count) hoặc aggfunc của `value`."""
        return pd.crosstab(df[row], df[col]) if value is None else pd.pivot_table(df, index=row, columns=col, values=value, aggfunc=aggfunc)

    @staticmethod
    def to_features_values(ct, id_name="id", var_name="feature", value_name="value", code_map=None):
        """Cross-tab -> Features-Values (melt). code_map đổi mã số về nhãn (vd {0:'Nam',1:'Nữ',2:'Không muốn trả lời'}) – THƯỜNG DÙNG."""
        out = ct.reset_index().melt(id_vars=ct.index.name or "index", var_name=var_name, value_name=value_name)
        if code_map: out[var_name] = out[var_name].map(code_map).fillna(out[var_name])
        return out.rename(columns={ct.index.name or "index": id_name})
# ----- chạy thử -----
t = NumericTransformer(dirty, "buy").minmax(["score"]).decimal_scaling(["income"]).zscore(["spend"]).robust(["age"])
print(t.df[["score", "income", "spend", "age"]].agg(["min", "max", "mean", "std"]).round(3))
print("z-score khớp StandardScaler:", np.allclose(NumericTransformer(dirty).zscore(["score"]).df["score"], StandardScaler().fit_transform(dirty[["score"]]).ravel()))
t2 = NumericTransformer(dirty.dropna(), "buy").add_polynomial(["age", "score"], 2).add_interaction("income", "score", "/").binarize("score", 60)
print(t2.df.columns.tolist()[-5:])
t3 = (NumericTransformer(dirty.dropna(), "buy").equal_width("age", 4).equal_freq("income", 4)
      .custom_bins("age", [-10, 18, 35, 60, 300], ["tre", "thanh nien", "trung nien", "cao tuoi"]).supervised_bins("score", "buy"))
print(t3.df[["age_w", "income_f", "age_g", "score_s"]].head(3))
t4 = NumericTransformer(dirty, "buy").sma("score", 7).centered_sma("score", 7).wma("score", 5)
print(t4.df[["score", "score_sma7", "score_csma15", "score_wma5"]].iloc[5:8].round(2))
ct = NumericTransformer.to_crosstab(dirty, "city", "edu"); print(ct.head(3)); print(NumericTransformer.to_features_values(ct, "city").head(3))
```

### 0.10 MÃ HOÁ DỮ LIỆU PHÂN LOẠI  *(BUỔI 5 §4 – phần 2)*
**BẢNG PHÂN BIỆT CÁC KIỂU MÃ HOÁ**
| Kiểu | Số cột mới | Giữ thứ tự? | Dùng cho | Rủi ro |
|---|---|---|---|---|
| **Label** | 0 (đổi tại chỗ) | **Tạo thứ tự giả** | Nhãn y; biến cho **cây** | Sai bản chất với mô hình tuyến tính/khoảng cách |
| **Ordinal** | 0 | **Có** (do bạn khai báo) | **Định tính / thứ bậc** | Khoảng cách giữa các mức phải hợp lý |
| **Frequency / count** | 0 | Theo tần suất | Biến nhiều giá trị | **Trùng tần suất** → 2 giá trị bị nhập |
| **One-hot** | **k** | Không | **Định danh ít giá trị** | Nhiều cột (chiều tăng) |
| **Dummy** | **k−1** | Không | Hồi quy tuyến tính (tránh **đa cộng tuyến**) | — |
| **Target / Mean / Likelihood / Leave-one-out** | 0 | Theo trung bình target | Biến **rất nhiều** giá trị | ⚠️ **Rò rỉ nhãn**: LOO/K-fold để không dùng chính nhãn của dòng đó |
| **Hashing** | **cố định n** | Không | Rất nhiều giá trị (hàng nghìn) | **Va chạm** hàm băm |

**Mã hoá hợp lệ khi (📕):** thuật toán **hội tụ** — kết quả độc lập với điểm khởi đầu. Với biến **định danh**, giá trị đổi khác nhau theo giá trị gốc (do gán số) nên **cân nhắc**; với **định tính** phải **đảm bảo thứ tự**.

```python
# ===== 0.10 MÃ HOÁ DỮ LIỆU PHÂN LOẠI (Word Buổi 5 – mục 4, phần 2) =====
from sklearn.feature_extraction import FeatureHasher
from sklearn.model_selection import KFold

class CategoricalEncoder(PrepBase):
    """Định DANH (nominal – không thứ tự) khác định TÍNH (ordinal – có thứ tự): mã hoá sai loại sẽ làm sai bản chất bài toán."""

    # ---- Label / Ordinal ----
    def ordinal(self, col, order, new_col=None):
        """ĐỊNH TÍNH (ordinal): phải ĐẢM BẢO THỨ TỰ, vd order=['Low','Medium','High'] -> 0,1,2. Khoảng cách giữa các mức cần cân nhắc (số năm, số bậc…)."""
        m = {v: i for i, v in enumerate(order)}; self.df[new_col or col] = self.df[col].map(m)
        self._note(f"Ordinal '{col}': {m}"); return self

    def label(self, col, new_col=None):
        """Label encoding cho ĐỊNH DANH: gán số 0..k-1. ⚠️ Số tạo ra THỨ TỰ GIẢ -> ảnh hưởng thuật toán dựa khoảng cách/tuyến tính. An toàn với CÂY quyết định."""
        cats = sorted(self.df[col].dropna().unique()); m = {v: i for i, v in enumerate(cats)}
        self.df[new_col or col] = self.df[col].map(m); self._note(f"Label '{col}': {m}"); return self

    def label_target(self, mapping):
        """Mã hoá NHÃN y chữ -> số, vd {'No':0,'Yes':1} (lớp dương/Positive = 1)."""
        self.df[self.target] = self.df[self.target].map(mapping); self._note(f"Nhãn {self.target}: {mapping}"); return self

    # ---- Tần suất ----
    def frequency(self, col, normalize=True, new_col=None):
        """Mã hoá theo TẦN SUẤT xuất hiện. ⚠️ Chỉ ổn khi các giá trị có tần suất KHÁC NHAU (trùng tần suất -> 2 giá trị bị nhập 1)."""
        f = self.df[col].map(self.df[col].value_counts(normalize=normalize)); self.df[new_col or col + "_freq"] = f
        self._note(f"Frequency encoding '{col}'"); return self

    # ---- One-hot / Dummy ----
    def onehot(self, cols, drop_first=False, drop_original=True):
        """One-hot: k giá trị -> k cột 0/1. Dùng cho định DANH ít giá trị (≲10-15). drop_first=True => DUMMY (k-1 cột) – tránh đa cộng tuyến cho hồi quy tuyến tính."""
        d = pd.get_dummies(self.df[cols], columns=cols, drop_first=drop_first, dtype=int)
        self.df = pd.concat([self.df.drop(columns=cols) if drop_original else self.df, d], axis=1)
        self._note(f"{'Dummy (k-1)' if drop_first else 'One-hot (k)'} cho {cols} -> +{d.shape[1]} cột"); return self

    # ---- Target / Mean / Leave-one-out ----
    def target_encode(self, col, method="loo", smoothing=10, n_splits=5, new_col=None):
        """Mã hoá theo TARGET (target y phải là SỐ, vd 0/1). Cần phòng 'rò rỉ nhãn' (leakage):
        'loo'   Leave-One-Out: giá trị dòng i = trung bình y của nhóm KHÔNG gồm chính dòng i (đúng ý Word: không dùng chính nhãn của nó).
        'kfold' Mean encoding tính trên các fold KHÁC (out-of-fold) + làm mượt về trung bình toàn cục.
        'mean'  trung bình nhóm đơn giản (⚠️ rò rỉ mạnh khi nhóm nhỏ)."""
        y = self.df[self.target]; g = self.df.groupby(col)[self.target]; mean_all = y.mean()
        if method == "mean":
            enc = self.df[col].map(g.mean())
        elif method == "loo":
            s, n = g.transform("sum"), g.transform("count"); enc = ((s - y) / (n - 1)).where(n > 1, mean_all)
        else:
            enc = pd.Series(np.nan, index=self.df.index)
            for tr, va in KFold(n_splits, shuffle=True, random_state=RANDOM_STATE).split(self.df):
                st = self.df.iloc[tr].groupby(col)[self.target].agg(["sum", "count"])
                sm = (st["sum"] + smoothing * mean_all) / (st["count"] + smoothing)
                enc.iloc[va] = self.df.iloc[va][col].map(sm).fillna(mean_all).values
        self.df[new_col or f"{col}_te"] = enc; self._note(f"Target encoding [{method}] cho '{col}'"); return self

    # ---- Hashing ----
    def hashing(self, col, n_features=8, drop_original=True):
        """Feature Hashing: băm chuỗi -> vector độ dài CỐ ĐỊNH n_features. Dùng khi feature có RẤT NHIỀU giá trị (hàng nghìn). Có thể va chạm (collision)."""
        H = FeatureHasher(n_features=n_features, input_type="string").transform(self.df[col].astype(str).map(lambda v: [v])).toarray()
        h = pd.DataFrame(H, columns=[f"{col}_h{i}" for i in range(n_features)], index=self.df.index)
        self.df = pd.concat([self.df.drop(columns=[col]) if drop_original else self.df, h], axis=1)
        self._note(f"Hashing '{col}' -> {n_features} cột"); return self
# ----- chạy thử -----
d = dirty.copy(); d["buy"] = (d["buy"] == "Yes").astype(int)               # target số 0/1 để thử target encoding
e = CategoricalEncoder(d, "buy").ordinal("edu", ["Low", "Medium", "High"]).frequency("city")
print(e.df[["edu", "city_freq"]].head(3).T)
e2 = CategoricalEncoder(d, "buy").onehot(["gender", "city"], drop_first=True); print(e2.df.columns.tolist())
for m in ["loo", "kfold", "mean"]:
    print(m, CategoricalEncoder(d, "buy").target_encode("city", m).df["city_te"].describe()[["mean", "std"]].round(3).tolist())
print(CategoricalEncoder(dirty.assign(city=dirty.city.fillna("na")), "buy").hashing("city", 4).df.filter(like="_h").shape)
```

### 0.11 THU GỌN DỮ LIỆU  *(BUỔI 5 §5)*
**Thu gọn (📕) = không thay đổi kết quả đáng kể.** Nén dữ liệu phục vụ **lưu trữ/truyền thông**, *không phải* mục tiêu giảm cho phân tích.

**BẢNG PHÂN BIỆT 4 NHÓM GIẢM CHIỀU**
| | **Filter** | **Wrapper** | **Embedded** | **PCA (rút trích)** |
|---|---|---|---|---|
| Ý tưởng | Chấm điểm từng feature **không cần mô hình** | Thử **tập feature** bằng mô hình, theo sai số | Chọn **trong lúc huấn luyện** | **Tạo cột mới** (tổ hợp tuyến tính) |
| Ví dụ | Variance, tương quan, ANOVA-F, chi², MI, p-value | **RFE**, **forward**, **backward** | **Cây quyết định**, **Lasso**, Ridge | PC1, PC2,… |
| Tốc độ | **Nhanh** | **Chậm** | Vừa | Nhanh |
| Giữ tên feature | Có | Có | Có | **Không** (mất ý nghĩa gốc) |
| Cần chuẩn hoá trước | Chi² cần **không âm** (Min-Max) | Có (mô hình tuyến tính) | Lasso: có | **Bắt buộc** |
| Hàm | `variance_filter`, `correlation_filter`, `filter_kbest`, `filter_pvalue` | `wrapper_rfe`, `wrapper_sequential` | `embedded_tree`, `embedded_lasso` | `pca` |

**Mẹo chọn target (📕):** target có **phương sai quá bé** → bỏ; **quá lớn** → cẩn thận khi lấy mẫu (**overfitting**). Khi bỏ cột phải **xét ý nghĩa**: bắt buộc giữ *số lượng, thành tiền*; nếu bỏ thì bỏ *đơn giá*.

```python
# ===== 0.11 THU GỌN DỮ LIỆU (Word Buổi 5 – mục 5): giảm SỐ DÒNG · giảm SỐ CỘT (chọn lọc) · rút trích (PCA) =====
from sklearn.feature_selection import (VarianceThreshold, SelectKBest, f_classif, f_regression, chi2,
                                       mutual_info_classif, mutual_info_regression, RFE, SequentialFeatureSelector)
from sklearn.linear_model import LogisticRegression, LassoCV
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor

class DataReducer(PrepBase):
    def __init__(self, df, target=None, task="classification"):
        super().__init__(df, target); self.task = task                 # task: 'classification' | 'regression'

    def _XY(self):
        X = self.df[self.num_cols()]
        if X.isna().any().any(): raise ValueError("Còn ô thiếu (NaN) – hãy xử lý dữ liệu thiếu trước khi chọn đặc trưng")
        return X, self.df[self.target]

    def _keep(self, cols, why):
        drop = [c for c in self.num_cols() if c not in cols]
        self.df = self.df.drop(columns=drop); self._note(f"{why}: giữ {len(cols)} cột {list(cols)}; bỏ {drop}"); return self

    # ================= A. GIẢM KHỐI LƯỢNG (số dòng) =================
    def sample(self, n=None, frac=None, stratify=True):
        """Lấy mẫu (sampling). stratify=True: lấy mẫu PHÂN TẦNG theo nhãn (giữ nguyên tỉ lệ nhãn)."""
        frac = frac if frac is not None else n / len(self.df)
        if stratify and self.target in self.df.columns:
            self.df = self.df.groupby(self.target, group_keys=False).sample(frac=frac, random_state=RANDOM_STATE)
        else: self.df = self.df.sample(frac=frac, random_state=RANDOM_STATE)
        self._note(f"Lấy mẫu {frac:.0%} -> {len(self.df)} dòng"); return self

    def cluster_reduce(self, cols, k=10):
        """Clustering: gom k cụm rồi CHỈ LƯU TRỌNG TÂM (+ kích thước cụm) thay cho toàn bộ dòng. TRẢ VỀ bảng mới."""
        sc = StandardScaler(); Z = sc.fit_transform(self.df[cols].dropna())
        km = KMeans(k, n_init=10, random_state=RANDOM_STATE).fit(Z)
        out = pd.DataFrame(sc.inverse_transform(km.cluster_centers_), columns=cols); out["size"] = np.bincount(km.labels_); return out

    def histogram_summary(self, col, bins=10):
        """Rời rạc hoá bằng histogram: lưu (khoảng, tần số) thay cho dữ liệu gốc. TRẢ VỀ bảng mới."""
        cnt, edges = np.histogram(self.df[col].dropna(), bins=bins)
        return pd.DataFrame({"từ": edges[:-1], "đến": edges[1:], "tần số": cnt})

    def parametric_regression(self, x, y):
        """Phương pháp CÓ THAM SỐ: chỉ lưu hàm sinh dữ liệu (hệ số hồi quy) thay vì lưu training set."""
        d = self.df[[x, y]].dropna(); b, a = np.polyfit(d[x], d[y], 1); r2 = np.corrcoef(d[x], d[y])[0, 1] ** 2
        print(f"{y} ≈ {a:.4g} + {b:.4g}·{x}   (R²={r2:.3f})"); return a, b, r2

    # ================= B. GIẢM CHIỀU: CHỌN LỌC THUỘC TÍNH =================
    # ---- FILTER (nhanh, không cần mô hình) ----
    def variance_filter(self, threshold=0.01, scale=True):
        """Bỏ feature phương sai < ngưỡng (gần hằng, không giúp phân biệt). scale=True: so trên thang Min-Max [0,1] để công bằng giữa các đơn vị."""
        X = self.df[self.num_cols()].dropna()
        Z = pd.DataFrame(MinMaxScaler().fit_transform(X), columns=X.columns) if scale else X
        v = Z.var(); print(v.round(4).to_dict()); return self._keep(v[v >= threshold].index, f"Variance filter (<{threshold})")

    def correlation_filter(self, thr=0.9):
        """Trong mỗi cặp |tương quan| > thr: bỏ cột TƯƠNG QUAN VỚI TARGET YẾU HƠN (nếu có target số); không có target -> bỏ cột sau.
        Ví dụ Word: không bỏ 'số lượng', 'thành tiền' -> nếu bỏ thì bỏ 'đơn giá' (cột suy ra được từ 2 cột kia)."""
        cols = self.num_cols(); C = self.df[cols].corr().abs()
        ty = self.df[cols + [self.target]].corr()[self.target].abs() if self.target and pd.api.types.is_numeric_dtype(self.df[self.target]) else None
        drop = set()
        for a, b in itertools.combinations(cols, 2):
            if a in drop or b in drop or not C.loc[a, b] > thr: continue
            drop.add(b if ty is None or ty[a] >= ty[b] else a)
        return self._keep([c for c in cols if c not in drop], f"Correlation filter (>{thr})")

    def _score_fn(self, score):
        return {"f_classif": f_classif, "chi2": chi2, "mutual_info_classif": mutual_info_classif,
                "f_regression": f_regression, "mutual_info_regression": mutual_info_regression}[score]

    def rank_features(self, score=None):
        """Bảng xếp hạng filter: ANOVA-F (phân lớp), F-regression (hồi quy), chi2 (đặc trưng KHÔNG ÂM), mutual information (phi tuyến)."""
        X, y = self._XY(); score = score or ("f_classif" if self.task == "classification" else "f_regression")
        Xs = MinMaxScaler().fit_transform(X) if score == "chi2" else X
        r = self._score_fn(score)(Xs, y); sc, p = (r if isinstance(r, tuple) else (r, [np.nan] * X.shape[1]))
        return pd.DataFrame({"score": sc, "p-value": p}, index=X.columns).sort_values("score", ascending=False)

    def filter_kbest(self, k=5, score=None):
        return self._keep(self.rank_features(score).index[:k], f"Filter top-{k}")

    def filter_pvalue(self, alpha=0.05, score=None):
        r = self.rank_features(score); return self._keep(r.index[r["p-value"] < alpha], f"Filter kiểm định p<{alpha}")

    # ---- WRAPPER (dựa hiệu quả mô hình – chính xác hơn nhưng chậm) ----
    def _est(self, estimator):
        if estimator is not None: return estimator
        return LogisticRegression(max_iter=1000) if self.task == "classification" else LinearRegression()

    def wrapper_rfe(self, n_features=5, estimator=None):
        """RFE (loại đệ quy): mỗi vòng bỏ thuộc tính kém nhất theo hệ số/độ quan trọng của mô hình. Chuẩn hoá trước."""
        X, y = self._XY(); Z = pd.DataFrame(StandardScaler().fit_transform(X), columns=X.columns)
        r = RFE(self._est(estimator), n_features_to_select=n_features).fit(Z, y)
        print("Hạng (1 = giữ):", {c: int(k) for c, k in zip(X.columns, r.ranking_)}); return self._keep(X.columns[r.support_], f"RFE top-{n_features}")

    def wrapper_sequential(self, n_features=5, direction="forward", estimator=None, cv=5):
        """forward = bổ sung dần thuộc tính hữu dụng · backward = loại dần thuộc tính (đúng 2 hướng trong Word)."""
        X, y = self._XY(); Z = pd.DataFrame(StandardScaler().fit_transform(X), columns=X.columns)
        s = SequentialFeatureSelector(self._est(estimator), n_features_to_select=n_features, direction=direction, cv=cv).fit(Z, y)
        return self._keep(X.columns[s.get_support()], f"Sequential {direction} top-{n_features}")

    # ---- EMBEDDED (chọn ngay trong lúc huấn luyện: Cây, Lasso) ----
    def embedded_tree(self, top=5):
        X, y = self._XY()
        m = (RandomForestClassifier if self.task == "classification" else RandomForestRegressor)(200, random_state=RANDOM_STATE).fit(X, y)
        imp = pd.Series(m.feature_importances_, index=X.columns).sort_values(ascending=False); print(imp.round(4).to_dict())
        return self._keep(imp.index[:top], f"Embedded (rừng/cây) top-{top}")

    def embedded_lasso(self):
        """Lasso (L1): ép hệ số các thuộc tính kém về ĐÚNG 0 -> tự loại. Chuẩn hoá trước."""
        X, y = self._XY(); Z = StandardScaler().fit_transform(X)
        if self.task == "classification":
            m = LogisticRegression(penalty="l1", solver="liblinear", C=0.5).fit(Z, y); coef = m.coef_.ravel()
        else: coef = LassoCV(cv=5, random_state=RANDOM_STATE).fit(Z, y).coef_
        s = pd.Series(coef, index=X.columns); print(s.round(3).to_dict()); return self._keep(s.index[s != 0], "Embedded Lasso (hệ số ≠ 0)")

    # ================= C. RÚT TRÍCH THUỘC TÍNH: PCA =================
    def pca(self, n_components=0.95, keep_target=True):
        """PCA: chuẩn hoá -> chiếu lên các thành phần chính (vuông góc, giữ nhiều phương sai nhất). n_components: số nguyên, hoặc tỉ lệ phương sai cần giữ (0.95).
        ⚠️ Cột mới PC1, PC2… mất ý nghĩa gốc (khác chọn lọc)."""
        cols = self.num_cols(); Z = StandardScaler().fit_transform(self.df[cols])
        p = PCA(n_components=n_components, random_state=RANDOM_STATE).fit(Z)
        out = pd.DataFrame(p.transform(Z), columns=[f"PC{i + 1}" for i in range(p.n_components_)], index=self.df.index)
        print("Tỉ lệ phương sai giải thích:", p.explained_variance_ratio_.round(3).tolist(), "| tích luỹ:", p.explained_variance_ratio_.cumsum().round(3)[-1])
        self.pca_model = p
        keep = self.df[[self.target]] if keep_target and self.target in self.df.columns else pd.DataFrame(index=self.df.index)
        self.df = pd.concat([out, keep], axis=1); self._note(f"PCA: {len(cols)} -> {p.n_components_} thành phần"); return self
# ----- chạy thử (cần dữ liệu SỐ, KHÔNG còn NaN, target số) -----
d = dirty.drop(columns=["id", "gender", "city", "edu", "const"]).copy(); d["buy"] = (d["buy"] == "Yes").astype(int)
d = MissingHandler(d, "buy").fill_stat().get(); d["noise"] = np.random.default_rng(1).normal(size=len(d))
print(DataReducer(d, "buy").rank_features().round(3))                       # bảng xếp hạng Filter (ANOVA-F)
for name, f in {"variance": lambda r: r.variance_filter(0.01), "correlation": lambda r: r.correlation_filter(0.9),
                "filter top-3": lambda r: r.filter_kbest(3), "RFE": lambda r: r.wrapper_rfe(3),
                "embedded tree": lambda r: r.embedded_tree(3), "Lasso": lambda r: r.embedded_lasso(), "PCA 90%": lambda r: r.pca(0.9)}.items():
    print("-----", name); print(f(DataReducer(d, "buy")).df.columns.tolist())
print(DataReducer(d, "buy").sample(frac=0.5).df["buy"].mean().round(3), "~", d["buy"].mean().round(3), "(giữ tỉ lệ nhãn)")
print(DataReducer(d, "buy").histogram_summary("score", 4))
```

### 0.12 THỐNG KÊ MÔ TẢ  *(BUỔI 7)*
**Phân tích mô tả (📕)** dùng: ① **giá trị số** (thống kê) · ② **bảng** (truy vấn CSDL) · ③ **biểu đồ**.

**BẢNG PHÂN BIỆT 4 LOẠI TRUNG BÌNH**
| | **Cộng** | **Nhân** | **Điều hoà** | **Trung vị** |
|---|---|---|---|---|
| Hợp khi | Biến thiên **ít** | Biến thiên **lớn**, tăng trưởng | **Tỷ số** 2 đại lượng **khác đơn vị** | Có outlier, lệch |
| Nhận biết | Gần trung vị | **Gần trung vị hơn** cộng khi lệch | Vận tốc, giá/đơn vị | Trung tâm **vị trí** |
| Điều kiện | — | `x > 0` | `x > 0` | — |

**Phương sai (📕):** chia **`n−1`** (bậc tự do: phần tử cuối bị ràng buộc → *không chệch*); chia `n` là coi mọi phần tử tự do. Bình phương để **khử dấu** (giá trị tuyệt đối không có đạo hàm bậc 2 → không có cực trị).
**Đường chéo:** ma trận **hiệp phương sai** = **phương sai** từng biến · ma trận **tương quan** = **1**.
**LU:** `A = P·L·U` · **QR:** `A = Q·R`, `Q` trực giao (`Q⁻¹ = Qᵀ`) · *ma trận hoán vị cũng là ma trận trực giao*; **vector** trực giao ⇔ tích vô hướng bằng 0.

```python
# ===== 0.12 THỐNG KÊ MÔ TẢ (Word Buổi 7): xu thế trung tâm · phân tán · hình dáng · tương quan · LU / QR =====
import scipy.linalg as sla

class DescriptiveStats:
    def __init__(self, df):
        self.df = df

    # ---------- 1. XU THẾ TRUNG TÂM ----------
    def central(self, col):
        """Trung bình cộng / nhân / điều hoà + trung vị + yếu vị. Quy tắc chọn đại diện (Word):
        • Trung bình CỘNG: dữ liệu BIẾN THIÊN ÍT.  • Trung bình NHÂN: biến thiên LỚN (xấp xỉ trung vị hơn), tốc độ tăng trưởng, hệ số nhân.
        • Trung bình ĐIỀU HOÀ: TỶ SỐ giữa 2 đại lượng KHÁC ĐƠN VỊ (vận tốc…).  • Trung vị: có outlier / lệch.  • Yếu vị: KHÔNG chọn khi kiểm tra (biến liên tục không có mode)."""
        x = self.df[col].dropna(); pos = (x > 0).all()
        t = {"mean (cộng)": x.mean(), "geometric (nhân)": stats.gmean(x) if pos else np.nan,
             "harmonic (điều hoà)": stats.hmean(x) if pos else np.nan, "median (trung vị)": x.median(), "mode (yếu vị)": x.mode().iloc[0]}
        s = pd.Series(t); cand = s[["mean (cộng)", "geometric (nhân)", "harmonic (điều hoà)"]].dropna()
        print(s.round(4).to_dict()); print("-> Đại diện gần TRUNG VỊ nhất:", (cand - x.median()).abs().idxmin()); return s

    @staticmethod
    def weighted_means(values, weights):
        """'Cách 2 – gộp mẫu chung': mỗi giá trị đại diện cho `weights` quan sát (vd nhóm tuổi: trung điểm × số người)."""
        x, w = np.asarray(values, float), np.asarray(weights, float)
        return {"cộng": float((w * x).sum() / w.sum()), "nhân": float(np.exp((w * np.log(x)).sum() / w.sum())), "điều hoà": float(w.sum() / (w / x).sum())}

    # ---------- 2. PHÂN TÁN ----------
    def dispersion(self, col, percentiles=(10, 25, 50, 75, 90)):
        """Khoảng biến thiên (range: chỉ 2 đầu mút), bách phân vị, khoảng trải giữa IQR, phương sai & độ lệch chuẩn (ddof=0 quần thể | ddof=1 mẫu)."""
        x = self.df[col].dropna(); q1, q3 = x.quantile([.25, .75])
        t = {"range": x.max() - x.min(), "IQR": q3 - q1, "var (ddof=0, quần thể)": x.var(ddof=0), "var (ddof=1, mẫu)": x.var(ddof=1),
             "std (ddof=0)": x.std(ddof=0), "std (ddof=1)": x.std(ddof=1), "CV = std/mean": x.std(ddof=1) / x.mean()}
        t.update({f"P{p}": np.percentile(x, p) for p in percentiles}); print("(pandas .var() mặc định ddof=1; numpy .var() mặc định ddof=0 – LUÔN ghi rõ ddof)")
        return pd.Series(t).round(4)

    @staticmethod
    def chebyshev(mean, std, lower, upper):
        """Tỉ lệ dữ liệu GIỮ LẠI trong [lower, upper] tối thiểu là 1 - 1/z², z = số độ lệch chuẩn. Cận lệch (không đối xứng) -> lấy z NHỎ HƠN."""
        z = min(abs(upper - mean), abs(mean - lower)) / std
        keep = 1 - 1 / z ** 2 if z > 1 else 0.0
        print(f"z = {z:.3f} -> giữ lại ít nhất {keep:.1%}"); return keep

    # ---------- 3. HÌNH DÁNG PHÂN PHỐI ----------
    def shape(self, col):
        """Skewness Pearson_2 = 3(mean - median)/s. Mean > Median: lệch PHẢI (đuôi bên phải, dương); Mean < Median: lệch TRÁI.
        Kurtosis so với 3 (phân phối chuẩn): >3 nhọn hơn chuẩn · <3 bẹt hơn chuẩn (không nói chung chung)."""
        x = self.df[col].dropna(); p2 = 3 * (x.mean() - x.median()) / x.std(ddof=1); k = kurtosis(x, fisher=False)
        side = "lệch PHẢI (đuôi phải)" if p2 > 0.1 else "lệch TRÁI (đuôi trái)" if p2 < -0.1 else "gần đối xứng"
        print(f"Pearson_2 = {p2:.3f} -> {side} | scipy skew = {skew(x):.3f} | kurtosis = {k:.3f} -> " + ("nhọn hơn chuẩn" if k > 3.2 else "bẹt hơn chuẩn" if k < 2.8 else "gần chuẩn"))
        return p2, k

    def summary(self):
        d = self.df.select_dtypes("number"); t = d.describe().T
        t["IQR"] = d.quantile(.75) - d.quantile(.25); t["skew"] = d.skew(); t["kurt(3=chuẩn)"] = d.apply(lambda s: kurtosis(s.dropna(), fisher=False)); return t.round(3)

    # ---------- 4. TƯƠNG QUAN ----------
    def relation(self, cols=None, ddof=1):
        """Hiệp phương sai (chiều cùng biến thiên, PHỤ THUỘC ĐƠN VỊ -> không so mạnh/yếu) · Pearson (khử đơn vị, tuyến tính) · Spearman (thứ hạng).
        Đường chéo ma trận covariance = PHƯƠNG SAI; đường chéo ma trận tương quan = 1."""
        d = self.df[cols] if cols else self.df.select_dtypes("number")
        return {"cov": d.cov(ddof=ddof).round(4), "pearson": d.corr().round(4), "spearman": d.corr(method="spearman").round(4)}

    @staticmethod
    def spearman_by_rank(x, y):
        """Spearman = Pearson của THỨ HẠNG; đồng hạng -> chia đều thứ hạng (rankdata 'average')."""
        rx, ry = stats.rankdata(x), stats.rankdata(y); return np.corrcoef(rx, ry)[0, 1]

    # ---------- 5. PHÂN RÃ MA TRẬN ----------
    @staticmethod
    def lu(A):
        """A = P·L·U (P hoán vị – cũng là ma trận TRỰC GIAO nên P⁻¹ = Pᵀ; L tam giác dưới; U tam giác trên).
        Không hoán vị được LU chỉ khi MỌI ma trận con chính cấp 1..n có định thức ≠ 0 -> kiểm tra bên dưới. scipy.linalg.lu_factor tiết kiệm bộ nhớ (ma trận gộp + chỉ số)."""
        A = np.asarray(A, float); P, L, U = sla.lu(A)
        minors = [np.linalg.det(A[:k, :k]) for k in range(1, min(A.shape) + 1)]
        print("Định thức các ma trận con chính:", np.round(minors, 4), "-> LU không cần hoán vị:", all(abs(m) > 1e-12 for m in minors))
        print("Kiểm tra A = P@L@U:", np.allclose(A, P @ L @ U)); return P, L, U

    @staticmethod
    def gram_schmidt(A):
        """QR bằng Gram-Schmidt (viết tay). A (m×n, các cột độc lập tuyến tính) = Q·R; Q có cột trực chuẩn; R tam giác trên."""
        A = np.asarray(A, float); m, n = A.shape; Q = np.zeros((m, n)); R = np.zeros((n, n))
        for j in range(n):
            v = A[:, j].copy()
            for i in range(j):
                R[i, j] = Q[:, i] @ A[:, j]; v -= R[i, j] * Q[:, i]
            R[j, j] = np.linalg.norm(v); Q[:, j] = v / R[j, j]
        return Q, R

    @staticmethod
    def qr(A):
        """QR dành cho ma trận CHỮ NHẬT bất kỳ. Q trực giao (Q⁻¹ = Qᵀ – 'tốt hơn' L vì nghịch đảo dễ). So với numpy.linalg.qr."""
        Q, R = DescriptiveStats.gram_schmidt(A); Qn, Rn = np.linalg.qr(np.asarray(A, float))
        print("Gram-Schmidt: A ≈ Q@R:", np.allclose(A, Q @ R), "| QᵀQ = I:", np.allclose(Q.T @ Q, np.eye(Q.shape[1])),
              "| khớp numpy (sai khác dấu):", np.allclose(np.abs(Q), np.abs(Qn))); return Q, R
# ----- chạy thử -----
ds = DescriptiveStats(dirty)
ds.central("income"); ds.central("score")                       # income lệch -> trung bình NHÂN gần trung vị hơn
print(ds.dispersion("score")); DescriptiveStats.chebyshev(60, 12, 36, 90)
ds.shape("income"); ds.shape("score")
r = ds.relation(["age", "score", "spend"]); print(r["cov"]); print("đường chéo tương quan:", np.diag(r["pearson"]))
x, y = [1, 2, 2, 4, 5], [2, 1, 3, 3, 5]; print("Spearman (rank):", round(DescriptiveStats.spearman_by_rank(x, y), 4), "| scipy:", round(stats.spearmanr(x, y)[0], 4))
print(DescriptiveStats.weighted_means([5, 15, 25], [10, 30, 10]))
P, L, U = DescriptiveStats.lu([[2, 1, 1], [4, 3, 3], [8, 7, 9]])
Q, R = DescriptiveStats.qr([[1, 1], [1, 0], [0, 1]]); print(Q.round(3), "\n", R.round(3))
```

### 0.13 PIPELINE AN TOÀN (CHỐNG RÒ RỈ DỮ LIỆU)  *+ chọn chuẩn hoá theo thuật toán*
| Cách làm | Kết quả |
|---|---|
| ❌ Điền thiếu / chuẩn hoá / mã hoá trên **toàn bộ** dữ liệu rồi mới chia train–test | Thông tin Test **lọt vào** Train → điểm Test **đẹp giả** |
| ✅ **Chia trước**, `fit` trên **TRAIN**, chỉ `transform` Test và dữ liệu thực tế | Đánh giá **trung thực**, giống lúc triển khai thật |

> 📕 Notes (Buổi 4–5) trình bày các kỹ thuật trên **toàn bộ dữ liệu** để dễ hiểu; 📙 khi **huấn luyện mô hình** thì dùng `PreprocessPipeline` bên dưới để tránh rò rỉ.

```python
# ===== 0.13 PIPELINE AN TOÀN (CHỐNG RÒ RỈ DỮ LIỆU) + CHỌN CHUẨN HOÁ THEO THUẬT TOÁN =====
# Quy tắc vàng: CHIA TẬP TRƯỚC, rồi FIT (điền thiếu / chuẩn hoá / mã hoá) CHỈ TRÊN TRAIN, chỉ TRANSFORM trên Test và dữ liệu thực tế.
# (Nếu fit trên toàn bộ dữ liệu, thông tin của Test lọt vào Train -> điểm Test đẹp giả.)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Thuật toán -> cách chuẩn hoá (📗 notes ML tr.8 cho NB · 📙 phần còn lại)
SCALER_FOR = {
    "knn": "standard", "svm": "standard", "kmeans": "standard", "kmedoids": "standard", "kmodes": "none (dữ liệu chữ)",
    "dbscan": "standard", "hierarchical": "standard", "fuzzy_cmeans": "standard", "pca": "standard",
    "logistic": "standard", "linear_gd": "standard (GD cần cùng thang để hội tụ)", "lasso_ridge": "standard",
    "multinomial_nb": "minmax (KHÔNG StandardScaler: cần không âm)", "bernoulli_nb": "nhị phân hoá / minmax",
    "gaussian_nb": "none (cần gần chuẩn -> có thể Yeo-Johnson)", "tree": "none", "random_forest": "none",
    "linear_ols": "none (chuẩn hoá chỉ để so sánh độ lớn hệ số)"}

class PreprocessPipeline:
    """fit() = chia train/test + học tham số trên TRAIN; transform() = áp dụng cho dữ liệu mới (có kiểm tra ràng buộc cột)."""

    def __init__(self, target, task="classification", scaler="standard", num_impute="median", cat_impute="most_frequent",
                 ordinal=None, dummy=False, power=None, drop_cols=(), positive=None, test_size=0.2):
        """scaler: 'standard'|'minmax'|'robust'|'none' · num_impute: 'mean'|'median'|'most_frequent'|'constant' · power: None|'yeo-johnson'
        ordinal: {'edu': ['Low','Medium','High']} (cột có THỨ TỰ) · dummy=True -> bỏ 1 cột one-hot · positive: nhãn dương, vd 'Yes' -> 1."""
        self.target, self.task, self.scaler, self.num_impute, self.cat_impute = target, task, scaler, num_impute, cat_impute
        self.ordinal, self.dummy, self.power, self.drop_cols, self.positive, self.test_size = ordinal or {}, dummy, power, list(drop_cols), positive, test_size

    def _build(self, X):
        num = [c for c in X.columns if pd.api.types.is_numeric_dtype(X[c])]
        ordc = [c for c in self.ordinal if c in X.columns]
        nom = [c for c in X.columns if c not in num and c not in ordc]
        sc = {"standard": StandardScaler(), "minmax": MinMaxScaler(), "robust": RobustScaler(), "none": "passthrough"}[self.scaler]
        steps = [("imp", SimpleImputer(strategy=self.num_impute))]
        if self.power: steps.append(("pow", PowerTransformer(method=self.power, standardize=False)))
        steps.append(("sc", sc))
        parts = [("num", Pipeline(steps), num)]
        if ordc: parts.append(("ord", Pipeline([("imp", SimpleImputer(strategy=self.cat_impute)),
                                                ("enc", OrdinalEncoder(categories=[self.ordinal[c] for c in ordc]))]), ordc))
        if nom: parts.append(("nom", Pipeline([("imp", SimpleImputer(strategy=self.cat_impute)),
                                               ("enc", OneHotEncoder(handle_unknown="ignore", sparse_output=False, drop="first" if self.dummy else None))]), nom))
        return ColumnTransformer(parts, verbose_feature_names_out=False).set_output(transform="pandas")

    def _encode_y(self, y):
        if self.task == "regression" or pd.api.types.is_numeric_dtype(y): return y
        if self.positive is not None: return (y == self.positive).astype(int)
        self.classes_ = sorted(y.unique()); return y.map({c: i for i, c in enumerate(self.classes_)})

    def fit(self, df, random_state=RANDOM_STATE):
        X = df.drop(columns=[self.target] + self.drop_cols); y = self._encode_y(df[self.target])
        ok = y.notna(); X, y = X[ok], y[ok]                                         # bỏ dòng không có nhãn
        strat = y if self.task == "classification" else None
        X_tr, X_te, self.y_train, self.y_test = train_test_split(X, y, test_size=self.test_size, stratify=strat, random_state=random_state)
        self.features_ = list(X.columns); self.ct_ = self._build(X_tr)
        self.X_train = self.ct_.fit_transform(X_tr)                                  # FIT chỉ trên TRAIN
        self.X_test = self.ct_.transform(X_te)                                       # Test chỉ TRANSFORM
        print(f"Train {self.X_train.shape} | Test {self.X_test.shape} | scaler={self.scaler} | số cột sau xử lý = {self.X_train.shape[1]}")
        return self

    def transform(self, df_new):
        miss = [c for c in self.features_ if c not in df_new.columns]
        if miss: raise ValueError(f"Dữ liệu mới thiếu cột: {miss}")                  # ràng buộc: đủ cột
        return self.ct_.transform(df_new[self.features_])                            # ràng buộc: đúng thứ tự cột
# ----- chạy thử -----
d = dirty.drop(columns=["id"])
for cfg in [dict(scaler="standard"), dict(scaler="minmax", dummy=True), dict(scaler="robust", power="yeo-johnson", num_impute="mean")]:
    p = PreprocessPipeline("buy", positive="Yes", ordinal={"edu": ["Low", "Medium", "High"]}, **cfg).fit(d)
print(p.X_train.iloc[:2].round(2).T.head(8))
print("Còn NaN? Train:", int(p.X_train.isna().sum().sum()), "Test:", int(p.X_test.isna().sum().sum()))
new = d.drop(columns=["buy"]).iloc[:3]; print("Dữ liệu mới ->", p.transform(new).shape)
try: p.transform(new.drop(columns=["age"]))
except ValueError as e: print("Ràng buộc hoạt động:", e)
```

## PHẦN 1. HỒI QUY TUYẾN TÍNH & ĐA THỨC  *(Notes tr.1–2, BUỔI 1)*
**Ý chính trong notes:**
- Hàm mất mát (loss) = tổng bình phương sai số → **OLS** (tr.1).
- **Gradient Descent:** `w(t+1) = w(t) − ρ · L'(w(t))`. ρ (learning rate) **quá lớn** → nhảy qua lại, không hội tụ; **quá nhỏ** → chậm, hết vòng lặp vẫn chưa tới cực trị (tr.1).
- Tự viết Gradient Descent với ma trận thì **BẮT BUỘC thêm cột 1** vào input (để có w0); `scikit-learn` tự thêm sẵn (tr.1).
- **Hồi quy đa thức** = tạo thêm biến (x², x1·x2, …) để đưa về hồi quy tuyến tính đa biến (tr.1).
- **Độ chệch (bias)** = sai số trên tập Train; **Phương sai (variance)** = sai số trên tập Test (tr.2). Bảng Underfitting/Overfitting (tr.2).

Bài toán minh hoạ: dự đoán `area_mean` từ các đặc trưng khác (biến số → bài toán **hồi quy**, tr.2).

```python
# ===== 1.1 HỒI QUY TUYẾN TÍNH BẰNG GRADIENT DESCENT (tự viết) =====
class LinearRegressionGD:
    """Hồi quy tuyến tính y = w0 + w1*x1 + ... + wd*xd, học w bằng Gradient Descent. (Notes tr.1)"""

    def __init__(self, rho=0.1, n_iter=1000):
        self.rho = rho          # ρ = learning rate (bước nhảy)
        self.n_iter = n_iter    # số vòng lặp
        self.w = None
        self.history = []       # lưu loss mỗi vòng để vẽ biểu đồ hội tụ

    def _add_ones(self, X):
        # BẮT BUỘC thêm cột 1 vào ma trận input để hệ số chặn w0 được học (Notes tr.1)
        return np.c_[np.ones(len(X)), X]

    def fit(self, X, y):
        X = np.asarray(X, float); y = np.asarray(y, float)
        # chuẩn hoá để GD hội tụ nhanh (máy học nên khử đơn vị – Notes tr.15)
        self.mu, self.sd = X.mean(0), X.std(0)
        self.ymu, self.ysd = y.mean(), y.std()
        Xb = self._add_ones((X - self.mu) / self.sd)
        yb = (y - self.ymu) / self.ysd
        n = len(yb)
        self.w = np.zeros(Xb.shape[1])           # khởi tạo w = 0
        self.history = []
        for _ in range(self.n_iter):
            err = Xb @ self.w - yb               # (y_hat - y)
            self.history.append(np.mean(err ** 2))   # loss (trung bình bình phương sai số)
            grad = (2 / n) * Xb.T @ err          # đạo hàm L'(w)
            self.w = self.w - self.rho * grad    # w(t+1) = w(t) - ρ.L'(w(t))
        return self

    def predict(self, X):
        Xb = self._add_ones((np.asarray(X, float) - self.mu) / self.sd)
        return (Xb @ self.w) * self.ysd + self.ymu      # đổi ngược về đơn vị gốc của y

# Dữ liệu hồi quy: 1 biến (đơn biến) và nhiều biến (đa biến)
single_col = ["radius_mean"]
multi_cols = ["radius_mean", "texture_mean", "smoothness_mean", "concavity_mean", "symmetry_mean"]
target_reg = "area_mean"
Xr_tr, Xr_te, yr_tr, yr_te = train_test_split(dm.df[multi_cols], dm.df[target_reg],
                                              test_size=0.2, random_state=RANDOM_STATE)

# ---- Thử 3 mức ρ: nhỏ / vừa / quá lớn (đúng ý notes tr.1) ----
fig, ax = plt.subplots(figsize=(6, 4))
for rho in [0.001, 0.1, 1.05]:
    m = LinearRegressionGD(rho=rho, n_iter=50).fit(Xr_tr[single_col], yr_tr)
    ax.plot(m.history, label=f"ρ={rho}")
ax.set_yscale("log"); ax.set_xlabel("vòng lặp"); ax.set_ylabel("loss"); ax.legend()
ax.set_title("ρ nhỏ: chậm | vừa: hội tụ | quá lớn: phân kỳ"); plt.show()

# ---- So sánh GD với OLS (nghiệm đóng) và scikit-learn ----
gd = LinearRegressionGD(rho=0.1, n_iter=1000).fit(Xr_tr, yr_tr)
sk = LinearRegression().fit(Xr_tr, yr_tr)             # sklearn tự thêm cột 1
Xb = np.c_[np.ones(len(Xr_tr)), Xr_tr.values]
w_ols = np.linalg.pinv(Xb) @ yr_tr.values              # OLS: tối thiểu tổng bình phương sai số
print("Dự đoán 3 mẫu đầu – GD :", gd.predict(Xr_te[:3]).round(1))
print("Dự đoán 3 mẫu đầu – sklearn:", sk.predict(Xr_te[:3]).round(1))
print("Dự đoán 3 mẫu đầu – OLS    :", (np.c_[np.ones(3), Xr_te[:3].values] @ w_ols).round(1))
```

```python
# ===== 1.2 HỒI QUY ĐA THỨC (tạo biến tương tác) – Notes tr.1 =====
class PolynomialRegressionModel:
    """Hồi quy đa thức: tạo thêm biến x^2, x1*x2,... rồi chạy hồi quy tuyến tính đa biến."""

    def __init__(self, degree=2):
        self.degree = degree
        # StandardScaler trước để các luỹ thừa không quá lớn; PolynomialFeatures tạo biến mới
        self.model = Pipeline([("scale", StandardScaler()),
                               ("poly", PolynomialFeatures(degree, include_bias=False)),
                               ("lin", LinearRegression())])

    def fit(self, X, y):
        self.model.fit(X, y); return self

    def predict(self, X):
        return self.model.predict(X)

# Minh hoạ: 3 biến -> y = f(x1², x2², x3², x1x2, x1x3, x2x3, ...) giống ví dụ trong notes
pf = PolynomialFeatures(2, include_bias=False).fit(pd.DataFrame(np.ones((1, 3)), columns=["x1", "x2", "x3"]))
print("Các biến sau khi tạo tương tác:", list(pf.get_feature_names_out(["x1", "x2", "x3"])))

# ---- Bias (sai số Train) vs Variance (sai số Test) theo bậc đa thức – Notes tr.2 ----
from sklearn.metrics import mean_squared_error
rows = []
for d in range(1, 9):
    pm = PolynomialRegressionModel(d).fit(Xr_tr[single_col], yr_tr)
    tr_err = mean_squared_error(yr_tr, pm.predict(Xr_tr[single_col]))     # độ chệch (Train)
    te_err = mean_squared_error(yr_te, pm.predict(Xr_te[single_col]))     # phương sai theo cách gọi của notes (Test)
    rows.append((d, tr_err, te_err, te_err - tr_err))
bv = pd.DataFrame(rows, columns=["bậc", "sai số Train (độ chệch)", "sai số Test (phương sai)", "chênh lệch Test-Train"])
print(bv.round(1).to_string(index=False))
print("\nĐọc bảng (Notes tr.2): Train cao + Test cao = Underfitting | Train thấp + Test cao = Overfitting | cả hai thấp = tốt.")
```

## 🎯 KHI NÀO DÙNG: HỒI QUY TUYẾN TÍNH ↔ ĐA THỨC (và đọc bias–variance)
| Tình huống | Chọn | Dấu hiệu kết quả |
|---|---|---|
| y **số liên tục**, quan hệ x–y ≈ **đường thẳng** | **Tuyến tính** (hàm bậc 1) 📗 tr.1 | Scatter y thực–y dự đoán bám phân giác; R² gần 1 |
| Quan hệ **cong / có tương tác** giữa biến | **Đa thức** = tạo biến x², x1·x2… rồi chạy lại hồi quy tuyến tính đa biến 📗 tr.1 | 📊 Bậc 1: Train MSE **3344** → bậc 2: **205** (giảm mạnh) ⇒ dữ liệu **không** tuyến tính thuần |
| Tăng bậc mà sai số **hết giảm đáng kể** | **Dừng**, chọn bậc nhỏ nhất đủ tốt (ELBOW) 📗 tr.16 | 📊 Bậc 2 → Test **267**; bậc 6 → **224**: giảm ít mà phức tạp hơn nhiều ⇒ ưu tiên **bậc 2** (đơn giản hơn 📗 tr.9) |
| Test − Train **dương và lớn dần** khi tăng bậc | Dấu hiệu bắt đầu **overfitting** | 📊 chênh bậc 2 = **+62**; bậc 3 = **+73** |

**Bảng bias–variance của notes (tr.2):**
| | Độ chệch **THẤP** (Train) | Độ chệch **CAO** |
|---|---|---|
| Phương sai **THẤP** (Test) | ✅ Tốt | **Underfitting** |
| Phương sai **CAO** | **Overfitting** | ❓ Tệ cả hai |

📙 Cách đọc thực dụng: **Train đã cao → underfit** (tăng độ phức tạp); **Train thấp nhưng Test cao hơn hẳn → overfit** (giảm độ phức tạp). *(Notes định nghĩa "phương sai" hai cách: sai số trên Test (tr.2) và chênh Test − Train (tr.16); ý nghĩa thực dụng như trên.)*

**Gradient Descent – dấu hiệu chỉnh `rho` (📗 tr.1):** loss **nhảy qua lại / tăng** ⇒ ρ **quá lớn**; loss **giảm rất chậm, hết vòng lặp vẫn chưa tới cực trị** ⇒ ρ **quá nhỏ**. 📙 Dữ liệu nhỏ–vừa dùng **OLS / sklearn** (nhanh, chính xác – 📊 3 cách ra cùng dự đoán); dữ liệu rất lớn / rất nhiều biến mới cần **GD**.

## PHẦN 2. PHÂN LỚP (CLASSIFICATION) – KHUNG CHUNG  *(Notes tr.2)*
**Ý chính:** Danh sách lớp phải biết trước; Training set gồm *quan sát (dòng)*, *target attribute* và *features*. Có **2 bước**: (1) xây dựng mô hình (huấn luyện) → (2) đánh giá mô hình bằng tập Test.
Phân lớp **nhị phân** (K=2), **đa lớp** (K>2), **đa nhãn** (1 item thuộc nhiều lớp). Bài này là **nhị phân** (Malignant / Benign).

Mình tạo **lớp cha `BaseClassifier`** (fit / predict / scores / accuracy) – mọi thuật toán phân lớp bên dưới đều **kế thừa** nó → đúng tinh thần OOP.

```python
class BaseClassifier:
    """Lớp cha cho mọi thuật toán phân lớp. Lớp con chỉ cần viết hàm build() trả về Pipeline."""
    name = "Base"

    def __init__(self):
        self.pipeline = None

    def build(self):                      # lớp con bắt buộc ghi đè
        raise NotImplementedError

    def fit(self, X, y):                  # Bước 1: xây dựng mô hình (Notes tr.2)
        self.pipeline = self.build()
        self.pipeline.fit(X, y)
        return self

    def predict(self, X):                 # gán nhãn cho quan sát mới
        return self.pipeline.predict(X)

    def scores(self, X):                  # điểm/xác suất lớp dương – dùng vẽ ROC (Notes tr.11)
        if hasattr(self.pipeline, "predict_proba"):
            return self.pipeline.predict_proba(X)[:, 1]
        return self.pipeline.decision_function(X)

    def accuracy(self, X, y):             # Bước 2: đánh giá bằng tập Test
        return accuracy_score(y, self.predict(X))

# ===================== PHẦN 2A. K-NN =====================
```

## PHẦN 2A. THUẬT TOÁN 1: K-NN (K-NEAREST NEIGHBORS) – LAZY LEARNING  *(Notes tr.3–4)*
**Ý chính:**
- Bắt chước *"ngưu tầm ngưu, mã tầm mã"*; **trì hoãn xây mô hình** đến khi có quan sát mới (Lazy ↔ Eager) (tr.3).
- **Phân lớp:** chọn lớp phổ biến (mode) trong k láng giềng. **Hồi quy:** lấy **trung bình** k láng giềng (tr.3).
- 2 vấn đề: **(1) Khoảng cách/độ tương đồng, (2) Láng giềng.** Dữ liệu số → **Euclid**; dữ liệu phân loại → **Hamming** (đếm giá trị khác nhau) (tr.3).
- Chọn k: theo thông lệ, hoặc theo tiêu chí đánh giá (vét cạn); nếu không biết chọn thì k ≈ √n (tr.3 – ký hiệu trong file bị mất, mình hiểu là căn bậc hai).
- **Hạn chế:** mọi láng giềng bình đẳng → biến thể **có trọng số theo khoảng cách** (gần: trọng số lớn) (tr.3).
- Cần chuẩn hoá dữ liệu vì dựa trên khoảng cách (nếu không, cột có số lớn như `area` lấn át cột nhỏ).

```python
def hamming_distance(a, b):
    """Hamming = tỉ lệ thuộc tính khác nhau (dữ liệu phân loại) – Notes tr.3."""
    a, b = list(a), list(b)
    return sum(x != y for x, y in zip(a, b)) / len(a)

# Ví dụ đúng như notes tr.3: (M, KHMT) và (M, KHDL): giới tính giống, ngành khác -> 1/2
print("Hamming((M,KHMT),(M,KHDL)) =", hamming_distance(("M", "KHMT"), ("M", "KHDL")))

class KNNClassifier(BaseClassifier):
    name = "K-NN"

    def __init__(self, k=5, metric="euclidean", weights="uniform"):
        super().__init__()
        self.k = k                  # số láng giềng
        self.metric = metric        # 'euclidean' (số) hoặc 'hamming' (phân loại)
        self.weights = weights      # 'uniform' = bình đẳng | 'distance' = gần thì trọng số lớn (khắc phục hạn chế)

    def build(self):
        return Pipeline([("scale", StandardScaler()),
                         ("clf", KNeighborsClassifier(n_neighbors=self.k, metric=self.metric,
                                                      weights=self.weights))])

    @staticmethod
    def sqrt_rule(n_train):
        """Quy tắc k ≈ √n khi không biết chọn k (Notes tr.3). Lấy số lẻ để tránh hoà phiếu."""
        k = int(round(np.sqrt(n_train)))
        return k if k % 2 == 1 else k + 1

    @staticmethod
    def find_best_k(X, y, k_values=range(1, 31, 2), cv=5):
        """Chọn k theo tiêu chí đánh giá bằng cross-validation (vét cạn – Notes tr.3, 11)."""
        res = {}
        for k in k_values:
            res[k] = cross_val_score(KNNClassifier(k).build(), X, y, cv=cv).mean()
        return pd.Series(res)

# ----- chạy thử -----
print("k theo quy tắc √n:", KNNClassifier.sqrt_rule(len(X_train)))
acc_k = KNNClassifier.find_best_k(X_train, y_train)
best_k = int(acc_k.idxmax())
print("Accuracy CV theo k:\n", acc_k.round(4).to_string())
print("=> k tốt nhất:", best_k)

knn = KNNClassifier(k=best_k).fit(X_train, y_train)
knn_w = KNNClassifier(k=best_k, weights="distance").fit(X_train, y_train)    # biến thể có trọng số
print(f"Accuracy Test – K-NN thường : {knn.accuracy(X_test, y_test):.4f}")
print(f"Accuracy Test – K-NN trọng số: {knn_w.accuracy(X_test, y_test):.4f}")

# K-NN hồi quy (lấy trung bình các láng giềng – Notes tr.3)
knn_reg = Pipeline([("s", StandardScaler()), ("r", KNeighborsRegressor(5))]).fit(Xr_tr, yr_tr)
print("K-NN hồi quy, dự đoán area_mean 3 mẫu đầu:", knn_reg.predict(Xr_te[:3]).round(1))
```

## 🎯 KHI NÀO DÙNG: **K-NN**
**DỮ LIỆU như thế nào thì chọn:**
- 📗 Features **số** → khoảng cách **Euclid** (chim bay, đường thẳng); features **phân loại** → **Hamming** (đếm giá trị khác nhau; 📊 `(M,KHMT)` vs `(M,KHDL)` = 1/2). Mã hoá nhãn thành số làm đổi tính chất bài toán → muốn giữ nguyên thì dùng Hamming.
- 📗 Cần mô hình **dễ giải thích** (k-NN, cây quyết định là 2 mô hình "dễ diễn giải" – tr.9).
- 📗 **Lazy learning**: không xây mô hình trước, chờ có quan sát mới mới tính (≠ Eager) → hợp khi dữ liệu **thay đổi liên tục**, không muốn huấn luyện lại. 📙 Dữ liệu **vừa/nhỏ**.
- 📗 Muốn chọn láng giềng theo **độ tương đồng** (VD: "giống ~50%") thay vì khoảng cách → **Nearest Neighbours (NN)** — phải qua bước **tiền xử lý** dữ liệu trước.
- 📙 **Bắt buộc chuẩn hoá** (cột `area` ~ hàng trăm lấn át cột ~ 0,1).

**Loại y → cách lấy kết quả (📗 tr.3):** y **nhãn** → chọn **lớp phổ biến (mode)** trong k láng giềng; y **số** → **trung bình** (average) k láng giềng.

**KẾT QUẢ như thế nào thì chọn / đổi:**
| Dấu hiệu | Kết luận |
|---|---|
| 📊 k=1 → CV **0,947** (thấp, nhạy nhiễu); k=7 → **0,969** (cao nhất); k=29 → **0,954** (giảm lại) | **k nhỏ quá = overfit; k lớn quá = underfit** → chọn k ở **vùng đỉnh** (hoặc điểm gãy ELBOW 📗 tr.16) |
| k theo thông lệ chưa chắc tốt | 📗 Cách 1: theo thông lệ; Cách 2: **theo tiêu chí đánh giá** (vét cạn thử nhiều k, dùng CV). Không biết chọn → k ≈ **√n** (📊 = 21; nhưng k=7 tốt hơn ⇒ **đừng tin mù quáng √n**) |
| Láng giềng gần/xa chênh nhiều mà vẫn tính "bình đẳng" | Dùng **K-NN có trọng số** (`weights="distance"`): gần ⇒ trọng số lớn 📗 tr.3. 📊 ở dữ liệu này 2 bản **bằng nhau (0,9561)** → không cần phức tạp hơn |
| 📙 Hoà phiếu giữa 2 lớp | Chọn **k lẻ** |

**KHÔNG nên dùng khi:** 📙 dữ liệu **rất lớn** (mỗi lần dự đoán phải tính khoảng cách tới mọi điểm) • **quá nhiều chiều** (khoảng cách mất ý nghĩa) • chưa chuẩn hoá được dữ liệu.
> ⚠️ Notes tr.3 ghi k-NN "sinh ra mô hình có số tham số cố định, biết trước" — câu này dễ **lệch** so với lý thuyết thông dụng (k-NN thường xếp **phi tham số**; `k` là **siêu tham số**). Nên **đối chiếu lại slide** trước khi thi.

## PHẦN 3. THUẬT TOÁN 2: CÂY QUYẾT ĐỊNH (DECISION TREE)  *(Notes tr.4–6)*
**Ý chính:**
- 2 loại node: **node thuộc tính (feature)** và **node lá (nhãn yes/no)**. Cách tiếp cận *tree induction* (tr.4).
- Xây cây theo **chia để trị** + **đệ quy từ trên xuống** (top-down) (tr.4). Mỗi node là 1 điều kiện **if-else** → biên quyết định "phẳng" (tr.4).
- Là phương pháp **Discriminative** (phân biệt), **phi tham số** (không biết trước số node) (tr.5).
- **Chọn thuộc tính tốt nhất** = chia ra các phân hoạch **đồng nhất nhất / hỗn tạp thấp nhất** (tr.5):
  - **Entropy** (Shannon 1948): `Entropy(D) = −Σ pᵢ·log₂ pᵢ`
  - **Information Gain** = Entropy(D) − Info(D,A) → chọn A có Gain **lớn nhất** → **ID3** (tr.5)
  - Hạn chế ID3: thiên vị thuộc tính chia nhiều nhóm → **C4.5** dùng **Gain Ratio = Gain / SplitInfo** (tr.6)
  - **Gini = 1 − Σ pᵢ²** (càng thấp càng đồng nhất) (tr.6)
- **Dữ liệu liên tục:** rời rạc hoá – cắt thành n đoạn, chọn mốc cho entropy nhỏ nhất (tr.6).
- **Cắt tỉa (pruning):** *"giải thích đơn giản thì đáng tin hơn giải thích phức tạp"* (tr.6).

```python
# ===== 3.1 TÍNH TAY Entropy / Gain / Gain Ratio / Gini (để thấy bản chất công thức) =====
class ImpurityCalculator:
    """Các độ đo 'độ hỗn tạp' dùng chọn thuộc tính tốt nhất (Notes tr.5–6)."""

    @staticmethod
    def entropy(y):                                  # Entropy(D) = -Σ p log2 p
        p = np.bincount(y) / len(y)
        p = p[p > 0]
        return -(p * np.log2(p)).sum()

    @staticmethod
    def gini(y):                                     # Gini = 1 - Σ p²
        p = np.bincount(y) / len(y)
        return 1 - (p ** 2).sum()

    @staticmethod
    def best_threshold(x, y, n_cuts=9):
        """Rời rạc hoá dữ liệu liên tục (Notes tr.6): thử n mốc cắt, chọn mốc cho entropy sau chia NHỎ nhất."""
        best_t, best_info = None, np.inf
        for q in np.linspace(0.1, 0.9, n_cuts):
            t = np.quantile(x, q)
            left, right = y[x <= t], y[x > t]
            if len(left) == 0 or len(right) == 0:
                continue
            info = (len(left) * ImpurityCalculator.entropy(left) +
                    len(right) * ImpurityCalculator.entropy(right)) / len(y)    # Info(D,A)
            if info < best_info:
                best_t, best_info = t, info
        return best_t

    @classmethod
    def evaluate_split(cls, x, y):
        """Trả về Gain, GainRatio, Gini(sau chia) cho 1 thuộc tính liên tục với mốc cắt tốt nhất."""
        t = cls.best_threshold(x, y)
        left, right = y[x <= t], y[x > t]
        w = np.array([len(left), len(right)]) / len(y)               # tỉ lệ 2 nhóm
        info = w[0] * cls.entropy(left) + w[1] * cls.entropy(right)  # Info(D,A)
        gain = cls.entropy(y) - info                                 # Information Gain (ID3)
        split_info = -(w * np.log2(w)).sum()                         # SplitInfo (phạt thuộc tính chia nhiều nhóm)
        gain_ratio = gain / split_info if split_info > 0 else 0      # Gain Ratio (C4.5)
        gini_after = w[0] * cls.gini(left) + w[1] * cls.gini(right)  # Gini sau chia (càng thấp càng tốt)
        return t, gain, gain_ratio, gini_after

    @classmethod
    def rank_features(cls, X, y, top=8):
        y = np.asarray(y)
        rows = []
        for col in X.columns:
            t, g, gr, gi = cls.evaluate_split(X[col].values, y)
            rows.append((col, t, g, gr, gi))
        out = pd.DataFrame(rows, columns=["feature", "mốc cắt", "Info Gain (ID3)", "Gain Ratio (C4.5)", "Gini sau chia"])
        return out.sort_values("Info Gain (ID3)", ascending=False).head(top)

print("Entropy toàn tập Train:", round(ImpurityCalculator.entropy(y_train.values), 4))
print("Gini toàn tập Train   :", round(ImpurityCalculator.gini(y_train.values), 4))
print("\nTop thuộc tính tốt nhất (Gain lớn nhất = nên chọn làm node gốc):")
print(ImpurityCalculator.rank_features(X_train, y_train).round(4).to_string(index=False))
```

```python
# ===== 3.2 CÂY QUYẾT ĐỊNH BẰNG SKLEARN (kế thừa BaseClassifier) =====
class DecisionTreeModel(BaseClassifier):
    name = "Decision Tree"

    def __init__(self, criterion="entropy", max_depth=None, ccp_alpha=0.0):
        super().__init__()
        self.criterion = criterion   # 'entropy' (tương đương ID3/C4.5 về độ đo) hoặc 'gini' (Notes tr.5–6)
        self.max_depth = max_depth   # giới hạn độ sâu = cắt tỉa trước (pre-pruning)
        self.ccp_alpha = ccp_alpha   # > 0 = cắt tỉa sau (post-pruning) (Notes tr.6)

    def build(self):
        return Pipeline([("clf", DecisionTreeClassifier(criterion=self.criterion, max_depth=self.max_depth,
                                                        ccp_alpha=self.ccp_alpha, random_state=RANDOM_STATE))])

    def show_rules(self, feature_names):
        """In cây dưới dạng if-else (Notes tr.4: mỗi node là 1 điều kiện if-else)."""
        return export_text(self.pipeline.named_steps["clf"], feature_names=list(feature_names), max_depth=3)

# So sánh Entropy – Gini – có/không cắt tỉa
for crit in ["entropy", "gini"]:
    for ccp in [0.0, 0.01]:
        m = DecisionTreeModel(crit, ccp_alpha=ccp).fit(X_train, y_train)
        n_leaves = m.pipeline.named_steps["clf"].get_n_leaves()
        print(f"criterion={crit:8s} ccp_alpha={ccp:<5} | số lá={n_leaves:3d} | Acc Train={m.accuracy(X_train, y_train):.3f} | Acc Test={m.accuracy(X_test, y_test):.3f}")

tree = DecisionTreeModel("entropy", ccp_alpha=0.01).fit(X_train, y_train)
print("\nLuật if-else (3 tầng đầu):\n", tree.show_rules(X_train.columns))
# Nhận xét: cây không cắt tỉa thường Train=1.0 (overfitting); cắt tỉa -> đơn giản hơn, Test thường tốt hơn/bằng.
```

## 🎯 KHI NÀO DÙNG: **CÂY QUYẾT ĐỊNH** (và chọn Entropy / Gain Ratio / Gini)
**DỮ LIỆU như thế nào thì chọn:**
- 📗 Cần **luật if–else đọc được** để giải thích cho người khác (📊 cây ở trên in ra được luật `perimeter_worst <= 114.45 → …`). Là mô hình **dễ diễn giải** (tr.9).
- 📗 Nhóm có thể cắt bằng các **biên "phẳng" song song trục** (*rectilinear decision boundary*): mỗi node = 1 điều kiện trên 1 feature (cây **đơn biến**).
- 📗 Là phương pháp **Discriminative** (phân lớp dựa vào **dấu hiệu khác biệt**; VD: câu có dấu → tiếng Việt; **không có dấu thì chưa kết luận được**) và **phi tham số** (số node không biết trước). 📙 Ngược lại **Generative** (như Naïve Bayes) mô hình hoá từng lớp.
- 📗 **Dữ liệu liên tục** → phải **rời rạc hoá**: cắt thành n đoạn, chọn **mốc cho entropy nhỏ nhất**. 📙 Không cần chuẩn hoá; chịu được trộn số + phân loại.

**Chọn TIÊU CHÍ tách node (📗 tr.5–6):**
| Tình huống | Chọn | Vì sao |
|---|---|---|
| Mặc định, đơn giản, hiệu quả | **ID3** – Entropy + Information Gain (chọn A có Gain **lớn nhất**) | Hỗn tạp thấp nhất sau khi chia |
| Có feature chia ra **rất nhiều nhóm** (kiểu mã số, ID, ngày) | **C4.5** – **Gain Ratio = Gain / SplitInfo** | ID3 **thiên vị** feature chia nhiều nhóm vì nhóm nhỏ thì dễ đồng nhất ("chơi không đẹp"); C4.5 chia thêm số nhóm để **cào bằng** |
| Số lớp **khá nhiều** *(notes ghi "khá…", hiểu là nhiều)* | **Gini = 1 − Σpᵢ²** (càng thấp càng đồng nhất) | Notes: hiệu quả khi nhiều lớp, nhưng độ chính xác có thể **không bằng** ID3/C4.5. 📙 Gini không tính log nên tính nhanh |

**KẾT QUẢ như thế nào thì làm gì:**
| Dấu hiệu | Kết luận / hành động |
|---|---|
| 📊 Acc **Train = 1,000** nhưng Test **0,956 / 0,930** | **Overfit** (cây thuộc lòng Train) → **cắt tỉa** (`ccp_alpha`, `max_depth`) 📗 tr.6: *"giải thích đơn giản đáng tin hơn giải thích phức tạp"* |
| 📊 entropy: ccp=0 → 17 lá, Test **0,956**; ccp=0,01 → 15 lá, Test **0,947** | Chênh **1 mẫu** nhưng cây gọn hơn ⇒ theo notes tr.9 có thể chọn cây **gọn** |
| 📊 gini + ccp=0,01 → chỉ **5 lá**, Test **0,912** (tụt) | Cắt tỉa **quá tay** → **underfit** → nới `ccp_alpha` |
| 📊 entropy (0,956) vs gini (0,930) | Ở dữ liệu này entropy nhỉnh hơn → đúng lời notes "Gini có thể kém chính xác hơn" — nhưng **luôn chạy cả hai rồi so** |

**KHÔNG nên dùng khi:** 📙 cần ranh giới **chéo/cong mượt** (cây phải cắt bậc thang → cần rất nhiều node) • dữ liệu ít nhưng cây sâu → **rất dễ overfit**.

## PHẦN 4. ÔN TẬP XÁC SUẤT  *(Notes tr.6–7, BUỔI 4)*
**Ý chính:** Phân phối đều liên tục (xác suất = diện tích hình chữ nhật) • Phân phối chuẩn → chuẩn hoá về **chuẩn tắc (mean=0, var=1)** để tra 1 bảng • **Pareto 80:20**, dùng cho survival analysis • **Luật số lớn** (n lớn → trung bình ≈ kỳ vọng) • **Định lý giới hạn trung tâm CLT** (tổng nhiều biến ngẫu nhiên → xấp xỉ chuẩn). Liên quan đến kiểm định giả thuyết ở phần tiền xử lý.

```python
class ProbabilityReview:
    """Các minh hoạ xác suất trong Notes tr.6–7."""
    rng = np.random.default_rng(RANDOM_STATE)

    @staticmethod
    def uniform_prob(a, b, x1, x2):
        """P(x1<X<x2) với X~U(a,b) = F(x2)-F(x1) = diện tích hình chữ nhật (Notes tr.6)."""
        d = stats.uniform(loc=a, scale=b - a)
        return d.cdf(x2) - d.cdf(x1)

    @staticmethod
    def normal_prob(mu, sigma, x1, x2):
        """Chuẩn hoá z=(x-mu)/sigma để dùng phân phối chuẩn tắc N(0,1) (Notes tr.6)."""
        z1, z2 = (x1 - mu) / sigma, (x2 - mu) / sigma
        return stats.norm.cdf(z2) - stats.norm.cdf(z1)

    @classmethod
    def pareto_80_20(cls, b=1.16, n=100000):
        """Pareto: ~20% giá trị lớn nhất chiếm ~80% tổng (Notes tr.7)."""
        data = np.sort(stats.pareto(b).rvs(n, random_state=RANDOM_STATE))[::-1]
        return data[: int(0.2 * n)].sum() / data.sum()

    @staticmethod
    def pareto_survival(b, t):
        """Survival: xác suất thời gian tồn tại >= t (bảo hành) (Notes tr.7)."""
        return stats.pareto(b).sf(t)

    @classmethod
    def law_of_large_numbers(cls, n=5000):
        """LLN: tung xúc xắc n lần, trung bình mẫu -> kỳ vọng 3.5 (Notes tr.7)."""
        rolls = cls.rng.integers(1, 7, n)
        running_mean = np.cumsum(rolls) / np.arange(1, n + 1)
        plt.plot(running_mean); plt.axhline(3.5, color="r", ls="--", label="kỳ vọng 3.5")
        plt.title("Luật số lớn"); plt.legend(); plt.show()
        return running_mean[-1]

    @classmethod
    def central_limit_theorem(cls, n_terms=30, n_samples=5000):
        """CLT: tổng của nhiều U(0,1) -> chuẩn hoá -> xấp xỉ N(0,1) (Notes tr.7)."""
        sums = cls.rng.random((n_samples, n_terms)).sum(axis=1)
        mu, sd = n_terms * 0.5, np.sqrt(n_terms / 12)      # kỳ vọng & độ lệch chuẩn của tổng
        z = (sums - mu) / sd                               # "ép về chuẩn" bằng cách đổi ngưỡng (Notes tr.7)
        plt.hist(z, bins=40, density=True, alpha=0.6)
        xs = np.linspace(-4, 4, 200); plt.plot(xs, stats.norm.pdf(xs), "r")
        plt.title("CLT: tổng chuẩn hoá ≈ N(0,1)"); plt.show()
        return z.mean(), z.std()

print("P(2.0<X<2.5), X~U(1,3)        =", ProbabilityReview.uniform_prob(1, 3, 2.0, 2.5), "(notes: 0.25)")
print("P(-1<Z<1), chuẩn tắc          =", round(ProbabilityReview.normal_prob(0, 1, -1, 1), 4))
print("Pareto: 20% lớn nhất chiếm    =", round(ProbabilityReview.pareto_80_20(), 3), "tổng")
print("Pareto: P(T>=2) với b=1.16    =", round(ProbabilityReview.pareto_survival(1.16, 2), 4))
print("LLN: trung bình sau 5000 lần  =", round(ProbabilityReview.law_of_large_numbers(), 3))
print("CLT: (mean, std) sau chuẩn hoá=", tuple(round(v, 3) for v in ProbabilityReview.central_limit_theorem()))
```

## PHẦN 5. THUẬT TOÁN 3: NAÏVE BAYES  *(Notes tr.7–8, BUỔI 5)*
**Ý chính:**
- Công thức Bayes: `P(A|B) = P(A)·P(B|A) / P(B)`. Mục tiêu: với quan sát X, chọn lớp **C\* = argmaxᵢ P(Cᵢ|X)** – xác suất **hậu nghiệm** (tr.7).
- **"Ngây thơ"** = giả định các thuộc tính **độc lập** với nhau (kiểm tra độc lập: 2 biến số → Pearson; phân loại vs số → T-test/ANOVA) (tr.8).
- **Hiệu chỉnh Laplace** (Laplace smoothing): cộng thêm 1 chút để xác suất không bao giờ = 0 (tr.8) → tham số `alpha=1`.
- **3 mô hình** (tr.8) – *"áp dụng cả 3 rồi so sánh sai số, chọn tốt nhất"*:
  - **Gaussian NB:** biến liên tục, phân phối chuẩn.
  - **Multinomial NB:** features rời rạc/không âm, chú ý Laplace, chuẩn hoá bằng **Min-Max (KHÔNG dùng StandardScaler** vì sinh số âm).
  - **Bernoulli NB:** như Multinomial nhưng biến **nhị phân** (0/1).

```python
class NaiveBayesModel(BaseClassifier):
    """3 biến thể Naïve Bayes (Notes tr.8)."""

    def __init__(self, variant="gaussian", alpha=1.0):
        super().__init__()
        self.variant = variant      # 'gaussian' | 'multinomial' | 'bernoulli'
        self.alpha = alpha          # alpha=1 -> hiệu chỉnh Laplace (tránh xác suất = 0)
        self.name = f"NB-{variant}"

    def build(self):
        if self.variant == "gaussian":      # liên tục, giả định chuẩn -> không cần scale
            return Pipeline([("clf", GaussianNB())])
        if self.variant == "multinomial":   # cần dữ liệu KHÔNG ÂM -> Min-Max (không dùng StandardScaler!)
            return Pipeline([("scale", MinMaxScaler()), ("clf", MultinomialNB(alpha=self.alpha))])
        if self.variant == "bernoulli":     # biến nhị phân: Min-Max rồi nhị phân hoá tại 0.5
            return Pipeline([("scale", MinMaxScaler()), ("clf", BernoulliNB(alpha=self.alpha, binarize=0.5))])
        raise ValueError("variant phải là gaussian / multinomial / bernoulli")

# Áp dụng cả 3, so sánh sai số (đúng yêu cầu notes tr.8)
nb_models = {}
for v in ["gaussian", "multinomial", "bernoulli"]:
    nb_models[v] = NaiveBayesModel(v).fit(X_train, y_train)
    print(f"{v:12s} Acc Train={nb_models[v].accuracy(X_train, y_train):.4f} | Acc Test={nb_models[v].accuracy(X_test, y_test):.4f}")

# Hiệu ứng Laplace: alpha≈0 ~ không hiệu chỉnh, alpha=1 = Laplace. (Dữ liệu này liên tục, ít ô = 0 nên 2 kết quả gần như bằng nhau;
# Laplace phát huy khi có đếm = 0, ví dụ dữ liệu văn bản/đếm từ.)
for a in [1e-10, 1.0]:
    m = NaiveBayesModel("multinomial", alpha=a).fit(X_train, y_train)
    print(f"Multinomial alpha={a:g}: Acc Test={m.accuracy(X_test, y_test):.4f}")
```

## PHẦN 6. THUẬT TOÁN 4: SUPPORT VECTOR MACHINE (SVM)  *(Notes tr.8–9)*
**6.1 SVM tuyến tính (tr.8):** tìm **siêu phẳng (hyperplane)** phân cách các lớp sao cho **biên (margin) lớn nhất** – đường tốt nhất là đường mà khoảng cách đến 2 nhóm **lớn nhất và xấp xỉ bằng nhau**. Chỉ các **support vector** (điểm sát biên) quyết định đường; nhiễu xa biên thì không ảnh hưởng.
- **Hard margin:** không cho điểm nào lấn biên → nhạy với outliers.
- **Soft margin (tr.9):** chấp nhận một số điểm lấn biên (margin violation) để tránh overfitting; dùng **siêu tham số `C`** + biến lỏng (slack) ξ để "trừng phạt" điểm lấn biên (ξ>1: nhảy sang phía sai; 0<ξ<1: còn trong vùng biên). **C lớn** → phạt nặng → gần hard margin; **C nhỏ** → biên rộng, dễ tính.

**6.2 SVM phi tuyến (tr.9):** dùng **Kernel Trick** (ánh xạ lên không gian nhiều chiều để tách tuyến tính) – các hàm: **Linear, Polynomial, RBF/Gaussian**; *"cứ chạy hết và xem hàm nào cho mô hình tốt nhất"*. Bậc đa thức nhỏ → kém với dữ liệu phức tạp; bậc lớn → chậm.

```python
class SVMModel(BaseClassifier):
    """SVM phân lớp với nhiều kernel (Notes tr.8–9). Cần chuẩn hoá vì SVM dựa trên khoảng cách."""

    def __init__(self, kernel="linear", C=1.0, degree=3, gamma="scale"):
        super().__init__()
        self.kernel = kernel   # 'linear' | 'poly' | 'rbf' (rbf = Gaussian kernel)
        self.C = C             # siêu tham số soft margin
        self.degree = degree   # bậc đa thức (chỉ dùng cho poly)
        self.gamma = gamma
        self.name = f"SVM-{kernel}"

    def build(self):
        return Pipeline([("scale", StandardScaler()),
                         ("clf", SVC(kernel=self.kernel, C=self.C, degree=self.degree,
                                     gamma=self.gamma, probability=True, random_state=RANDOM_STATE))])

    def n_support_vectors(self):
        return int(self.pipeline.named_steps["clf"].n_support_.sum())

# ---- 6.2 Chạy hết các kernel, xem kernel nào tốt (Notes tr.9) ----
svm_models = {}
for k in ["linear", "poly", "rbf"]:
    svm_models[k] = SVMModel(kernel=k).fit(X_train, y_train)
    print(f"kernel={k:6s} | số support vector={svm_models[k].n_support_vectors():3d} | Acc Test={svm_models[k].accuracy(X_test, y_test):.4f}")

# ---- 6.1 Ảnh hưởng của C (soft margin) ----
print("\nẢnh hưởng siêu tham số C (kernel linear):")
for C in [0.01, 1, 100]:
    m = SVMModel("linear", C=C).fit(X_train, y_train)
    print(f"C={C:<6} | support vector={m.n_support_vectors():3d} | Acc Train={m.accuracy(X_train, y_train):.4f} | Acc Test={m.accuracy(X_test, y_test):.4f}")
# C nhỏ -> biên rộng, nhiều support vector, cho phép lấn biên; C lớn -> biên hẹp, gần hard margin
```

**DỮ LIỆU như thế nào thì chọn:**
- 📗 **2 nhóm tách được bằng siêu phẳng** → **SVM tuyến tính**. Đường tốt nhất = đường có **khoảng cách tới 2 nhóm lớn nhất và xấp xỉ bằng nhau** (margin cực đại). Điểm có f(x)=0 nằm trên đường; dấu +/− cho biết nằm **phía nào**.
- 📗 Nhóm **lẫn / cong** → **phi tuyến**: bổ sung feature (VD vector (x, x²)) hoặc **Kernel Trick** (ánh xạ lên chiều cao hơn để thành tách tuyến tính, VD siêu phẳng ↔ hình tròn).
- 📗 **Bậc đa thức nhỏ** → kém với dữ liệu phức tạp; **bậc lớn** → mô hình **chậm**. → *"Cứ chạy hết Linear / Polynomial / RBF(Gaussian) và xem hàm nào tốt nhất."*

- 📙 Cần **chuẩn hoá** (dựa trên khoảng cách). Hợp khi **nhiều chiều**, dữ liệu **vừa**.

**Hard hay Soft margin?** 📗 Có **outlier** / nhóm chồng lấn → **Soft margin** (cho phép một số điểm lấn biên – *margin violation*; biến lỏng ξ>1: nhảy sang phía sai, 0<ξ<1: còn trong vùng biên). **Hard margin** chỉ dùng khi dữ liệu **sạch, tách hoàn toàn** (nhạy outlier).

RBF hơn linear chỉ ~**1 mẫu** ⇒ theo 📗 tr.9 có thể chọn **linear (đơn giản hơn)**; Poly bậc 3 vừa **chậm vừa kém** ⇒ loại |
| 📊 `C=100`: Train **0,9978** nhưng Test **0,9298** (tụt) | **Overfit** (phạt lấn biên quá nặng ≈ hard margin) → **giảm C** |

**KHÔNG nên dùng khi:** muốn mô hình **diễn giải được** (SVM kernel là "hộp đen" hơn cây/k-NN); dữ liệu **rất lớn** (huấn luyện chậm).

## PHẦN 7. ĐÁNH GIÁ MÔ HÌNH PHÂN LỚP  *(Notes tr.9–11, BUỔI 5 – ĐÁNH GIÁ GIẢI PHÁP)*
**Tiêu chí chọn giải pháp (tr.9):** (1) hiệu quả–chi phí, (2) độ chính xác (sai số), (3) độ nhạy với nhiễu, (4) khả năng diễn giải (dễ: k-NN, cây quyết định). Khi độ chính xác chênh không nhiều → cân nhắc độ phức tạp.

**Ma trận nhầm lẫn (tr.10)** – hàng = thực tế, cột = dự đoán:
| | Dự đoán + | Dự đoán − |
|---|---|---|
| **Thực tế +** | TP (đúng) | **FN – Sai lầm loại II (silence)** |
| **Thực tế −** | **FP – Sai lầm loại I (noise)** | TN (đúng) |

`accuracy = (TP+TN)/n` • `error = (FP+FN)/n` • `precision = TP/(TP+FP)` • `recall = TP/(TP+FN)`
**Precision & Recall nghịch biến** → dùng **trung bình điều hoà**: `F = precision·recall / ((1−α)·precision + α·recall)` (α=0.5 → F1).
**ROC (tr.11):** điểm hoàn hảo (FPR=0, TPR=1); đường/diện tích **AUC** càng lớn càng tốt; chọn thuật toán gần điểm hoàn hảo nhất.

```python
class ClassificationEvaluator:
    """Tính mọi độ đo đánh giá phân lớp (Notes tr.10–11). Lớp dương (+) = Malignant = 1."""

    @staticmethod
    def confusion(y_true, y_pred):
        # labels=[1,0] để đúng bố cục notes: hàng/cột đầu = POSITIVE -> [[TP, FN],[FP, TN]]
        cm = confusion_matrix(y_true, y_pred, labels=[1, 0])
        return pd.DataFrame(cm, index=["Thực tế +", "Thực tế -"], columns=["Dự đoán +", "Dự đoán -"])

    @staticmethod
    def f_measure(precision, recall, alpha=0.5):
        """F = P*R / ((1-α)P + αR) – đúng công thức notes tr.10 (α=0.5 là F1)."""
        denom = (1 - alpha) * precision + alpha * recall
        return precision * recall / denom if denom > 0 else 0.0

    @classmethod
    def summary(cls, y_true, y_pred):
        tp, fn = cls.confusion(y_true, y_pred).iloc[0]
        fp, tn = cls.confusion(y_true, y_pred).iloc[1]
        n = tp + tn + fp + fn
        precision = tp / (tp + fp) if tp + fp else 0.0
        recall = tp / (tp + fn) if tp + fn else 0.0
        return {"TP": tp, "FN": fn, "FP": fp, "TN": tn,
                "accuracy": (tp + tn) / n, "error": (fp + fn) / n,
                "precision": precision, "recall": recall,
                "F1": cls.f_measure(precision, recall, 0.5)}

    @staticmethod
    def roc_auc(y_true, scores):
        fpr, tpr, _ = roc_curve(y_true, scores)
        return fpr, tpr, auc(fpr, tpr)

    @staticmethod
    def dist_to_perfect(fpr, tpr):
        """Khoảng cách nhỏ nhất tới điểm hoàn hảo (0,1) (Notes tr.11)."""
        return np.sqrt(fpr ** 2 + (1 - tpr) ** 2).min()

# ----- chạy thử cho 1 mô hình -----
y_pred = knn.predict(X_test)
print("Ma trận nhầm lẫn K-NN:\n", ClassificationEvaluator.confusion(y_test, y_pred))
s = ClassificationEvaluator.summary(y_test, y_pred)
print({k: (round(v, 4) if isinstance(v, float) else int(v)) for k, v in s.items()})
# Đối chiếu với sklearn để chắc chắn đúng:
print("Đối chiếu sklearn  precision/recall/F1:", round(precision_score(y_test, y_pred), 4),
      round(recall_score(y_test, y_pred), 4), round(f1_score(y_test, y_pred), 4))
print("Với F-measure α=0.5 tự viết:", round(ClassificationEvaluator.f_measure(s["precision"], s["recall"]), 4))
```

```python
# ===== SO SÁNH TẤT CẢ THUẬT TOÁN PHÂN LỚP + VẼ ROC (Notes tr.9, 11) =====
class ModelComparison:
    """Huấn luyện nhiều mô hình, so sánh Accuracy/Precision/Recall/F1/AUC và vẽ ROC."""

    def __init__(self, models):
        self.models = models            # dict {tên: đối tượng BaseClassifier}

    def run(self, X_tr, y_tr, X_te, y_te, plot=True):
        rows = []
        if plot:
            plt.figure(figsize=(6, 5))
        for name, m in self.models.items():
            m.fit(X_tr, y_tr)
            pred = m.predict(X_te)
            s = ClassificationEvaluator.summary(y_te, pred)
            fpr, tpr, a = ClassificationEvaluator.roc_auc(y_te, m.scores(X_te))
            rows.append((name, s["accuracy"], s["precision"], s["recall"], s["F1"], a,
                         ClassificationEvaluator.dist_to_perfect(fpr, tpr)))
            if plot:
                plt.plot(fpr, tpr, label=f"{name} (AUC={a:.3f})")
        if plot:
            plt.plot([0, 1], [0, 1], "k--"); plt.scatter([0], [1], c="r", marker="*", s=120, label="điểm hoàn hảo")
            plt.xlabel("FPR (1 - specificity)"); plt.ylabel("TPR (recall)"); plt.title("Đường cong ROC")
            plt.legend(fontsize=7); plt.show()
        return pd.DataFrame(rows, columns=["model", "accuracy", "precision", "recall", "F1", "AUC",
                                           "khoảng cách tới điểm hoàn hảo"]).set_index("model")

all_models = {
    "K-NN":          KNNClassifier(k=best_k),
    "Cây (entropy)": DecisionTreeModel("entropy", ccp_alpha=0.01),
    "Cây (gini)":    DecisionTreeModel("gini", ccp_alpha=0.01),
    "NB Gaussian":   NaiveBayesModel("gaussian"),
    "NB Multinomial": NaiveBayesModel("multinomial"),
    "NB Bernoulli":  NaiveBayesModel("bernoulli"),
    "SVM linear":    SVMModel("linear"),
    "SVM poly":      SVMModel("poly"),
    "SVM rbf":       SVMModel("rbf"),
}
table = ModelComparison(all_models).run(X_train, y_train, X_test, y_test)
print(table.round(4).sort_values("AUC", ascending=False))
# Chọn mô hình: AUC lớn + gần điểm hoàn hảo; nếu chênh ít thì chọn mô hình ĐƠN GIẢN hơn (Notes tr.9)
```

## 🎯 KHI NÀO DÙNG: **TỪNG ĐỘ ĐO ĐÁNH GIÁ PHÂN LỚP**
**Nguyên tắc gốc (📗 tr.10):** *Accuracy không phân biệt* **silence (FN – sai lầm loại II)** *và* **noise (FP – sai lầm loại I)** — mà 2 loại sai này **nguy hiểm khác nhau** → phải nhìn thêm Precision / Recall.

| Độ đo | Dùng khi | Dấu hiệu "đừng tin" | Nguồn |
|---|---|---|---|
| **Accuracy** (TP+TN)/n | 2 loại sai **nguy hiểm như nhau** và lớp **cân bằng** | Lớp lệch → đoán toàn lớp đa số vẫn "cao" (📊 đoán toàn Benign đã ≈ **62,7%**) | 📗 tr.10 · 📙 |
| **Recall** TP/(TP+FN) | **Bỏ sót nguy hiểm** (ung thư, gian lận) | 📊 NB Multinomial: Precision **1,00** nhưng Recall **0,52** ⇒ bỏ sót **gần nửa** ca ác tính → **không chọn** dù precision đẹp | 📗 tr.10 |
| **Precision** TP/(TP+FP) | **Báo nhầm tốn kém** (lọc spam, cảnh báo giả) | Precision cao mà Recall thấp = mô hình "nhát" | 📗 tr.10 |
| **F-measure** | Cần **cân bằng** P và R (2 chỉ số **nghịch biến**) | — | 📗 tr.10 |
| **ROC / AUC** | So thuật toán **không phụ thuộc ngưỡng** | 2 đường ROC **cắt nhau** ⇒ dùng **AUC** (diện tích) | 📗 tr.11 |

📗 **Vì sao dùng trung bình điều hoà cho F:** Precision và Recall **cùng tử, khác mẫu** → trung bình cộng chỉ hợp khi **cùng mẫu số**.
📗 **Đọc α trong công thức notes** `F = P·R / ((1−α)·P + α·R) = 1 / (α/P + (1−α)/R)`: **α lớn → F nghiêng về Precision**; **α nhỏ → nghiêng về Recall**; α = 0,5 → **F1**. *(Bài ung thư: muốn nhấn Recall ⇒ chọn α < 0,5.)*

**ROC (📗 tr.11):** điểm hoàn hảo = **(FPR=0, TPR=1)**; chọn thuật toán có đường **gần điểm hoàn hảo nhất / diện tích lớn hơn**. Ngưỡng quyết định là P(y=1|x).
📊 Cẩn thận: SVM poly có AUC **0,9967** và "khoảng cách tới điểm hoàn hảo" nhỏ nhất, nhưng Accuracy/Recall lại **thấp nhất** → ROC/AUC xét **mọi ngưỡng**, còn Accuracy/Recall xét **1 ngưỡng (0,5)** ⇒ hai bên có thể mâu thuẫn → **chọn theo mục tiêu bài toán** (bỏ sót nguy hiểm ⇒ ưu tiên Recall).

**Quy trình chọn giải pháp (📗 tr.9):** tiêu chí = (1) chi phí (thời gian, bộ nhớ) • (2) **độ chính xác** • (3) **độ nhạy với dữ liệu nhiễu** (nhiễu đúng điểm yếu của mô hình là "rất sợ") • (4) **khả năng diễn giải**. **Chênh độ chính xác ít → cân nhắc độ phức tạp / diễn giải.** So sánh **khác nhóm** (parametric ↔ non-parametric) hoặc **cùng nhóm** (so parameters / hyperparameters).
📗 Với thuật toán **cổ điển** (sắp xếp…) sai số = **0** nên chỉ so độ phức tạp; thuật toán **máy học** sai số **> 0** vì Train chỉ là **mẫu**, không phải toàn bộ quần thể.

## PHẦN 8. CHỌN MÔ HÌNH & CROSS-VALIDATION  *(Notes tr.9–11)*
**Dữ liệu thực nghiệm (tr.9–10):** **Training set** xây mô hình ứng viên • **Validation/development set** (tùy chọn) chọn mô hình & tinh chỉnh **hyperparameters** (khác *parameters*) • **Test set** đánh giá mô hình cuối.
**Các phương pháp (tr.11):**
- **Hold-out:** chia ngẫu nhiên, thường 80–20.
- **k-Fold CV:** chia k phần, lần lượt mỗi phần làm test, **cộng/trung bình sai số** các lần. *Ghi chú notes: "đưa vào một class: chia tập thành k, có nhiều thuật toán phân lớp, có method đánh giá độ chính xác – mọi chức năng tổ chức thành 1 class".*
- **Grid Search:** vét cạn mọi tổ hợp siêu tham số (như chọn k của k-NN), chọn sai số thấp nhất.
- **Random Search:** chọn ngẫu nhiên giá trị siêu tham số (cho phép giá trị liên tục).
- **Hand-tuning:** chọn theo kinh nghiệm / giá trị thông lệ.

```python
class CrossValidator:
    """Gom MỌI chức năng đánh giá vào 1 class (đúng ghi chú notes tr.11):
    chia tập (hold-out, k-fold) + nhiều thuật toán phân lớp + method đo độ chính xác."""

    def __init__(self, models, k=5):
        self.models = models      # dict {tên: BaseClassifier}
        self.k = k                # số fold

    def hold_out(self, X, y, test_size=0.2):
        """Hold-out 80-20 (Notes tr.11)."""
        X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=test_size, stratify=y, random_state=RANDOM_STATE)
        return {n: m.fit(X_tr, y_tr).accuracy(X_te, y_te) for n, m in self.models.items()}

    def k_fold(self, X, y):
        """k-Fold: mỗi fold lần lượt làm test, tính sai số rồi trung bình (Notes tr.11)."""
        skf = StratifiedKFold(n_splits=self.k, shuffle=True, random_state=RANDOM_STATE)
        result = {}
        for name, m in self.models.items():
            fold_acc = []
            for tr_idx, te_idx in skf.split(X, y):
                m.fit(X.iloc[tr_idx], y.iloc[tr_idx])                      # huấn luyện trên k-1 fold
                fold_acc.append(m.accuracy(X.iloc[te_idx], y.iloc[te_idx]))  # kiểm tra trên fold còn lại
            result[name] = (np.mean(fold_acc), np.std(fold_acc))
        return pd.DataFrame(result, index=["accuracy TB", "độ lệch chuẩn"]).T

cv = CrossValidator({"K-NN": KNNClassifier(best_k), "Cây": DecisionTreeModel("entropy", ccp_alpha=0.01),
                     "NB Gaussian": NaiveBayesModel("gaussian"), "SVM rbf": SVMModel("rbf")}, k=5)
print("Hold-out 80-20:", {n: round(a, 4) for n, a in cv.hold_out(X, y).items()})
print("\n5-Fold CV:\n", cv.k_fold(X, y).round(4))
```

```python
class HyperparameterTuner:
    """Tinh chỉnh siêu tham số (hyperparameters) bằng Grid / Random / Hand-tuning (Notes tr.11)."""

    def __init__(self, X, y, cv=5):
        self.X, self.y, self.cv = X, y, cv
        self.pipe = KNNClassifier().build()        # ví dụ tinh chỉnh K-NN (như notes nhắc)

    def grid_search(self):
        """Vét cạn mọi tổ hợp, chọn tổ hợp sai số thấp nhất."""
        grid = {"clf__n_neighbors": list(range(1, 26, 2)), "clf__weights": ["uniform", "distance"]}
        gs = GridSearchCV(self.pipe, grid, cv=self.cv).fit(self.X, self.y)
        return gs.best_params_, gs.best_score_

    def random_search(self, n_iter=15):
        """Chọn ngẫu nhiên giá trị siêu tham số (có thể liên tục) – ở đây thử tham số C (liên tục) của SVM."""
        from scipy.stats import loguniform
        svm_pipe = SVMModel("rbf").build()
        dist = {"clf__C": loguniform(1e-2, 1e2), "clf__gamma": loguniform(1e-4, 1e0)}
        rs = RandomizedSearchCV(svm_pipe, dist, n_iter=n_iter, cv=self.cv, random_state=RANDOM_STATE).fit(self.X, self.y)
        return rs.best_params_, rs.best_score_

    def hand_tuning(self):
        """Chọn theo kinh nghiệm/thông lệ (k=5, hoặc k≈√n) rồi đo CV (Notes tr.11)."""
        res = {}
        for k in [5, KNNClassifier.sqrt_rule(len(self.X))]:
            res[f"k={k}"] = cross_val_score(KNNClassifier(k).build(), self.X, self.y, cv=self.cv).mean()
        return res

tuner = HyperparameterTuner(X_train, y_train)
print("Grid Search  (K-NN):", tuner.grid_search())
print("Random Search(SVM) :", tuner.random_search())
print("Hand-tuning  (K-NN):", {k: round(v, 4) for k, v in tuner.hand_tuning().items()})

# ---- Train / Validation / Test (Notes tr.9–10): chọn k trên VALIDATION, đánh giá cuối trên TEST ----
Xa, Xv, Xt, ya, yv, yt = dm.split_train_val_test()
val_acc = {k: KNNClassifier(k).fit(Xa, ya).accuracy(Xv, yv) for k in range(1, 22, 2)}
k_val = max(val_acc, key=val_acc.get)
print(f"\nk chọn trên validation = {k_val}; Acc Test cuối cùng = {KNNClassifier(k_val).fit(Xa, ya).accuracy(Xt, yt):.4f}")
```

## PHẦN 9. HỌC KHÔNG GIÁM SÁT – GOM CỤM (CLUSTERING) & CÁCH TIẾP CẬN PHÂN HOẠCH  *(Notes tr.11–13, BUỔI 6)*
**Ý chính chung (tr.11–12):**
- Gom cụm **không cần training set**, **không dùng nhãn**; khó gán nhãn chủ đề cho cụm. Học giám sát = *dự đoán* cụm nào; không giám sát = *mô tả* cụm có ý nghĩa gì (tìm trọng tâm, mean, median, mode…).
- Chất lượng cụm: **cohesion** (kết dính trong cụm) và **distinction** (khác biệt giữa các cụm).
- Đo khoảng cách giữa 2 cụm (tr.13): **Single-link** (ngắn nhất), **Complete-link** (dài nhất), **Average-link**, **Centroids**, **Medoids**, **Ward**.
- **Outlier có 3 loại (tr.12):** khác biệt toàn thể; khác biệt so với lân cận; vô ích (phương sai = 0).

**Nhóm thuật toán phân hoạch (tr.13):**
- **K-means:** dữ liệu **số**, tập **lồi**, phải **biết k**. 2 thao tác lặp lại: **(1) tính trọng tâm, (2) "tô màu"** (gán mỗi điểm vào tâm gần nhất).
- **K-modes:** dữ liệu **phân loại**, dùng **Hamming**. **K-prototypes:** dữ liệu hỗn hợp (số + phân loại). **K-medoids (PAM):** tâm là **điểm thật** có tổng khoảng cách tới các điểm khác ngắn nhất (bền với outlier).
- **Fuzzy C-means:** 1 đối tượng thuộc **nhiều cụm** theo **mức độ tham gia** wᵢⱼ ∈ [0,1]; **tổng mỗi dòng = 1**, tổng mỗi cột > 0. Lặp: tính trọng tâm → cập nhật membership → **dừng khi ‖W(t) − W(t−1)‖_Frobenius nhỏ**.

```python
# ===== DỮ LIỆU CHO GOM CỤM: bỏ nhãn, chuẩn hoá (nhãn y chỉ dùng để đánh giá ngoài ở Phần 13) =====
X_clu = StandardScaler().fit_transform(dm.X)       # chuẩn hoá vì dựa trên khoảng cách
y_true_clu = dm.y.values
X_2d = PCA(n_components=2, random_state=RANDOM_STATE).fit_transform(X_clu)   # chỉ để VẼ HÌNH 2D

class BaseClusterer:
    """Lớp cha cho mọi thuật toán gom cụm: fit_predict(X) -> nhãn cụm."""
    name = "Base"
    def __init__(self): self.labels_ = None
    def fit_predict(self, X): raise NotImplementedError
    def plot(self, labels=None, title=None):
        labels = self.labels_ if labels is None else labels
        plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels, cmap="viridis", s=12)
        plt.title(title or self.name); plt.xlabel("PC1"); plt.ylabel("PC2"); plt.show()

class KMeansModel(BaseClusterer):
    """K-means: (1) tính trọng tâm  (2) gán điểm vào tâm gần nhất, lặp đến hội tụ (Notes tr.13)."""
    def __init__(self, k=2):
        super().__init__(); self.k = k; self.name = f"K-means (k={k})"
    def fit_predict(self, X):
        self.model = KMeans(n_clusters=self.k, n_init=10, random_state=RANDOM_STATE).fit(X)
        self.labels_ = self.model.labels_
        self.inertia_ = self.model.inertia_      # tổng bình phương khoảng cách tới tâm (SSE) – dùng ELBOW
        return self.labels_

class KMedoidsModel(BaseClusterer):
    """K-medoids (PAM đơn giản): tâm cụm là ĐIỂM THẬT có tổng khoảng cách tới các điểm trong cụm nhỏ nhất (Notes tr.13)."""
    def __init__(self, k=2, max_iter=20):
        super().__init__(); self.k, self.max_iter = k, max_iter; self.name = f"K-medoids (k={k})"
    def fit_predict(self, X):
        D = cdist(X, X)                                              # ma trận khoảng cách (đối xứng, không âm – tr.13)
        rng = np.random.default_rng(RANDOM_STATE)
        medoids = rng.choice(len(X), self.k, replace=False)          # chọn ngẫu nhiên k điểm thật làm tâm
        for _ in range(self.max_iter):
            labels = D[:, medoids].argmin(axis=1)                    # gán điểm vào medoid gần nhất
            new = medoids.copy()
            for j in range(self.k):
                members = np.where(labels == j)[0]
                if len(members):                                     # medoid mới = điểm có tổng khoảng cách nhỏ nhất
                    new[j] = members[D[np.ix_(members, members)].sum(axis=1).argmin()]
            if np.array_equal(new, medoids): break
            medoids = new
        self.medoids_ = medoids
        self.labels_ = D[:, medoids].argmin(axis=1)
        return self.labels_

class KModesModel(BaseClusterer):
    """K-modes cho dữ liệu PHÂN LOẠI: khoảng cách Hamming, tâm cụm = giá trị phổ biến (mode) (Notes tr.13).
    (K-prototypes = kết hợp K-means + K-modes cho dữ liệu hỗn hợp; cần thư viện 'kmodes', không viết ở đây.)"""
    def __init__(self, k=2, max_iter=20):
        super().__init__(); self.k, self.max_iter = k, max_iter; self.name = f"K-modes (k={k})"
    def fit_predict(self, Xcat):
        Xcat = np.asarray(Xcat)
        rng = np.random.default_rng(RANDOM_STATE)
        modes = Xcat[rng.choice(len(Xcat), self.k, replace=False)]
        for _ in range(self.max_iter):
            dist = np.array([(Xcat != m).mean(axis=1) for m in modes]).T      # Hamming = tỉ lệ thuộc tính khác nhau
            labels = dist.argmin(axis=1)
            new = modes.copy()
            for j in range(self.k):
                if (labels == j).any():
                    new[j] = stats.mode(Xcat[labels == j], axis=0).mode        # mode theo từng cột
            if np.array_equal(new, modes): break
            modes = new
        self.modes_, self.labels_ = modes, labels
        return labels

class FuzzyCMeans(BaseClusterer):
    """Fuzzy C-means (Notes tr.13–14). W[i,j] = mức độ điểm i thuộc cụm j, tổng mỗi dòng = 1."""
    def __init__(self, c=2, m=2.0, tol=1e-4, max_iter=100):
        super().__init__(); self.c, self.m, self.tol, self.max_iter = c, m, tol, max_iter
        self.name = f"Fuzzy C-means (c={c})"
    def fit_predict(self, X):
        rng = np.random.default_rng(RANDOM_STATE)
        W = rng.random((len(X), self.c)); W /= W.sum(axis=1, keepdims=True)      # khởi tạo, ép tổng dòng = 1
        for it in range(self.max_iter):
            Wm = W ** self.m
            centers = (Wm.T @ X) / Wm.sum(axis=0)[:, None]                       # Bước 1: tính trọng tâm (có trọng số w^m)
            d = cdist(X, centers) + 1e-10
            inv = d ** (-2 / (self.m - 1))
            W_new = inv / inv.sum(axis=1, keepdims=True)                         # Bước 2: cập nhật membership ("tô màu")
            if np.linalg.norm(W_new - W, "fro") < self.tol:                      # Bước 3: dừng khi ‖W(t)-W(t-1)‖_F nhỏ
                W = W_new; break
            W = W_new
        self.W_, self.centers_, self.n_iter_ = W, centers, it + 1
        self.labels_ = W.argmax(axis=1)                                          # gán cứng = cụm có membership lớn nhất
        return self.labels_

# ----- chạy thử -----
km = KMeansModel(2); km.fit_predict(X_clu); km.plot()
kmed = KMedoidsModel(2); kmed.fit_predict(X_clu)
fcm = FuzzyCMeans(2); fcm.fit_predict(X_clu)
print("Fuzzy: số vòng lặp =", fcm.n_iter_, "| tổng mỗi dòng W (phải =1):", fcm.W_.sum(axis=1)[:3].round(3))
print("Membership 3 điểm đầu (mỗi dòng: mức độ thuộc cụm 0, cụm 1):\n", fcm.W_[:3].round(3))

# K-modes cần dữ liệu phân loại -> rời rạc hoá 6 cột đầu thành 3 mức (thấp/vừa/cao) bằng qcut
X_cat = np.column_stack([pd.qcut(dm.X.iloc[:, j], 3, labels=False) for j in range(6)])
kmo = KModesModel(2); kmo.fit_predict(X_cat)
print("K-modes: số điểm mỗi cụm =", np.bincount(kmo.labels_), "| mode mỗi cụm:\n", kmo.modes_)
print("Số điểm mỗi cụm – K-means:", np.bincount(km.labels_), "| K-medoids:", np.bincount(kmed.labels_))
```

## 🎯 KHI NÀO DÙNG: **GOM CỤM PHÂN HOẠCH** (K-means / K-modes / K-prototypes / K-medoids / Fuzzy C-means)
**DỮ LIỆU như thế nào thì chọn (📗 tr.13):**
| Dữ liệu | Chọn | Điều kiện / lưu ý |
|---|---|---|
| Features **số**, cụm **tròn / lồi**, **biết (hoặc thử được) k** | **K-means** | Tập phải **lồi** và phải biết k |
| Features **phân loại** | **K-modes** | Khoảng cách **Hamming**; tâm = **mode** |
| **Hỗn hợp** số + phân loại | **K-prototypes** | Khoảng cách số + khoảng cách phân loại |
| Cần tâm là **điểm thật** (diễn giải được), dữ liệu có **outlier** | **K-medoids (PAM)** | Tâm = điểm có tổng khoảng cách tới điểm khác **ngắn nhất**. 📙 bền với outlier hơn K-means vì tâm không bị kéo lệch |
| **Ranh giới mờ**, 1 điểm có thể thuộc **nhiều cụm** | **Fuzzy C-means** | Ma trận `wᵢⱼ ∈ [0,1]` **tổng mỗi dòng = 1**, tổng mỗi cột > 0; **dừng** khi ‖W(t)−W(t−1)‖ (Frobenius) nhỏ |

**KẾT QUẢ như thế nào thì kết luận:**
- 📊 Fuzzy: điểm đầu có membership **(0,295 ; 0,705)** → thuộc cụm 1 ~70% nhưng vẫn "dính" cụm 0 ~30% ⇒ **điểm gần ranh giới**. Nếu **hầu hết** membership ≈ 0,5/0,5 ⇒ cụm **chưa tách** (hoặc k chưa hợp). Muốn kết luận cứng ⇒ gán vào cụm có membership **lớn nhất**.
- 📊 K-means **[375 ; 194]** và K-medoids **[385 ; 184]** gần nhau ⇒ **ít outlier chi phối**; nếu hai kết quả khác hẳn ⇒ nghi có **outlier** → ưu tiên **K-medoids**.
- 📗 **Chọn k:** chạy nhiều k, dùng **ELBOW + Silhouette/CH/DB/Dunn** (xem ô 🎯 Phần 13).

**KHÔNG nên dùng K-means khi:** cụm **không lồi** (vòng, trăng khuyết) → dùng DBSCAN; **không biết k** → thử phân cấp / DBSCAN; có nhiều **outlier** → K-medoids.

📚 **Lý thuyết bổ sung từ notes (tr.11–13):**
- 📗 Học giám sát = **dự đoán** thuộc cụm nào (theo chủ đề); không giám sát = **mô tả** cụm đó có ý nghĩa gì (trọng tâm, mean, median, mode…). Mức mô tả: **vector trọng tâm** + **bán kính** (độ lệch chuẩn = khoảng cách TB từ 1 điểm tới tâm; phương sai = TB giữa 2 cặp điểm bất kỳ).
- 📗 **Similarity** (độ tương đồng) / **Distance – dissimilarity** (khoảng cách) / **Proximity** (gộp cả hai; với thuộc tính định tính đã số hoá thì giá trị **phụ thuộc chủ quan** của mình). Ma trận khoảng cách là ma trận **tam giác, không âm**; đo chuẩn vector bằng **p-norm / Frobenius norm**.
- 📗 Chất lượng cụm: **intra-class similarity** (kết dính – cohesion) phải **cao**; **extra-class** (khác biệt – distinction) phải **cao**.

## PHẦN 10. CÁCH TIẾP CẬN PHÂN CẤP: AGNES, DIANA, DENDROGRAM  *(Notes tr.13–14)*
- Thay đổi k để chạy thuật toán rồi chọn phân loại tốt nhất (tr.14).
- **AGNES (Agglomerative – gộp từ dưới lên):** bắt đầu mỗi điểm là 1 cụm, **gộp dần cụm gần nhất** cho tới khi còn 1 cụm.
- **DIANA (Divisive – chia từ trên xuống):** bắt đầu 1 cụm lớn rồi **tách dần**.
- **Dendrogram:** cây phân cấp; **cắt ngang** tại 1 mức → các nhánh còn dính nhau tạo thành 1 cụm.
- Khoảng cách giữa 2 cụm: single / complete / average / centroid / **Ward** (tr.13).

```python
class AgnesModel(BaseClusterer):
    """AGNES = gom cụm phân cấp gộp dần (Agglomerative Nesting) – Notes tr.14."""
    def __init__(self, k=2, link="ward"):
        super().__init__(); self.k, self.link = k, link          # link: 'single' | 'complete' | 'average' | 'ward'
        self.name = f"AGNES-{link} (k={k})"
    def fit_predict(self, X):
        self.labels_ = AgglomerativeClustering(n_clusters=self.k, linkage=self.link).fit_predict(X)
        return self.labels_
    @staticmethod
    def plot_dendrogram(X, link="ward", last=20):
        """Vẽ dendrogram (Notes tr.14). Chỉ hiện 'last' nhánh cuối cho gọn."""
        Z = linkage(X, method=link)
        dendrogram(Z, truncate_mode="lastp", p=last, no_labels=True)
        plt.title(f"Dendrogram ({link}-link)"); plt.ylabel("khoảng cách gộp"); plt.show()

class DianaModel(BaseClusterer):
    """DIANA = gom cụm phân cấp chia dần (Divisive Analysis) – Notes tr.14. Bản đơn giản:
    lặp: chọn cụm có đường kính lớn nhất -> tách 'nhóm tách rời' (splinter group) khỏi phần còn lại."""
    def __init__(self, k=2):
        super().__init__(); self.k = k; self.name = f"DIANA (k={k})"

    def _split(self, D, idx):
        avg = D[np.ix_(idx, idx)].sum(axis=1) / max(len(idx) - 1, 1)
        A = [idx[int(avg.argmax())]]                     # điểm "khác biệt nhất" bắt đầu nhóm tách rời
        B = [i for i in idx if i not in A]
        while len(B) > 1:
            dB = D[np.ix_(B, B)].sum(axis=1) / (len(B) - 1)    # TB khoảng cách tới phần còn lại
            dA = D[np.ix_(B, A)].mean(axis=1)                  # TB khoảng cách tới nhóm tách
            diff = dB - dA
            j = int(diff.argmax())
            if diff[j] <= 0: break                             # không còn điểm nào gần nhóm tách hơn
            A.append(B.pop(j))
        return np.array(A), np.array(B)

    def fit_predict(self, X):
        D = cdist(X, X)
        clusters = [np.arange(len(X))]                         # bắt đầu: 1 cụm chứa tất cả
        while len(clusters) < self.k:
            diam = [D[np.ix_(c, c)].max() for c in clusters]   # đường kính mỗi cụm
            c = clusters.pop(int(np.argmax(diam)))             # tách cụm có đường kính lớn nhất
            a, b = self._split(D, list(c))
            clusters += [a, b]
        self.labels_ = np.zeros(len(X), int)
        for j, c in enumerate(clusters): self.labels_[c] = j
        return self.labels_

# ----- chạy thử -----
AgnesModel.plot_dendrogram(X_clu, "ward")
for link in ["single", "complete", "average", "ward"]:       # 4 cách đo khoảng cách giữa cụm (tr.13)
    lab = AgnesModel(2, link).fit_predict(X_clu)
    print(f"AGNES {link:8s}: số điểm mỗi cụm = {np.bincount(lab)} | silhouette = {silhouette_score(X_clu, lab):.4f}")
dia = DianaModel(2); dia.fit_predict(X_clu)
print("DIANA: số điểm mỗi cụm =", np.bincount(dia.labels_), "| silhouette =", round(silhouette_score(X_clu, dia.labels_), 4))
```

## 🎯 KHI NÀO DÙNG: **GOM CỤM PHÂN CẤP** (AGNES / DIANA / Dendrogram) & chọn **linkage**
**DỮ LIỆU / TÌNH HUỐNG chọn phân cấp (📗 tr.13–14):** **chưa biết k**, muốn **xem cấu trúc lồng nhau** → vẽ **Dendrogram**, **cắt ngang** ở mức mong muốn (các nút còn dính nhau = 1 cụm); thay đổi k rồi chọn phân loại tốt nhất.
- **AGNES** = gộp **từ dưới lên** (mỗi điểm 1 cụm → gộp dần). **DIANA** = chia **từ trên xuống**.
- 📙 Dữ liệu **nhỏ–vừa** (cần ma trận khoảng cách n×n → rất tốn bộ nhớ khi n lớn).

**Chọn LINKAGE (📗 tr.13 là định nghĩa; 📙 lời khuyên dùng):**
| Linkage | Định nghĩa (notes) | Khi dùng / đặc điểm |
|---|---|---|
| **Single** | khoảng cách **ngắn nhất** giữa 2 item | Bắt được cụm **dài/chuỗi**, nhưng dễ bị **outlier / "chaining"** kéo |
| **Complete** | khoảng cách **dài nhất** | Cụm **gọn**, nhưng **nhạy outlier** |
| **Average** | khoảng cách **trung bình** | Thoả hiệp giữa 2 loại trên |
| **Centroids / Medoids** | khoảng cách 2 **trọng tâm / 2 đối tượng trung tâm** | Diễn giải rõ |
| **Ward** | gộp 2 cụm rồi xét **phương sai** (độ lệch chuẩn của tập hội) tăng ít nhất | 📙 Cụm **đều, gọn** → lựa chọn mặc định tốt cho dữ liệu số |

**KẾT QUẢ như thế nào thì kết luận (📊 ví dụ thật):**

- 📊 `ward` → **[184 ; 385]** (Silhouette **0,34** thấp hơn, nhưng **2 cụm cân đối, có ý nghĩa**) ⇒ **chọn Ward**.
- ⇒ **Bài học:** *Silhouette cao chưa chắc tốt* — luôn xem **kích thước từng cụm** (đừng chỉ nhìn 1 con số). Notes tr.17 cũng nói: hình "dao bầu" (các cụm đều nhau) mới là tốt.

## PHẦN 11. DỰA TRÊN MẬT ĐỘ (DBSCAN) & DỰA TRÊN LƯỚI (CLIQUE)  *(Notes tr.14–15)*
**DBSCAN (tr.14–15):** vùng **dày đặc** ngăn cách bởi vùng thưa. Dùng **bán kính ε (eps)** và **ngưỡng số láng giềng (min_samples)**:
- **Core point (điểm lõi):** có số láng giềng trong ε **vượt ngưỡng**.
- **Border point (điểm biên):** nằm trong ε của điểm lõi nhưng không đủ ngưỡng; các cụm có thể **lan truyền** nối đuôi qua các điểm lõi.
- **Noise point (điểm nhiễu):** xa cả hai loại trên (nhãn −1).
- Ngưỡng chọn tự do hoặc theo tiêu chí giống K-NN (→ dùng **k-distance** để chọn ε). **KHÔNG dùng cho dữ liệu lồi** *(đúng nguyên văn notes)*.

**CLIQUE (tr.15):** gom cụm trong **không gian con** bằng lưới; áp dụng **tính chất Apriori**: ô trong không gian k chiều đủ dày (≥ n điểm) **chỉ khi** mọi hình chiếu xuống k−1 chiều cũng đủ dày. Làm từng chiều trước → giữ ô đủ mật độ → ghép sang chiều sau → lặp.

```python
class DBSCANModel(BaseClusterer):
    """DBSCAN: core / border / noise (Notes tr.14–15)."""
    def __init__(self, eps=2.5, min_samples=5):
        super().__init__(); self.eps, self.min_samples = eps, min_samples      # min_samples = ngưỡng k
        self.name = f"DBSCAN (eps={eps}, min={min_samples})"
    def fit_predict(self, X):
        self.model = DBSCAN(eps=self.eps, min_samples=self.min_samples).fit(X)
        self.labels_ = self.model.labels_                                     # -1 = điểm nhiễu
        n = len(self.labels_)
        self.core_mask_ = np.zeros(n, bool); self.core_mask_[self.model.core_sample_indices_] = True
        return self.labels_
    def summary(self):
        lab = self.labels_
        return {"số cụm": len(set(lab)) - (1 if -1 in lab else 0),
                "điểm lõi (core)": int(self.core_mask_.sum()),
                "điểm biên (border)": int(((lab != -1) & ~self.core_mask_).sum()),
                "điểm nhiễu (noise)": int((lab == -1).sum())}
    @staticmethod
    def k_distance(X, k=5):
        """Chọn eps theo kiểu K-NN: sắp xếp khoảng cách tới láng giềng thứ k; chỗ 'gãy' (elbow) gợi ý eps."""
        d = NearestNeighbors(n_neighbors=k).fit(X).kneighbors(X)[0][:, -1]
        d = np.sort(d)
        plt.plot(d); plt.title(f"{k}-distance (chọn eps tại điểm gãy)"); plt.ylabel("khoảng cách"); plt.show()
        return d

kd = DBSCANModel.k_distance(X_clu, 5)
print("Gợi ý eps (phân vị 90% của k-distance):", round(np.quantile(kd, 0.90), 2))

# Quét thử nhiều eps (như 'cứ chạy hết xem cái nào tốt'): dữ liệu 30 chiều nên eps lớn -> gần như 1 cụm lớn + ít nhiễu
for e in [2.0, 2.5, 3.0, 4.0]:
    t = DBSCANModel(eps=e, min_samples=5); t.fit_predict(X_clu)
    print(f"eps={e}: {t.summary()}")

db = DBSCANModel(eps=2.5, min_samples=5); db.fit_predict(X_clu)    # chọn eps=2.5 để có >= 2 cụm minh hoạ
print("\nMô hình chọn:", db.name, "->", db.summary())
db.plot(title="DBSCAN (màu vàng/xanh = cụm, tím đậm = nhiễu -1)")

class CliqueModel(BaseClusterer):
    """CLIQUE (bản 2 chiều đơn giản) – Notes tr.15. xi = số ô mỗi chiều, tau = ngưỡng mật độ (số điểm tối thiểu/ô)."""
    def __init__(self, xi=10, tau=15):
        super().__init__(); self.xi, self.tau = xi, tau; self.name = f"CLIQUE (xi={xi}, tau={tau})"
    def fit_predict(self, X2):
        n = len(X2)
        # chia lưới: mỗi chiều cắt thành xi đoạn bằng nhau
        idx = np.zeros((n, 2), int)
        for d in range(2):
            edges = np.linspace(X2[:, d].min(), X2[:, d].max(), self.xi + 1)
            idx[:, d] = np.clip(np.digitize(X2[:, d], edges) - 1, 0, self.xi - 1)
        # Bước 1 (làm trên 1 chiều trước): ô 1 chiều đủ dày (>= tau điểm)
        dense1 = [{b for b in range(self.xi) if (idx[:, d] == b).sum() >= self.tau} for d in range(2)]
        # Bước 2 (Apriori): ô 2 chiều chỉ xét nếu CẢ HAI hình chiếu 1 chiều đều đủ dày, và bản thân đủ dày
        cells = {}
        for i, (b0, b1) in enumerate(idx):
            if b0 in dense1[0] and b1 in dense1[1]:
                cells.setdefault((b0, b1), []).append(i)
        dense2 = {c: pts for c, pts in cells.items() if len(pts) >= self.tau}
        # Bước 3: nối các ô đủ dày kề nhau (trên/dưới/trái/phải) thành 1 cụm (duyệt BFS)
        labels = -np.ones(n, int); cid = 0; seen = set()
        for start in dense2:
            if start in seen: continue
            queue = [start]; seen.add(start)
            while queue:
                c = queue.pop()
                labels[dense2[c]] = cid
                for nb in [(c[0]+1, c[1]), (c[0]-1, c[1]), (c[0], c[1]+1), (c[0], c[1]-1)]:
                    if nb in dense2 and nb not in seen:
                        seen.add(nb); queue.append(nb)
            cid += 1
        self.labels_, self.n_dense_cells_ = labels, len(dense2)
        return labels

# Chạy CLIQUE trên 2 chiều (radius_mean, texture_mean) đã chuẩn hoá
cl = CliqueModel(xi=10, tau=15); cl.fit_predict(X_clu[:, [0, 1]])
print(cl.name, "| số ô đủ dày:", cl.n_dense_cells_, "| số cụm:", len(set(cl.labels_)) - (1 if -1 in cl.labels_ else 0),
      "| điểm không thuộc ô dày (nhiễu):", int((cl.labels_ == -1).sum()))
plt.scatter(X_clu[:, 0], X_clu[:, 1], c=cl.labels_, cmap="viridis", s=12)
plt.title("CLIQUE trên (radius_mean, texture_mean)"); plt.show()
```

## PHẦN 12. ĐÁNH GIÁ MÔ HÌNH HỒI QUY  *(Notes tr.15–17, BUỔI 6 – ĐÁNH GIÁ GIẢI PHÁP MÔ HÌNH, mục 3)*
**Mô hình tốt cần (tr.15):** *không chệch* (sai số bé), *vững chắc* (phương sai bé khi đổi mẫu), *hiệu quả* (cân nhắc chi phí – sai số).
**Các độ đo (tr.15–16):** **SSE** (tổng bình phương sai số) • **MSE** (L2, chia N để công bằng khi tập train khác kích thước) • **MAE** (L1) • **RSE** (không phụ thuộc đơn vị đo) • **R²** (mức độ phù hợp).
> 📝 Notes ghi *"mô hình tốt với R²>1"* – thực tế **R² ≤ 1 và càng gần 1 càng tốt** (có thể là lỗi gõ trong ghi chú), code bên dưới dùng định nghĩa chuẩn.

**Phân rã sai số (tr.16):** Kỳ vọng sai số = *độ chệch (sai số Train)* + *phương sai (Test − Train)* + *sai số không thể giảm*.
**4 công cụ chọn/so sánh mô hình (tr.16–17):** (1) **ELBOW** – điểm gãy, ưu tiên k nhỏ; (2) **Regression Plot** (x vs y + đường hồi quy; dùng **cả R² và đường hồi quy**); (3) **Scatter Plot** *(CHÚ Ý CHO KIỂM TRA)* – trục hoành = y thực, trục tung = y dự đoán, càng bám đường phân giác càng tốt; (4) **Distribution Plot** – mỗi thuật toán 1 đường so với phân phối y thực; nếu R² gần nhau thì xem độ bám, bám quá giống thì chọn R² lớn hơn.

```python
class RegressionEvaluator:
    """Các độ đo & biểu đồ đánh giá hồi quy (Notes tr.15–17)."""

    @staticmethod
    def sse(y, yh):  return float(np.sum((y - yh) ** 2))                   # Sum of Squared Error
    @staticmethod
    def mse(y, yh):  return float(np.mean((y - yh) ** 2))                  # Mean Squared Error (L2 loss)
    @staticmethod
    def mae(y, yh):  return float(np.mean(np.abs(y - yh)))                 # Mean Absolute Error (L1 loss)
    @staticmethod
    def rse(y, yh):  return float(np.sum((y - yh) ** 2) / np.sum((y - np.mean(y)) ** 2))   # Relative Squared Error
    @staticmethod
    def r2(y, yh):   return 1 - RegressionEvaluator.rse(y, yh)             # R² = 1 - RSE

    @classmethod
    def summary(cls, y, yh):
        y, yh = np.asarray(y, float), np.asarray(yh, float)
        return {"SSE": cls.sse(y, yh), "MSE": cls.mse(y, yh), "MAE": cls.mae(y, yh),
                "RSE": cls.rse(y, yh), "R2": cls.r2(y, yh)}

    @staticmethod
    def regression_plot(x, y, model, xlabel="radius_mean", ylabel="area_mean"):
        """Regression Plot (tr.16): x độc lập (hoành) – y phụ thuộc (tung) + đường hồi quy."""
        xs = np.linspace(x.min(), x.max(), 200).reshape(-1, 1)
        plt.scatter(x, y, s=10, alpha=0.5)
        plt.plot(xs, model.predict(pd.DataFrame(xs, columns=[xlabel])), "r")
        plt.xlabel(xlabel); plt.ylabel(ylabel); plt.title("Regression Plot"); plt.show()

    @staticmethod
    def scatter_plot(y_true, preds):
        """Scatter Plot (tr.16 – CHÚ Ý KIỂM TRA): hoành = y thực, tung = y dự đoán, so với đường phân giác y=x."""
        lo, hi = float(np.min(y_true)), float(np.max(y_true))
        plt.figure(figsize=(5, 5))
        for name, yh in preds.items():
            plt.scatter(y_true, yh, s=8, alpha=0.5, label=name)
        plt.plot([lo, hi], [lo, hi], "k--", label="đường phân giác")
        plt.xlabel("y thực tế"); plt.ylabel("y dự đoán"); plt.legend(fontsize=7); plt.title("Scatter Plot"); plt.show()

    @staticmethod
    def distribution_plot(y_true, preds):
        """Distribution Plot (tr.17): mỗi thuật toán 1 đường phân phối so với y thực."""
        bins = np.linspace(np.min(y_true), np.max(y_true), 30)
        centers = (bins[:-1] + bins[1:]) / 2
        plt.plot(centers, np.histogram(y_true, bins, density=True)[0], "k", lw=3, label="y thực")
        for name, yh in preds.items():
            plt.plot(centers, np.histogram(yh, bins, density=True)[0], label=name)
        plt.legend(fontsize=7); plt.title("Distribution Plot"); plt.show()

# ----- so sánh nhiều thuật toán hồi quy trên cùng tập (đa biến) -----
reg_models = {
    "Linear (GD)":     LinearRegressionGD(rho=0.1, n_iter=1000),
    "Linear (sklearn)": LinearRegression(),
    "Poly bậc 2":      PolynomialRegressionModel(2),
    "K-NN (k=5)":      Pipeline([("s", StandardScaler()), ("r", KNeighborsRegressor(5))]),
    "Cây (depth=4)":   DecisionTreeRegressor(max_depth=4, random_state=RANDOM_STATE),
}
rows, preds = [], {}
for name, m in reg_models.items():
    m.fit(Xr_tr, yr_tr)
    preds[name] = m.predict(Xr_te)
    s = RegressionEvaluator.summary(yr_te, preds[name])
    tr_mse = RegressionEvaluator.mse(yr_tr.values, m.predict(Xr_tr))
    rows.append({"model": name, **s, "Train MSE (chệch)": tr_mse, "Test-Train (phương sai)": s["MSE"] - tr_mse})
print(pd.DataFrame(rows).set_index("model").round(4).to_string())

# 3 loại biểu đồ (tr.16–17)
RegressionEvaluator.regression_plot(Xr_tr[single_col].values.ravel(), yr_tr.values,
                                    LinearRegression().fit(Xr_tr[single_col], yr_tr))
RegressionEvaluator.scatter_plot(yr_te.values, {k: preds[k] for k in ["Linear (sklearn)", "Poly bậc 2", "K-NN (k=5)"]})
RegressionEvaluator.distribution_plot(yr_te.values, {k: preds[k] for k in ["Linear (sklearn)", "K-NN (k=5)", "Cây (depth=4)"]})

# ELBOW chọn k cho K-NN hồi quy (tr.16): sai số Test theo k, chọn điểm gãy / k nhỏ mà sai số chấp nhận được
ks = list(range(1, 31))
errs = [RegressionEvaluator.mse(yr_te.values,
        Pipeline([("s", StandardScaler()), ("r", KNeighborsRegressor(k))]).fit(Xr_tr, yr_tr).predict(Xr_te)) for k in ks]
plt.plot(ks, errs, "o-"); plt.axvline(find_elbow(ks, errs), color="r", ls="--")
plt.xlabel("k"); plt.ylabel("MSE Test"); plt.title("ELBOW cho K-NN hồi quy"); plt.show()
print("k tại điểm gãy (elbow):", find_elbow(ks, errs))
```

| Độ đo | Dùng khi | Ghi chú |

| **Distribution plot** | **So nhiều thuật toán** (mỗi thuật toán 1 đường) | R² **gần nhau** → xem đường nào **bám phân phối y thực** hơn; **bám y hệt** thì chọn mô hình có **R² lớn hơn** |

## PHẦN 13. ĐÁNH GIÁ MÔ HÌNH GOM CỤM  *(Notes tr.17–18)*
**Không có đáp án để so sánh** → dùng tiêu chí **nội tại** (tr.17): **Compactness** (điểm trong cụm giống nhau) & **Separation** (các cụm tách xa nhau).
1. **Cluster inertia** (tổng bình phương khoảng cách tới tâm): k tăng thì inertia giảm → dùng **ELBOW**.
2. **Silhouette, Calinski-Harabasz (CH), Davies-Bouldin (DB), Dunn** — **⭐ notes ghi: 3 chỉ số cuối sẽ vào bài kiểm tra** (CH, DB, Dunn; Silhouette cũng giải thích rất kỹ).
   - **Silhouette:** `s = (b − a) / max(a, b)`; **a** = khoảng cách TB tới "đồng đội" cùng cụm, **b** = khoảng cách TB tới cụm láng giềng gần nhất. **s>0:** gần đồng đội, xa láng giềng (tốt); **s=0:** nằm ở biên; **s<0:** gần láng giềng hơn đồng đội (xếp nhầm cụm). Mẫu số chỉ để chuẩn hoá về [−1,1]. Không phải k tăng là silhouette tốt → **chọn k ở giữa**; biểu đồ các cụm đều nhau (hình "dao bầu") là tốt.
   - **CH:** càng **lớn** càng tốt. **DB:** càng **nhỏ** càng tốt. **Dunn:** càng **lớn** càng tốt.
3. **Đánh giá ngoài** (tr.18; chủ yếu lý thuyết, cần nhãn có sẵn): **Rand index** (so 2 phân hoạch R và S) và **Jaccard**. Dùng khi k=5 và k=7 có silhouette xấp xỉ nhau.

```python
class ClusterEvaluator:
    """Đánh giá gom cụm: inertia/elbow, Silhouette, CH, DB, Dunn (nội tại) + Rand, Jaccard (ngoài). Notes tr.17–18."""

    @staticmethod
    def dunn_index(X, labels):
        """Dunn = (khoảng cách NHỎ NHẤT giữa 2 cụm) / (đường kính LỚN NHẤT của 1 cụm). Càng lớn càng tốt."""
        groups = [X[labels == c] for c in np.unique(labels) if c != -1]       # bỏ điểm nhiễu (-1) nếu có
        min_inter = min(cdist(a, b).min() for i, a in enumerate(groups) for b in groups[i + 1:])
        max_intra = max(cdist(g, g).max() for g in groups)
        return min_inter / max_intra

    @staticmethod
    def silhouette_one_point(X, labels, i):
        """Tính TAY silhouette cho 1 điểm theo đúng các bước notes tr.17–18 (để giải thích trong bài)."""
        same = np.where(labels == labels[i])[0]; same = same[same != i]
        a = cdist(X[[i]], X[same]).mean()                                     # a: TB khoảng cách tới đồng đội
        b = min(cdist(X[[i]], X[labels == c]).mean() for c in np.unique(labels) if c != labels[i])   # b: tới cụm láng giềng gần nhất
        return (b - a) / max(a, b), a, b

    @classmethod
    def internal(cls, X, labels):
        mask = labels != -1                                                   # loại điểm nhiễu khi đo
        Xm, lm = X[mask], labels[mask]
        return {"Silhouette (lớn tốt)": silhouette_score(Xm, lm),
                "Calinski-Harabasz (lớn tốt)": calinski_harabasz_score(Xm, lm),
                "Davies-Bouldin (nhỏ tốt)": davies_bouldin_score(Xm, lm),
                "Dunn (lớn tốt)": cls.dunn_index(Xm, lm)}

    @staticmethod
    def external(y_true, labels):
        """Rand index & Jaccard dựa trên các CẶP điểm (Notes tr.18)."""
        pc = pair_confusion_matrix(y_true, labels)       # [[cặp khác-khác, cặp khác-cùng],[cặp cùng-khác, cặp cùng-cùng]]
        tp, fp, fn = pc[1, 1], pc[0, 1], pc[1, 0]
        return {"Rand index": rand_score(y_true, labels), "Jaccard": tp / (tp + fp + fn)}

    @staticmethod
    def plot_silhouette(X, labels):
        """Biểu đồ silhouette từng cụm (hình 'dao bầu' – Notes tr.17)."""
        s = silhouette_samples(X, labels); y0 = 0
        for c in np.unique(labels):
            sc = np.sort(s[labels == c]); plt.fill_betweenx(np.arange(y0, y0 + len(sc)), 0, sc, alpha=0.7)
            y0 += len(sc) + 10
        plt.axvline(s.mean(), color="r", ls="--"); plt.xlabel("silhouette"); plt.title("Silhouette từng cụm"); plt.show()

# ---- (a) Elbow + 4 chỉ số nội tại khi thay k = 2..8 (K-means) ----
rows, inertias = [], []
for k in range(2, 9):
    m = KMeansModel(k); lab = m.fit_predict(X_clu); inertias.append(m.inertia_)
    rows.append({"k": k, "inertia": m.inertia_, **ClusterEvaluator.internal(X_clu, lab)})
res = pd.DataFrame(rows).set_index("k")
print(res.round(4).to_string())
plt.plot(range(2, 9), inertias, "o-"); plt.axvline(find_elbow(range(2, 9), inertias), color="r", ls="--")
plt.xlabel("k"); plt.ylabel("inertia (SSE)"); plt.title("ELBOW – Cluster inertia"); plt.show()
print("k tại điểm gãy:", find_elbow(range(2, 9), inertias),
      "| k có Silhouette cao nhất:", res["Silhouette (lớn tốt)"].idxmax(),
      "| k có DB thấp nhất:", res["Davies-Bouldin (nhỏ tốt)"].idxmin())
# Lưu ý notes tr.17: không phải k tăng thì silhouette tốt; không chọn k quá bé, nên cân nhắc k ở giữa.
```

```python
# ---- (b) Silhouette tính tay cho 1 điểm, đối chiếu sklearn ----
lab2 = KMeansModel(2).fit_predict(X_clu)
s_hand, a, b = ClusterEvaluator.silhouette_one_point(X_clu, lab2, 0)
print(f"Điểm 0: a={a:.3f}, b={b:.3f} -> s=(b-a)/max(a,b)={s_hand:.4f} | sklearn: {silhouette_samples(X_clu, lab2)[0]:.4f}")
ClusterEvaluator.plot_silhouette(X_clu, lab2)

# ---- (c) So sánh các thuật toán gom cụm (k=2): chỉ số nội tại + đánh giá ngoài so với nhãn thật ----
algos = {
    "K-means":   KMeansModel(2), "K-medoids": KMedoidsModel(2), "Fuzzy C-means": FuzzyCMeans(2),
    "AGNES-ward": AgnesModel(2, "ward"), "DIANA": DianaModel(2), "DBSCAN": db,
}
rows = []
for name, m in algos.items():
    lab = m.fit_predict(X_clu)
    if len(set(lab) - {-1}) < 2:           # DBSCAN có thể chỉ ra 1 cụm -> không tính được chỉ số
        continue
    rows.append({"thuật toán": name, **ClusterEvaluator.internal(X_clu, lab), **ClusterEvaluator.external(y_true_clu, lab)})
print(pd.DataFrame(rows).set_index("thuật toán").round(4).to_string())
# Rand index/Jaccard cao = phân hoạch trùng với nhãn thật nhiều (chỉ dùng được khi có nhãn – Notes tr.18)
```

## 🎯 KHI NÀO DÙNG: **CHỈ SỐ ĐÁNH GIÁ GOM CỤM** & cách **chọn k**
📗 Gom cụm **không có đáp án để so** ⇒ dùng tiêu chí **nội tại** (không cần nhãn ngoài): **Compactness** (điểm trong cụm giống nhau) và **Separation** (các cụm **tách xa nhau**). *(Notes tr.17 ghi "độ phân tách càng **nhỏ** càng tốt" — mình hiểu là gõ nhầm; ý đúng: các cụm càng **tách xa / ít chồng lấn** càng tốt.)*

| Chỉ số | Hướng tốt | Dùng khi | Dấu hiệu |
|---|---|---|---|
| **Inertia** (tổng bình phương tới tâm) | **Nhỏ**, nhưng **luôn giảm khi k tăng** | Chọn k bằng **ELBOW** (📗) — **không** chọn k làm inertia nhỏ nhất | Điểm gãy của đường cong |
| **Silhouette** s=(b−a)/max(a,b) ∈[−1,1] | **Lớn** | Ý nghĩa từng điểm và cả mô hình | 📗 **s>0**: gần đồng đội, xa láng giềng • **s=0**: ở biên • **s<0**: xếp nhầm cụm. **Không phải k tăng là tốt** ⇒ chọn k **ở giữa**; các cụm đều nhau (hình **"dao bầu"**) là tốt |
| **Calinski–Harabasz (CH)** | **Lớn** | ⭐ Vào bài kiểm tra | — |
| **Davies–Bouldin (DB)** | **Nhỏ** | ⭐ Vào bài kiểm tra | — |
| **Dunn** | **Lớn** (min khoảng cách giữa cụm / max đường kính cụm) | ⭐ Vào bài kiểm tra | 📙 nhạy outlier |
| **Rand / Jaccard** (đánh giá **ngoài**) | Gần 1 | **Chỉ khi có nhãn ngoài**; 📗 dùng khi 2 mô hình (VD k=5 và k=7) có **Silhouette xấp xỉ nhau** | Chủ yếu lý thuyết, ít thực hành vì **không có thang đo** |

## PHẦN 14. `ExamPipeline` – QUY TRÌNH TRỌN GÓI CHO BÀI KIỂM TRA  *(Notes tr.4 – "Phải làm bằng OOP mới có điểm")*
Đúng thứ tự đề bài kiểm tra mô tả: **(1) đọc tập huấn luyện → (2) EDA + tiền xử lý → (3) phân tích hướng đối tượng + ràng buộc → (4) đưa Test + đánh giá mô hình → (5) trả kết quả cho tập thực tế**.
Khi vào phòng thi: chỉ cần đổi `path`, `target` và chạy `ExamPipeline(...).run()`; phần dự đoán tập mới là `predict_new()`.

```python
class ExamPipeline:
    """Quy trình phân lớp trọn gói: EDA -> tiền xử lý -> huấn luyện nhiều mô hình -> đánh giá -> chọn tốt nhất -> dự đoán dữ liệu mới."""

    def __init__(self, path, target="Class", positive="Malignant"):
        self.dm = DataManager(path, target)
        self.positive = positive                       # tên nhãn "dương" (Positive) khi in kết quả
        self.models = {}
        self.best_name = None

    # ---- (1)+(2) đọc, EDA, tiền xử lý ----
    def prepare(self):
        self.dm.load()
        self.dm.eda()
        self.dm.preprocess()
        self.X_tr, self.X_te, self.y_tr, self.y_te = self.dm.split()
        self.features = list(self.X_tr.columns)        # nhớ tên & thứ tự cột để kiểm tra ràng buộc khi dự đoán mới

    # ---- (3)+(4) huấn luyện + đánh giá ----
    def train_and_evaluate(self):
        k = KNNClassifier.find_best_k(self.X_tr, self.y_tr).idxmax()      # chọn k bằng CV
        self.models = {
            "K-NN": KNNClassifier(int(k)),
            "Decision Tree": DecisionTreeModel("entropy", ccp_alpha=0.01),
            "NB Gaussian": NaiveBayesModel("gaussian"),
            "SVM rbf": SVMModel("rbf"),
        }
        self.table = ModelComparison(self.models).run(self.X_tr, self.y_tr, self.X_te, self.y_te, plot=False)
        # chọn mô hình: AUC cao nhất (nếu hoà thì lấy F1 cao hơn)
        self.best_name = self.table.sort_values(["AUC", "F1"], ascending=False).index[0]
        print(self.table.round(4))
        print("=> Mô hình được chọn:", self.best_name)

    # ---- (5) dự đoán cho tập thực tế (có ràng buộc đầu vào) ----
    def predict_new(self, df_new):
        missing = [c for c in self.features if c not in df_new.columns]
        if missing:
            raise ValueError(f"Dữ liệu mới thiếu cột: {missing[:3]}...")           # ràng buộc 1: đủ cột
        if df_new[self.features].isna().any().any():
            raise ValueError("Dữ liệu mới có giá trị thiếu (NaN)")                  # ràng buộc 2: không thiếu giá trị
        Xn = df_new[self.features]                                                   # ràng buộc 3: đúng thứ tự cột
        model = self.models[self.best_name]
        pred = model.predict(Xn)
        inv = {v: k for k, v in self.dm.label_map.items()}                           # số -> chữ
        out = pd.DataFrame({"dự đoán": [inv[p] for p in pred]}, index=df_new.index)
        if hasattr(model.pipeline, "predict_proba"):
            out[f"P({self.positive})"] = model.pipeline.predict_proba(Xn)[:, 1].round(4)
        return out

    def run(self):
        self.prepare(); self.train_and_evaluate()

# ----- chạy thử toàn bộ -----
exam = ExamPipeline(CSV_PATH)
exam.run()

# "Tập thực tế": giả lập 5 mẫu mới chưa có nhãn (lấy từ Test, bỏ nhãn)
new_data = exam.X_te.iloc[:5]
print("\nKết quả dự đoán tập thực tế:\n", exam.predict_new(new_data))
print("Nhãn thật để đối chiếu     :", [("Malignant" if v == 1 else "Benign") for v in exam.y_te.iloc[:5]])

# Thử ràng buộc: thiếu cột -> báo lỗi
try:
    exam.predict_new(new_data.drop(columns=["radius_mean"]))
except ValueError as e:
    print("Ràng buộc hoạt động:", e)
```
